```python
# Variable declarations
flora_name = "Majegau"          # String
flora_count = 15                # Integer
latitude_coordinate = -8.3405   # Float
is_protected = True             # Boolean

print(flora_name)
print(latitude_coordinate)
print(flora_count)
print(is_protected)
```

```python
# Creating a list of strings
endemic_plants = ["Majegau", "Bunga Sandat", "Kantong Semar"]

# Accessing items via index (index starts at 0)
print(endemic_plants[0])  # Output: Majegau
print(endemic_plants[2])  # Output: Kantong Semar
```

```python
# Creating a dictionary for a single plant
plant_profile = {
    "nama": "Bunga Bangkai",
    "status": "Terancam Punah",
    "lat": -0.7893,
    "lon": 100.9649
}

# Accessing values using keys
print(plant_profile["nama"])    # Output: Bunga Bangkai
print(plant_profile["status"])  # Output: Terancam Punah
```

```python
current_status = "Rentan"

if current_status == "Terancam Punah":
    print("Urgent protection needed.")
elif current_status == "Rentan":
    print("Monitoring required.")
else:
    print("Population is stable.")

# Output : 
# Monitoring required.
```

```python
categories = ["Aman", "Rentan", "Terancam Punah"]

for category in categories:
    print(category)

# Output : 
# Aman
# Rentan
# Terancam Punah
```

```python
DATA_FLORA = [
    {
        "nama": "Majegau",
        "status": "Rentan",
        "lat": -8.3405, 
        "lon": 115.0919 
    },
    {
        "nama": "Bunga Sandat",
        "status": "Aman",
        "lat": -8.4095,
        "lon": 115.1889
    }
]

print(DATA_FLORA[1]["nama"])  # Output: Bunga Sandat

for tanaman in DATA_FLORA:
    nama_tanaman = tanaman["nama"]
    status_konservasi = tanaman["status"]
    
    print(nama_tanaman)
    print(status_konservasi)

# Output : 
# Majegau
# Rentan
# Bunga Sandat
# Aman
```
