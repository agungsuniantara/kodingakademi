```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service

chrome_driver_path = "/Applications/MAMP/htdocs/local/kodingakademi/Python/ivan/scrapping/chromedriver-mac-arm64/chromedriver"

service = Service(chrome_driver_path)
driver = webdriver.Chrome(service=service)

driver.get("https://www.wikipedia.org/")

print("judul halaman : ", driver.title)
```

```python
from selenium.webdriver.common.by import By

element = driver.find_element(By.CLASS_NAME, "central-textlogo")
print(element.text)
```

```python
element = driver.find_element(By.ID, "searchLanguage")
print(element.text)
```

```python
elements = driver.find_elements(By.TAG_NAME, "div")

for element in elements:
    print(element.text)
```

```python
element = driver.find_element(By.NAME, "search")
element.send_keys("Selenium Python")
element.submit()
```

```python
button = driver.find_element(By.CLASS_NAME, "pure-button")
button.click()
```

```python
element = driver.find_element(By.CSS_SELECTOR, "footer.footer")
print(element.text)
```

```python
link = driver.find_element(By.LINK_TEXT, "Download Wikipedia for Android or iOS")
print(link.get_attribute("href"))
```

XPATH

```python
element = driver.find_element(By.XPATH, "//p[@class='site-license']")
print(element.text)
```

```python
link = driver.find_element(By.XPATH, "//a[strong[text() = 'English']]")
print(link.get_attribute("href"))
```

```python
driver.implicitly_wait(10)

element = driver.find_element(By.CLASS_NAME, "central-textlogo")
print(element.text)
```

```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

wait = WebDriverWait(driver, 10)

element = wait.until(EC.text_to_be_present_in_element((By.CLASS_NAME, "central-textlogo"), "Wikipedia"))

print("elemen teks ditemukan")
```
