```python
def total_sum(a, b):
    sum_a = a.sum()
    sum_b = b.sum()
    return sum_a, sum_b

array1 = np.array([1,2,3,4,5])
array2 = np.array([10,11,12,13,14])

sum_a, sum_b = total_sum(array1, array2)
print(f"sum a : {sum_a}, sum b : {sum_b}")
```

```python
menu = ["nasi goreng", "mie goreng", "es jeruk", "jus mangga", "pisang goreng"]
pesanan = []

print("Menu :")
for i in range(len(menu)):
    print(f"{i + 1}. {menu[i]}")
    
while(True):
    memilih = input("pilih menu : ")
    pesanan.append(memilih)
    
    kondisi = input("masih ingin memilih ? ")
    if kondisi == "tidak":
        break

print("pesanan anda  : ")
for i in range(len(pesanan)):
    print(f"{i + 1}. {pesanan[i]}")

print(f"total harga = {len(pesanan) * 20000}")
```

```python
data = [
    [1,2],
    [3,4]
]

det = data[0][0] * data[1][1] - data[0][1] * data[1][0]

ops1 = 1 / det 
adjoin = [
    [ops1 * data[1][1], -(ops1 * data[0][1])],
    [-(ops1 * data[1][0]), ops1 * data[0][0]]
]
print(adjoin)
```
