```python
import pandas as pd
import pandas as pd
import warnings

df = pd.read_excel("penumpang transjakarta desember 2021.xlsx")
df
```

```python
df2 = pd.read_csv("penumpang transjakarta november 2021.csv")
df2
```

```python
# Hapus kolom dengan nama "Unnamed" pada data 1

df = df.drop(columns=[col for col in df.columns if 'Unnamed' in col])

df
```

```python
# Menemukan indeks baris "pimpinan"

idx_pimpinan = df[df["tahun"] == "pimpinan"].index[0]

# Menghapus baris pimipnan

df.drop(idx_pimpinan, inplace=True)

# df.drop(df.index[idx_pimpinan:], inplace=True) # Menghapus baris dari indeks "pimpinan" ke bawah

df
```

```python
# Membersihkan data dengan menghapus karakter ';;;;' pada penamaan kolom 'jumlah_penumpang'

df.columns = df.columns.str.replace(';', '')

# tampilkan data

df
```

```python
# Membersihkan data dengan menghapus baris yang kosong

df = df.dropna()

df
```

```python
# Menggunakan apply untuk mengubah setiap kolom menjadi string dan menghilangkan tanda ;
df = df.apply(lambda x: x.astype(str).apply(lambda y: y.replace(';', '')))

# Tampilkan data
df
```

```python
# Mengubah kolom 'jumlah_penumpang' menjadi float dengan mengubah yang tidak bisa menjadi NaN

df['jumlah_penumpang'] = pd.to_numeric(df['jumlah_penumpang'], errors='coerce')

# tampilkan data

df
```

```python
# hapus data jumlah_penumpang yang berisi NaN

df = df.dropna()

df
```

```python
import dataframe_image as dfi

dfi.export(df, "data_bersih.png", max_rows=-1)
```

```python
#gabungkan data desember dan data november menggunakan concat

#ignore_index=True digunakan untuk mengatur ulang indeks setelah penggabungan, sehingga indeks tidak berulang.

df = pd.concat([df, df2], ignore_index=True)

df
```

```python
dfi.export(df, "data_final.png", max_rows=-1)
```

```python
# final ubah data yang sudah dibersihkan dalam bentuk excel

df.to_excel('data_final.xlsx', index=False)
```
