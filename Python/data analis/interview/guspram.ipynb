```python
def hitung(*a): # [[1,2,3,4,5], ]
    hasil = 0 # 1 # 3
    for i in a: # i = [1,2,3,4,5]
        for z in i: # z = 1, z = 2
            hasil += z # hasil = 1 + 2
    return hasil
    
data = []

for i in range(2):
    angka = int(input("masukan angka : "))
    data.append(angka)

func = hitung(data)

rata = lambda a: print(a / len(data))

rata(func)
```
