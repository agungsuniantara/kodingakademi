```python
import numpy as np

dim1 = np.array([1,2,3,4])
dim1.sum()
```

```python
# sum

dim2 = np.array([
    [1,2,3],
    [4,5,6]
])
print(dim2.sum())

# baris
print(dim2.sum(axis = 1))

# kolom
print(dim2.sum(axis = 0))
```

```python
# mean

print(dim2.mean())
print(dim2.mean(axis = 1))
print(dim2.mean(axis = 0))
```

```python
# varian
dim2 = np.array([
    [1,2,3],
    [4,5,6]
])
print(dim2)
print(dim2.var())
print(dim2.var(axis = 1))
print(dim2.var(axis = 0))
```

```python
# standar deviasi

print(dim2)
print(dim2.std())
print(dim2.std(axis = 1))
print(dim2.std(axis = 0))
```

```python
# standar deviasi 2

dim2 = np.array([
    [1,2,3],
    [4,5,6]
])

print(dim2)
print(dim2.mean())
print(dim2.std())

print("hasil : ")
print("rentang terkecil : ", dim2.mean() - dim2.std())
print("rentang terbesar : ", dim2.mean() + dim2.std())
print(f"rentang nilai : {dim2.mean() - dim2.std()} - {dim2.mean() + dim2.std()}")

# print("outlier bawah : ", dim2[dim2 < dim2.mean() - dim2.std()])
# print("outlier atas : ", dim2[dim2 > dim2.mean() + dim2.std()])
```

```python
# assignment 1

data_s = np.array([
    [1, "Januari", 4799],
    [2, "Februari", 3154],
    [3, "Maret", 4238],
    [4, "April", 2929],
    [5, "Mei", 1615]
])

# assignment 2

hasil = 0
for i in data_s[0:, 2]:
  angka = int(i)
  hasil += angka

print(hasil)

# assignment 3

rata = hasil / len(data_s[0:, 2])
print(rata)

# assignment 4

for baris in range(len(data_s)):
  for kolom in range(len(data_s[baris])):
    if kolom == 2:
      angka = int(data_s[baris][kolom]) * 2
      data_s[baris][kolom] = angka

print(data_s)

# assignment 5

for baris in range(len(data_s)):
  for kolom in range(len(data_s[baris])):
    if kolom == 2:
      data_s[baris][kolom] = 5000

print(data_s)
```
