```python
import numpy as np

def cari(a):
    total = a.sum()
    rata = a.mean()
    standar = np.std(a)
    return total, rata, standar

data1 = []

for i in range(5):
    data1.append(i)

angka1 = np.array(data1)

angka1_sum, angka1_mean, angka1_std = cari(angka1)
hasil_angka1 = angka1_sum + angka1_mean + angka1_std
```
