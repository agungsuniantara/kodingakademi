```python
class person:
    nama = "dino"
    alamat = "denpasar"
    umur = 20
    
    def tampil(self):
        print("nama = ", self.nama)
        print("alamat = ", self.alamat)
        print("umur = ", self.umur)

class mahasiswa:
    def __init__(self, a, b, c):
        self.nama = a
        self.alamat = b 
        self.umur = c
    
    def tampil(self):
        print("nama = ", self.nama)
        print("alamat = ", self.alamat)
        print("umur = ", self.umur)

data_diri = person()
data_diri.tampil()
print(data_diri.nama)

data_mahasiswa1 = mahasiswa("agung", "denpasar", 30)
data_mahasiswa2 = mahasiswa("bagus", "tabanan", 30)
data_mahasiswa1.tampil()
print(data_mahasiswa1.nama)
data_mahasiswa2.tampil()
print(data_mahasiswa2.nama)
```

```python
class person:
    def __init__(self, a, b, c):
        self.nama = a
        self.alamat = b 
        self.umur = c
        
    def tampil(self):
        print("nama = ", self.nama)
        print("alamat = ", self.alamat)
        print("umur = ", self.umur)

class mahasiswa(person):
    def __init__(self, a, b, c, d):
        super().__init__(a, b, c)
        self.nim = d
    
    def tampil_nim(self):
        print("nim = ", self.nim)

class bayar(mahasiswa):
    def __init__(self, a, b, c, d, e):
        super().__init__(a, b, c, d)
        self.ukt = e
    
    def tampil_ukt(self):
        print("ukt = ", self.ukt)


data1 = bayar("dino", "denpasar", 20, "10000", 9000000)
print(data1.nama)
data1.tampil()
data1.tampil_nim()
```
