CREATE SERIES DATA

```python
import pandas as pd 
import numpy as np 
```

```python
# series index

data = np.array([10,20,30,40])
series = pd.Series(data, index=[1,2,3,4])
series
```

```python
# series list 

data = [10,20,30,40]
series = pd.Series(data, index=[5,6,7,8])
series
```

```python
# series dictionary 

data = {
    "koding" : "python",
    "Tema" : "Pandas"
}

series = pd.Series(data)
series
```

```python
# series scalar 

series = pd.Series(20, index = [1,2,3,4])
series
```

SLICING DATA SERIES

```python
# select index series

data = ["Numpy", "Pandas", "Python"]
series = pd.Series(data, index = [1,2,3])
print(series)
print(f"label index 1 : {series[1]}")

series_str = pd.Series(data, index = ["satu", "dua", "tiga"])
print(series_str)
print(f"label index dua : {series_str["dua"]}")
```

```python
# slicing series 

data = ["Numpy", "Pandas", "Python"] # 0, 1, 2
series = pd.Series(data, index = [1,2,3])
print(series[0:2])
print(series[0:3]) 

# slicing tetap menggunakan index asli yang dimulai dari 0
```

```python
# slicing loc & iloc

data = ["Numpy", "Pandas", "Python"]
series = pd.Series(data, index = [1,2,3])
series2 = pd.Series(data, index = ["satu","dua","tiga"])

print(series.loc[1:3])
print(series2.loc["satu":"tiga"])

print(series.iloc[0:3])
print(series2.iloc[0:3])

# loc : slicing dengan index yang ditentukan
# iloc : slicing dengan index sebenarnya atau mulai dari 0
```
