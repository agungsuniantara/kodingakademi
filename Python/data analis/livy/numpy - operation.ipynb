```python
import numpy as np 

# sum 1 dimensi 

data = np.array([1,2,3,4])
print(data)
print("Jumlah data : ", data.sum())
```

```python
# sum 2 dimensi

data2d = np.array([
    [1,2,3,4],
    [5,6,7,8]
])

print("Jumlah : ", data2d.sum())
print("Jumlah baris : ", data2d.sum(axis=1))
print("Jumlah kolom : ", data2d.sum(axis=0))
```

```python
# mean

print("Rata-rata : ", data2d.mean())
print("Rata-rata baris : ", data2d.mean(axis=1))
print("Rata-rata kolom : ", data2d.mean(axis=0))
```

```python
# varian

data2d = np.array([
    [1,2,3,4],
    [5,6,7,8]
])

print(data2d)
print(data2d.var())
```

```python
# standar deviasi

rata = data2d.mean()
std = data2d.std()
print("std : ", std)
print("Jarak terkecil array : ", rata - std)
print("Jarak terbesar array : ", rata + std)
```

```python
print(data * 2)
print(data2d[1] * 3)
print(np.add(data, 3))
```
