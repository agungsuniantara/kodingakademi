```python
import numpy as np 
import random

def rata(a,b):
    rata1 = np.mean(a)
    rata2 = np.mean(b)
    print("rata1 : ", rata1)
    print("rata2 : ", rata2)
    return rata1 + rata2

data1 = np.random.randint(100, size=(1,5))
data2 = np.random.randint(100, size=(1,5))

hasil_mean = rata(data1, data2)
print("hasil mean = ", hasil_mean)

```

```python
def test1(): # function tanpa parameter
    print("hello")

def test2(a,b): # function dengan parameter
    print(a + b)
    
def test3(a,b): # function return
    return a * b

data1 = 10
data2 = 20

test1()
test2(data1, data2)
hasil_test3 = test3(data1, data2)

print(hasil_test3)
```
