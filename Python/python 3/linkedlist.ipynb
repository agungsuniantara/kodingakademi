```python
# Membuat linked list kosong
linked_list = []

# Menambahkan node pertama
node1 = {"data": "Agung", "next": None}
linked_list.append(node1)

# Menambahkan node kedua
node2 = {"data": "Anang", "next": None}
linked_list[-1]["next"] = len(linked_list)  # Menghubungkan node pertama ke node kedua
linked_list.append(node2)

# Menambahkan node ketiga
node3 = {"data": "Siti", "next": None}
linked_list[-1]["next"] = len(linked_list)  # Menghubungkan node kedua ke node ketiga
linked_list.append(node3)

# Menampilkan isi linked list
current_index = 0
while current_index is not None:
    current_node = linked_list[current_index]
    print(current_node["data"], end=" -> ")
    current_index = current_node["next"]
print("None")
```
