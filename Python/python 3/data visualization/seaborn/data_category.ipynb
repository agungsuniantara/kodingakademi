LOAD DATA


```python
import pandas as pd

csv = pd.read_csv('penyakit_jantung.csv')
source = csv[["sex","age","cp","trestbps","chol"]]
source.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

BARPLOT


```python
# barplot rata-rata

sns.barplot(x = "sex", y = "chol", data = source)
plt.title('Bar plot')
plt.show()
```

COUNTPLOT


```python
sns.countplot(x="sex", data = source)
plt.title('Count plot')
plt.show()
```

BOXPLOT


```python
sns.boxplot(x = "sex", y = "trestbps", data = source)
plt.title('Box plot')
plt.show()
```

VIOLINPLOT


```python
sns.violinplot(x = "sex", y = "trestbps", data = source)
plt.title('Violinplot')
plt.show()
```

STRIP PLOT


```python
sns.stripplot(x = "sex", y = "trestbps", data = source)
plt.title('Strip plot')
plt.show()
```

SWARM PLOT


```python
sns.swarmplot(x = "sex", y = "trestbps", data = source)
plt.title('Swarm Plot')
plt.show()
```

HEATMAP

```python
pivot = source.pivot_table(index="sex", columns="age", values="chol")
pivot = pivot.fillna(0)
pivot
```

```python
sns.heatmap(pivot)
plt.title('Heatmap Plot')
plt.show()
```

```python
sns.barplot(x = "sex", y = "chol", hue = "cp", data = source)
plt.title('Barplot hue')
plt.show()
```
