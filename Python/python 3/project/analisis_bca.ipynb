```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 50)
```

```python
data = pd.read_csv("bca/tabel 1")
data
```

```python
data['ASET'] = data['ASET'].str.strip()
data['ASET'] = data['ASET'].str.replace('a.', '').str.replace('b.', '').str.replace('c.', '').str.replace('d.', '') 
data['ASET'] = data['ASET'].str.replace('-/-', '').str.replace('\r', '').str.strip()
data = data.map(lambda x: x.replace('(', '').replace(')', '').replace('-', '0') if isinstance(x, str) else x)
data
```
