```python
class mahasiswa: 
    def __init__(self):
        self.data = {}
    def tambah(self, nama, nim, alamat, jurusan, usia):
        self.data[nim[0:2]] = [nama, alamat, jurusan, usia]
    def tampil(self):
        for key, value in self.data.items():
            print(f"Nim : 20{key}")
            print(f"Nama = {value[0]}")
            print(f"Alamat = {value[1]}")
            print(f"Jurusan = {value[2]}")
            print(f"Usia = {value[3]}")

murid = mahasiswa()
murid.tambah("agung", "2020", "denpasar", "sistem", "10 tahun")
murid.tambah("wahyu", "1111", "denpasar", "sistem", "10 tahun")
murid.tampil()
        
```
