```python
import yfinance as yf

# Mengambil data saham BBCA dari Yahoo Finance
ticker = yf.Ticker("BBCA.JK")

# Mendapatkan data harga historisd
data = ticker.history(period="2y")  # 1 tahun terakhir
print(data)

# Mengambil data fundamental
info = ticker.info
print("Nama Perusahaan:", info["longName"])
print("Harga Saat Ini:", info["currentPrice"])
```
