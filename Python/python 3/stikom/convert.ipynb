```python
import pandas as pd

# Membaca file Excel
df = pd.read_excel("Prak. Basis Data - DI243 (1).xlsx", engine="openpyxl")

# Menghapus baris pertama (indeks 0) dan mengatur ulang indeks
df = df.drop(index=0).reset_index(drop=True)

# Menggunakan baris pertama setelah penghapusan sebagai header baru
df.columns = df.iloc[0].fillna("Unknown")  # Mengganti NaN dengan "Unknown"

# Menghapus baris pertama karena sudah digunakan sebagai header
df = df[1:].reset_index(drop=True)

# Menghapus kolom dengan nama NaN (jika masih ada)
df = df.loc[:, ~df.columns.astype(str).str.lower().str.contains("nan")]

df = df.drop(columns=["Unknown"])

df = df.fillna("-")

df
```

```python
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4, landscape
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle
import pandas as pd

# Nama file PDF
pdf_file = "nilai_algoritma.pdf"

# Membuat dokumen PDF
pdf = SimpleDocTemplate(pdf_file, 
                        pagesize=landscape(A4),
                        )

table_data = [df.columns.tolist()] + df.values.tolist()

# Membuat tabel
table = Table(table_data)

table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
    ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
    ('BACKGROUND', (0, 1), (-1, -1), colors.beige),
    ('GRID', (0, 0), (-1, -1), 1, colors.black),
]))

# Menyimpan PDF
pdf.build([table])

print(f"File PDF berhasil disimpan: {pdf_file}")
```
