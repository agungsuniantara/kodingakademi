```python
import hvplot.pandas
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("penyakit_jantung.csv")
data
```

LINE PLOT


```python
data.hvplot(y = "chol", 
            by = "sex", 
            title='Line Plot', 
            legend='right', 
            kind="line")

# x = jumlah data
```

SCATTER PLOT


```python
data.hvplot(y = "chol", 
            by = "sex", 
            title='Scatter plot', 
            legend='right', 
            kind="scatter")
```

PLOT HISTOGRAM


```python
data.hvplot(y = ["trestbps", "chol"], 
            title='Line Plot', 
            legend='right', 
            alpha = 0.5,
            kind="hist")
```

BAR PLOT


```python
data.hvplot(y = "chol", 
            x = "sex", 
            title='Bar plot', 
            legend='bottom', 
            kind="bar",
            width=800, height=400)
```

PLOT AREA


```python
area_plot = data.hvplot.area(x='index', y='C', title='Area Plot', ylabel='Value')
area_plot
```

BOX PLOT


```python
box_plot = data.hvplot.box(y='A', title='Box Plot', ylabel='Value')
box_plot
```

HEATMAP PLOT


```python
heatmap_data = pd.DataFrame(np.random.rand(10, 10), columns=list('ABCDEFGHIJ'))
heatmap = heatmap_data.hvplot.heatmap(title='Heatmap', xlabel='Columns', ylabel='Rows')
heatmap
```

KOMBINASI PLOT


```python
# Membuat plot garis untuk kolom 'A' dan 'B'
line_plot_A = data.hvplot.line(y='B', line_color='blue', legend='top_left', title='Combined Line Plot')
line_plot_B = data.hvplot.line(y='C', line_color='red', legend='top_left')

# Menggabungkan kedua plot dalam satu canvas
combined_plot = line_plot_A * line_plot_B
combined_plot
```
