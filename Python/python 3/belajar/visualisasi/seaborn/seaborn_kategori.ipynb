SEABORN DATA KATEGORI


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from warnings import filterwarnings
filterwarnings('ignore')

source = pd.read_csv('penyakit_jantung.csv')
source
```

BARPLOT


```python
# barplot rata-rata

sns.barplot(x = "sex", y = "chol", data = source)
```

```python
# barplot sum

sns.barplot(x = "sex", y = "chol", data = source, estimator=sum)
```

```python
# barplot standar deviasi

sns.barplot(x = "sex", y = "chol", data = source, estimator=np.std)
```

COUNTPLOT


```python
# countplot

sns.countplot(x = "sex", data = source)
```

```python
sns.countplot(x = "sex", hue = "cp", data = source)
```

```python
sns.countplot(x = "sex", hue = "cp", order = ["male", "female"], data = source)
```

BOXPLOT


```python
# boxplot

sns.boxplot(x="sex", y = "chol", data = source, color="#3d4f8f")
```

```python
source["restecg"].unique()
```

```python
sns.boxplot(x="sex", y = "chol", hue = "cp", data = source, color="red")
```

```python
source["cp"]
```

```python
sns.boxplot(x="sex", y = "chol", hue = "cp", hue_order = [3, 2, 1, 0], data = source)
```

VIOLIN PLOT


```python
# violin plot

sns.violinplot(x="sex", y = "chol", data = source)
```

```python
sns.violinplot(x="sex", y = "chol", hue = "cp", order = ["male", "female"], data = source)

# order digunakan untuk mengatur tata letak x
```

```python
sns.violinplot(x="sex", y = "chol", hue = "cp", split = True, data = source)

# split menggabungkan grafik kolom kategori (hue)
# split hanya dapat digunakan ketika hue nya hanya 2 kategori
```

STRIP PLOT


```python
sns.stripplot(x="sex", y="chol", data=source)
```

```python
sns.stripplot(x="sex", y="chol", hue="cp", data=source)
```

```python
sns.stripplot(x="sex", y="chol", hue="cp", jitter = True, data=source)

# jitter digunakan agar lebih menyebar
```

SWARM PLOT


```python
plt.figure(figsize=(6,4))
sns.swarmplot(x="sex", y="chol", hue="cp", data=source)

# swarmplot itu gabungan scatterplot dan violin plot
```

FACETGRID


```python
source.head()
```

```python
p = sns.FacetGrid(source, row="sex", col="cp")
p.map(sns.distplot, "chol", kde = False) # "chol" = x
plt.show()
```

```python
p = sns.FacetGrid(source, row="sex", col="cp", hue="age")
p.map(sns.scatterplot, "chol","trestbps") # "chol" = x, "trestbps" = y
plt.legend()
plt.show()
```
