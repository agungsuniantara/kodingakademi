```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1,20,0.5)
y = x**2
```

```python
fig1 = plt.figure(figsize=(8,4), dpi = 100) # figsize = (width, height)
ax = fig1.add_axes([0,0,1,1])
ax.plot(x,y)
```

```python
fig2, axes = plt.subplots(2,1, figsize=(4,3), dpi=100)

axes[0].plot(x,y)
axes[0].set_title('Grafik 1')

axes[1].plot(x,y)
axes[1].set_title('Grafik 2')

plt.tight_layout()

plt.show()
```

```python
fig2.savefig('test.png')
```

LEGEND

```python
fig3 = plt.figure(facecolor='yellow',dpi = 200)

axes2 = fig3.add_axes([0,0,1,1])

axes2.plot(x,y, label = "Pangkat 2")

axes2.legend(loc=(0.4, 0.8)) # loc = (left, bottom)
axes2.grid(True)

plt.show()
```

```python
fig4 = plt.figure(dpi = 200)

axes3 = fig4.add_axes([0,0,1,1])

axes3.plot(x,y, label = "Pangkat 2", lw=1, ls='-') # lw = lineweight, ls = linestyle
axes3.plot(x,x**3, label = "Pangkat 3", lw=2, ls='--') # lw = lineweight, ls = linestyle
axes3.plot(x,x**4, label = "Pangkat 4", lw=3, ls='-.') # lw = lineweight, ls = linestyle

axes3.legend(loc=(0.4, 0.8)) # loc = (left, bottom)
axes3.grid(True)

plt.show()
```

```python
fig5 = plt.figure(dpi = 200)

axes4 = fig5.add_axes([0,0,1,1])

axes4.plot(x,y, label = "Pangkat 2", marker='o', markersize=4, mfc = 'black', mew=7, mec='red')
axes4.plot(x,x**3, label = "Pangkat 3", marker='+', markersize=1, mfc = 'black', mew=15, mec='blue')

axes4.legend(loc=(0.4, 0.8)) # loc = (left, bottom)
axes4.grid(True)

plt.show()
```

LIMIT

```python
fig6 = plt.figure(dpi = 200)

axes5 = fig6.add_axes([0,0,1,1])

axes5.plot(x,y, label = "Pangkat 2", lw=1, ls='-') 
axes5.plot(x,x**3, label = "Pangkat 3", lw=2, ls='--') 
axes5.plot(x,x**4, label = "Pangkat 4", lw=3, ls='-.') 

axes5.legend(loc=(0.4, 0.8)) # loc = (left, bottom)
axes5.grid(True)
axes5.set_xlim([0, 12]) # limit x (0 - 12)
axes5.set_ylim([0, 8000]) # limit y (0 - 8000)

plt.show()
```
