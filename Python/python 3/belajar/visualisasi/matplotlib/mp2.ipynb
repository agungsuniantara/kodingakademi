```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1,20,0.5)
y = x**2
```

```python
x
```

```python
y
```

```python
# versi 1

fig = plt.figure()
axes = fig.add_axes([0,0,0.9,0.9]) # (left, bottom, width, height)
axes.plot(x,y)
axes.set_xlabel('Sumbu X')
axes.set_ylabel('Sumbu Y')
axes.set_title('Judul grafik')
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0,0,0.9,0.9]) # (left, bottom, width, height)
axes1.plot(x,y, color='red')
axes1.set_xlabel('Sumbu X')
axes1.set_ylabel('Sumbu Y')
axes1.set_title('Judul grafik')

axes2 = fig.add_axes([0.1,0.5,0.3,0.3]) # (left, bottom, width, height) 0.1 = 10%
axes2.plot(y,x, color = 'blue')
axes2.set_xlabel('Sumbu X')
axes2.set_ylabel('Sumbu Y')
axes2.set_title('Judul grafik kecil')
```

```python
# versi 2

fig, axes3 = plt.subplots(3,2) # (baris, kolom)
fig.tight_layout()
```

```python
fig, axes3 = plt.subplots(1,2) # (baris, kolom)

axes3[0].plot(x,y, color='red')
axes3[0].set_xlabel('Sumbu X')
axes3[0].set_ylabel('Sumbu Y')
axes3[0].set_title('Judul grafik')

axes3[1].plot(y,x, color='pink')
axes3[1].set_xlabel('Sumbu X')
axes3[1].set_ylabel('Sumbu Y')
axes3[1].set_title('Judul grafik')

fig.tight_layout()
```

```python
fig, axes3 = plt.subplots(2,2) # (baris, kolom)

# baris 1 kolom 1
axes3[0][0].plot(x,y, color='red')
axes3[0][0].set_xlabel('Sumbu X')
axes3[0][0].set_ylabel('Sumbu Y')
axes3[0][0].set_title('Judul grafik')

# baris 1 kolom 2
axes3[0][1].plot(y,x, color='blue')
axes3[0][1].set_xlabel('Sumbu X')
axes3[0][1].set_ylabel('Sumbu Y')
axes3[0][1].set_title('Judul grafik')

# baris 2 kolom 1
axes3[1][0].plot(y,x, color='orange')
axes3[1][0].set_xlabel('Sumbu X')
axes3[1][0].set_ylabel('Sumbu Y')
axes3[1][0].set_title('Judul grafik')

# baris 2 kolom 2
axes3[1][1].plot(y,x, color='pink')
axes3[1][1].set_xlabel('Sumbu X')
axes3[1][1].set_ylabel('Sumbu Y')
axes3[1][1].set_title('Judul grafik')

fig.tight_layout()
```
