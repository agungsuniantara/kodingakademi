distribusi uniform

```python
import numpy as np 

# bilangan distribusi uniform (0-1)

np.random.rand(7)
```

```python
# bilangan distribusi uniform (0-1) matriks 2 dimensi

np.random.rand(5,4)
```

distribusi normal (gaussian)

```python
# bilangan distribusi normal (gaussian)

# gausian dari - tak terhingga sampai + tak terhingga

np.random.randn(5)
```

```python
# bilangan distribusi normal (gaussian) matriks 2 dimensi

np.random.randn(4,4)
```

DEKLARASI LANJUT

```python
from numpy.random import rand, randn, randint

randn(6,2)
```

```python
randint(0,10)
```

```python
randint(0,10,5)
```

```python
bilangan = randint(0, 100, 20) # (awal, batas, jumlah)
bilangan
```

```python
bilangan.reshape(4,5) # (baris, kolom) jumlah baris dan kolom harus sesuai dengan jumlah data yang di reshape
```

```python
coba = randint(0,10,20).reshape(5,4)

coba
```

```python
coba = coba.reshape(2,10)

coba
```
