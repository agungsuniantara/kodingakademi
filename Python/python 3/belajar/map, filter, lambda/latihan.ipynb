MAP

```python
# map

def hitung(angka):
    return angka * 10

data = [1,2,3]

print(map(hitung, data))

for i in map(hitung, data):
    print(i)

hasil = list(map(hitung, data))
hasil
```

```python
# map

def cek_string(kata):
    if len(kata) % 2 :
        return "kata genap"
    else:
        return "kata ganjil"

data2 = ["dasda", "dasdsada", "sss"]

list(map(cek_string, data2))
```

FILTER

```python
# filter

def cek_angka(angka):
    return angka % 2 == 0 # kalau hasil == 0 (True)

angka_baru = [12,13,14,121,11,323,4]

list(filter(cek_angka, angka_baru))
```

map = digunakan ketika return tidak boolean, filter = digunakan ketika return boolean

LAMBDA

```python
# lambda

fungsi = lambda x,y: print(f"x = {x * 10} y = {y * 20}")
fungsi(10, 20)
```

```python
# Lambda map

lambda_map = list(map(lambda x: x * 10, data))
lambda_map
```

```python
# Lambda filter

lambda_filter = list(filter(lambda x: x % 2 == 0, angka_baru))
lambda_filter
```

Perbedaan lambda dan def, lambda digunakan untuk sekali pakai tidak seperti def yang dapat digunakan berulang kali serta def dapat memiliki fungsi yang lebih banyak.
