```python
# slicing numpy

import numpy as np

data = np.array([1,2,3,4])
print(data[0:]) # zero until finish
print(data[0:2]) # zero until before two
```

```python
# numpy arange

data_arange1 = np.arange(20)
data_arange2 = np.arange(1, 10)
data_arange3 = np.arange(1, 20, 2)

print(data_arange1)
print(data_arange2)
print(data_arange3)
```

```python
lst = []

for i in data_arange2:
    if i % 2 == 0:
        lst.append(i)

print(f"D : ", lst)
```
