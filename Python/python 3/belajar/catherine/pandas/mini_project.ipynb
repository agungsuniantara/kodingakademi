```python
import pandas as pd 

data = pd.read_csv("bali_tourist (3).csv")
data
```

```python
area = data["area"]
area
```

```python
area = data[["area", "country"]]
area
```

```python
import matplotlib.pyplot as plt

data_2018 = data[data["year"] == 2018] 
mean_2018 = data_2018.iloc[0:, 3:].sum().mean()

data_2019 = data[data["year"] == 2019] 
mean_2019 = data_2019.iloc[0:, 3:].sum().mean()

# Buatkan visualisasi
years = ['2018', '2019']
means = [mean_2018, mean_2019]

plt.figure(figsize=(10, 5))
plt.bar(years, means, color=['blue', 'orange'])
plt.xlabel('Tahun')
plt.ylabel('Rata-rata')
plt.title('Rata-rata Kunjungan Wisatawan per Tahun')
plt.show()
```
