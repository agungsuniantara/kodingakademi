DATA CLEANING WITH PANDAS


CHECK DATA NAN

```python
import pandas as pd
import numpy as np

dict = {  
    'I':[100, 90, np.nan, 95],
    'II': [30, 45, 56, np.nan],
    'III':[np.nan, 40, 80, 98]
}
df = pd.DataFrame(dict)
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.notnull()
```

FILL DATA NAN

```python
dict = {  
    'I':[100, 90, np.nan, 95],
    'II': [30, 45, 56, np.nan],
    'III':[np.nan, 40, 80, 98]
}
df = pd.DataFrame(dict)
df
```

```python
df = df.fillna(1)
df
```

```python
df = df.fillna(df.mean())
df
```

DELETE DATA NAN

```python
dict = {  
    'I':[100, 90, np.nan, 95],
    'II': [30, 45, 56, np.nan],
    'III':[np.nan, 40, 80, 98]
}
df = pd.DataFrame(dict)
df
```
