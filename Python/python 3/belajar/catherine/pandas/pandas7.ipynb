DATA CLEANING DUPLICATE

```python
import pandas as pd

country= pd.Series([
    'France',
    'United Kingdom',
    'United Kingdom',
    'Italy',
    'Germany',
    'Germany',
    'Germany',
], index=[
    'Country 1',
    'Country 2',
    'Country 3',
    'Country 4',
    'Country 5',
    'Country 6',
    'Country 7'
])

country
```

CHECK SERIES DATA DUPLICATE

```python
country.duplicated()
```

```python
country.duplicated(keep="last")
```

DELETE SERIES DATA DUPLICATE

```python
country.drop_duplicates()
```

```python
country.drop_duplicates(keep="last")
```

```python
country.drop_duplicates(keep=False)
```

CHECK DATAFRAME DUPLICATE

```python
students = pd.DataFrame({
    'Name': [
        'Kobe Bryant',
        'LeBron James',
        'Kobe Bryant',
        'Carmelo Anthony',
        'Kobe Bryant',
    ],
    'Course': [
        'Python',
        'Unity',
        'Python',
        'Web Design',
        'Web Programming'
    ]
})

students
```

```python
students.duplicated()
```

```python
students.duplicated(subset=['Name'])
```
