DATA SERIES OPERATION

```python
import numpy as np 
import pandas as pd 

data1 = np.array([1,2,3,4])
data2 = np.array([12,1,22,21])
series1 = pd.Series(data1)
series2 = pd.Series(data2)
print(series1)
print(series2)
```

```python
# series logic

logic1 = series2[series2 > 10]
print(series2[series2 > 10])
print(series2[series2 > series2.mean()])
print(series2[(series2 > 10) & (series2 < 20)])
```

```python
# series math

print(series2 + 10)
print(series2 - 10)
print(series1.add(series2))
print(series2.subtract(series1))
```

```python
# series konversi

print(f"before : {series1.dtype}")
series1 = series1.astype(str)
print(f"after : {series1.dtype}")
```

```python
# series change data

series2[1] = 100
print(series2)
```
