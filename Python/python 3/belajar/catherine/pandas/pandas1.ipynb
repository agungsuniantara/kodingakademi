DATA SERIES

```python
import numpy as np 
import pandas as pd 

data_series = np.array([1,2,3,4])
series = pd.Series(data_series)
series
```

```python
# series with string

data_series = np.array(["satu", "dua", "tiga"])
series = pd.Series(data_series)
series
```

```python
# series with list

data_series = ["satu", "dua", "tiga"]
series = pd.Series(data_series, index = [10,11,12])
series
```

```python
# series with dictionary

data_dictionary = {
    "koding" : 90,
    "akademi" : 100
}
series = pd.Series(data_dictionary)
series
```

```python
# series with skalar

series = pd.Series(99, index = [10,11,12])
series
```

```python
# edit index 1

data_series = np.array(["satu", "dua", "tiga"])
series = pd.Series(data_series, index = [10,11,12])
series
```

```python
# edit index 2

data_series = np.array([1,2,3])
series = pd.Series(data_series, index = ["satu","dua","tiga"])
series
```

```python
# akses elemen series

data_series = np.array([1,2,3])
series = pd.Series(data_series, index = ["satu","dua","tiga"])
print(series[1])
print(series[0:2])
print(series["tiga"])
```

```python
# loc & iloc

data_series = np.array([1,2,3])
series = pd.Series(data_series, index = ["satu","dua","tiga"])
print("loc : ")
print(series.loc["satu" : "tiga"])
print("iloc : ")
print(series.iloc[0 : 3])
```
