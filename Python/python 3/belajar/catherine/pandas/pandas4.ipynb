DATAFRAME CONDITIONAL


```python
import pandas as pd 

df = pd.DataFrame({
    "population" : [10,22,33,44],
    "gdp" : [11,99,22,33],
    "surface area": [88,66,77,44],
    "hdi" : [11,22,33,44],
    "continent" : [19,18,17,12]
}, columns=["population","gdp","surface area","hdi","continent"], 
                  index = ["canada", "france", "germany", "italy"])

df
```

```python
# kondisional dataframe

print("logic 1 : ")
print(df["population"] > 30)
print("logic 2 : ")
print(df.loc[df["population"] > 30])
print("logic 3 : ")
print(df.loc[df["population"] > 30, "population"])
```

```python
# filter kondisi dataframe

logic = df[df["population"] > 20]
print(logic)
```

```python
# filter multi kondisi dataframe

logic2 = df[(df["population"] > 40) & (df["surface area"] > 10)]
print(logic2)
```

```python
# delete index / line data dataframe 

print("del col : ")
print(df.drop(columns="continent"))
print(df.drop(["population", "gdp"], axis="columns"))
print(df.drop(["population", "gdp"], axis=1))
print(df.drop(["germany", "italy"], axis=0))
```

```python
# broadcasting dataframe 

print(df["population"] + 2)
```

```python
# biner operation dataframe

series = pd.Series([10,50], index=['gdp', 'hdi'])
df[["gdp", "hdi"]] + series
```

MODIFICATION DATAFRAME


```python
# add data column with series

series = pd.Series([11,89,22,25], index = ["canada", "france", "germany", "italy"])
df["new1"] = series
df
```

```python
# add data column with operation

df["new2"] = df["population"] * df["gdp"]
df
```

```python
# change column name 

df = df.rename(columns={"new1" : "ganti1"})
df
```

```python
# change index name

df = df.rename(index={"germany" : "indonesia"})
df
```

```python
# change column and index name

df = df.rename(
    columns={
        "new2" : "ganti2"
    },
    index={
        "italy" : "kalimantan"
    }
)

df
```
