DATA DATAFRAME

```python
import pandas as pd 

data = [10,4,19,20,40]
df = pd.DataFrame(data)
df
```

```python
# dataframe with list 2 dimension

data = [
    [10,9,22],
    [20,11,23]
]

df = pd.DataFrame(data)
df
```

```python
# dataframe with key and value

df = pd.DataFrame({
    "population" : [10,22,33,44],
    "gdp" : [11,99,22,33],
    "surface area": [88,66,77,44],
    "hdi" : [11,22,33,44],
    "continent" : [19,18,17,12]
}, columns=["population","gdp","surface area","hdi","continent"])

df
```

```python
# set index dataframe

df.index = ["canada", "france", "germany", "italy"]
df
```

```python
# check dataframe columns name

df.columns
```

```python
# select column data

df[["population", "gdp"]]
```

```python
# select line data

print("with loc : ")
print(df.loc["canada"])
print("with iloc : ")
print(df.iloc[0])
```

```python
# slicing line data

print("with loc : ")
print(df.loc["canada" : "germany"])
print("with iloc : ")
print(df.iloc[0 : 3])
```

```python
# slicing line and column data

print("with loc : ")
print(df.loc["canada" : "germany", "population" : "hdi"])
print("with iloc : ")
print(df.iloc[0 : 3, 0 : 4])
```
