```python
import matplotlib.pyplot as plt
import pandas as pd

plt.plot([1,2,3,4], [10,5,30,20], label = "Plot 1")
plt.plot([1,2,3,4], [15,20,5,10], label = "Plot 2")
plt.title('Title')
plt.xlabel('Sumbu x')
plt.ylabel('Sumbu y')
plt.xlim(0,5)
plt.ylim(0,40)
plt.xticks([1, 2, 3, 4])
plt.yticks([0, 10, 20, 30, 40])
plt.legend(loc = "upper left")
plt.show()
```

PENDEKATAN PROSEDURAL

```python
plt.subplot(121) # (baris, kolom, letak plot)
plt.plot([1,2,3,4], [10,5,30,20], color='red')
plt.xlabel('Sumbu x')
plt.ylabel('Sumbu y')

plt.subplot(122)
plt.plot([1,2,3,4], [15,20,5,10], color='blue')
plt.xlabel('Sumbu x')
plt.ylabel('Sumbu y')
plt.tight_layout()
plt.show()
```

PENDEKATAN OBJECT ORIENTED

```python
fig = plt.figure()
axes = fig.add_axes([0,0,0.4,0.9]) # (left, bottom, width, height)
axes.plot([1,2,3,4], [10,5,30,20], color="orange")
axes.set_xlabel('Sumbu x')
axes.set_ylabel('Sumbu y')
plt.show()
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0,0,1.3,1.2])
axes1.plot([1,2,3,4], [10,5,30,20], color='red')
axes1.set_xlabel('Sumbu x')
axes1.set_ylabel('Sumbu y')
axes1.set_title('Axes 1')

axes2 = fig.add_axes([0.2,0.8,0.3,0.3])
axes2.plot([1,2,3,4], [15,20,5,10], color = 'blue')
axes2.set_xlabel('Sumbu x')
axes2.set_ylabel('Sumbu y')
axes2.set_title('Axes 2')

plt.show()
```

```python
fig, axes = plt.subplots(1,2) # (baris, kolom)

# kolom 1
axes[0].plot([1,2,3,4], [10,5,30,20], color='red')
axes[0].set_xlabel('Sumbu x')
axes[0].set_ylabel('Sumbu y')
axes[0].set_title('Kolom 1')

# kolom 2
axes[1].plot([1,2,3,4], [15,20,5,10], color='orange')
axes[1].set_xlabel('Sumbu x')
axes[1].set_ylabel('Sumbu y')
axes[1].set_title('Kolom 2')

fig.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,2) # (baris, kolom)

# baris 1 kolom 1
axes[0][0].plot([1,2,3,4], [10,5,30,20], color='red')
axes[0][0].set_xlabel('Sumbu x')
axes[0][0].set_ylabel('Sumbu y')
axes[0][0].set_title('Baris 1 kolom 1')

# baris 1 kolom 2
axes[0][1].plot([1,2,3,4], [15,20,5,10], color='orange')
axes[0][1].set_xlabel('Sumbu x')
axes[0][1].set_ylabel('Sumbu y')
axes[0][1].set_title('Baris 1 kolom 2')

# baris 2 kolom 1
axes[1][0].plot([1,2,3,4], [5,30,10,14], color='blue')
axes[1][0].set_xlabel('Sumbu x')
axes[1][0].set_ylabel('Sumbu y')
axes[1][0].set_title('Baris 2 kolom 1')

# baris 2 kolom 2
axes[1][1].plot([1,2,3,4], [21,11,24,13], color='black')
axes[1][1].set_xlabel('Sumbu x')
axes[1][1].set_ylabel('Sumbu y')
axes[1][1].set_title('Baris 2 kolom 2')

fig.tight_layout()
plt.show()
```
