```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Membuat objek analyzer
analyzer = SentimentIntensityAnalyzer()

# Teks contoh
text = "The movie was amazing! I loved it so much."

# Tokenisasi kata
words = text.split()

# Akses kamus VADER
lexicon = analyzer.lexicon

word_scores = {word: lexicon.get(word.lower(), 0) for word in words}

print(word_scores)

# Hitung skor positif, negatif, dan total magnitudo
positive_total = sum(score for score in word_scores.values() if score > 0)
negative_total = sum(abs(score) for score in word_scores.values() if score < 0)
total_magnitude = positive_total + negative_total

# Hitung compound score
compound_score = (positive_total - negative_total) / total_magnitude if total_magnitude != 0 else 0

# Klasifikasi sentimen
if compound_score > 0.05:
    sentiment = "Positive"
elif compound_score < -0.05:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

# Hasil
print("Word Scores:", word_scores)
print("Positive Total:", positive_total)
print("Negative Total:", negative_total)
print("Compound Score:", compound_score)
print("Sentiment:", sentiment)
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from googletrans import Translator

# Membuat objek untuk translasi dan VADER
translator = Translator()
analyzer = SentimentIntensityAnalyzer()

# Teks dalam bahasa Bali
text_bali = "saya demen gati sama kamu"

# Terjemahkan ke bahasa Inggris
translated = translator.translate(text_bali, src='id', dest='en')

# Ambil teks hasil terjemahan
translated_text = translated.text

# Analisis sentimen
scores = analyzer.polarity_scores(translated_text)

# Hasil
print("Teks Asli (Bali):", text_bali)
print("Teks Terjemahan (Inggris):", translated_text)
print("Sentimen:", scores)
```
