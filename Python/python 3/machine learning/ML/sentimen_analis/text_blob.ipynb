Objek textblob

```python
from textblob import TextBlob

# Membuat objek TextBlob
text = "TextBlob is a simple library for processing text."

blob = TextBlob(text)
print(blob)
```

```python
text = "I love TextBlob. It is amazingly simple and efficient."
blob = TextBlob(text)

# Analisis Sentimen
sentiment = blob.sentiment
print("Polarity:", sentiment.polarity)
print("Subjectivity:", sentiment.subjectivity)
```
