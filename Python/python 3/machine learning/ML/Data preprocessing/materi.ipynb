LOAD DATA


```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
data['species'] = iris.target
data.head()
```

DATA CLEANING


```python
# Memeriksa nilai yang hilang
print(data.isnull().sum())

# Memeriksa dan menghapus duplikasi
data = data.drop_duplicates()
print(f"Jumlah data setelah menghapus duplikasi: {len(data)}")
```

DATA TRANSFORM


normalisasi


```python
from sklearn.preprocessing import MinMaxScaler

X = data.drop('species', axis=1)
y = data['species']

# Normalisasi fitur
scaler_normal = MinMaxScaler()
X_normalized = scaler_normal.fit_transform(X)

# Mengubah hasil normalisasi menjadi DataFrame
X_normalized = pd.DataFrame(X_normalized, columns=X.columns)
X_normalized.head()
```

Standardisasi


```python
from sklearn.preprocessing import StandardScaler

X = data.drop('species', axis=1)
y = data['species']

# Standardisasi fitur
scaler_standard = StandardScaler()
X_standardized = scaler_standard.fit_transform(X)

# Mengubah hasil standardisasi menjadi DataFrame
X_standardized = pd.DataFrame(X_standardized, columns=X.columns)
X_standardized.head()
```

Encoder


```python
from sklearn.preprocessing import OneHotEncoder

# Misalkan kita memiliki kolom kategorikal 'species' yang perlu di-encode
encoder = OneHotEncoder()
species_encoded = encoder.fit_transform(data[['species']]).toarray()

# Mengubah hasil encoding menjadi DataFrame
species_encoded = pd.DataFrame(species_encoded, columns=encoder.get_feature_names_out(['species']))
species_encoded.head()
```

reduksi dimensi PCA


```python
from sklearn.decomposition import PCA

# Reduksi dimensi dengan PCA menggunakan data yang distandarisasi
pca_standardized = PCA(n_components=2)
X_reduced_standardized = pca_standardized.fit_transform(X_standardized)

# Mengubah hasil PCA menjadi DataFrame
X_reduced_standardized = pd.DataFrame(X_reduced_standardized, columns=['Principal Component 1', 'Principal Component 2'])
X_reduced_standardized['species'] = y
X_reduced_standardized.head()
```

```python
from sklearn.model_selection import train_test_split

# Membagi data yang distandarisasi menjadi training dan test set
X_train_standardized, X_test_standardized, y_train_standardized, y_test_standardized = train_test_split(X_reduced_standardized.drop('species', axis=1), y, test_size=0.2, random_state=42)

print(f"Jumlah data training (standardized): {len(X_train_standardized)}")
print(f"Jumlah data testing (standardized): {len(X_test_standardized)}")
```
