SHAPIRO WILK


```python
sampel = data["thalach"][0:24]

x = sampel.sort_values().reset_index(drop = True)

thalach = {
    "x": x,
    "x - xmean": x - x.mean(),
    "(x - xmean)²": pow(x - x.mean(), 2),
    "ai" : koefisien.loc["a1":"a12", 24].reset_index(drop = True),
    "X n+1-i" : x[::-1].reset_index(drop = True)[0:12],
    "Xi" : x[0:12],
    "(X n+1-i - Xi)" : x[::-1].reset_index(drop = True)[0:12] - x[0:12],
}

data_thalach = pd.DataFrame(thalach)
data_thalach.fillna("-", inplace = True)
data_thalach
```

```python
# W_pembilang b2

ai = koefisien.loc["a1":"a12", 24].reset_index(drop = True)
z = x[::-1].reset_index(drop = True)[0:12] - x[0:12]
b = ai * z
b2 = pow(b.sum(), 2)

print(f"b2 = {b2}")
```

```python
# W_penyebut SS

ss = pow(x - x.mean(), 2).sum()

print(f"SS = {ss}")
```

```python
# W_thalach

w_thalach = b2 / ss

print(f"W = {w_thalach}")
```

```python
# linear interpolasi

tabel = p_value.loc[24]
n_24 = tabel.values

b = [n_24[n_24 < w_thalach][-1], n_24[n_24 > w_thalach][0]]
c = [float(key) for key, val in tabel.items() if val in b]

x = w_thalach
x1 = b[0]
x2 = b[1]
y1 = c[0]
y2 = c[1]

interpolasi = y1 + (x - x1) / (x2 - x1) * (y2 - y1)
interpolasi
```

```python
# logarithmic interpolasi

# Locate the appropriate row in the p-value table
tabel = p_value.loc[24]
a = tabel.values

# Find the two bounding p-values around w_thalach
w = [a[a < w_thalach][-1], a[a > w_thalach][0]]
p = [float(key) for key, val in tabel.items() if val in b]

# Interpolate to find the p-value using logarithmic interpolation
w1 = w[0]
w2 = w[1]
p1 = p[0]
p2 = p[1]

# Using the correct logarithmic interpolation formula
log_p1 = np.log(p1)
log_p2 = np.log(p2)
log_w1 = np.log(w1)
log_w2 = np.log(w2)
log_w_thalach = np.log(w_thalach)

log_p_value_interpolated = log_p1 + (log_w_thalach - log_w1) / (log_w2 - log_w1) * (log_p2 - log_p1)
p_value_interpolated = np.exp(log_p_value_interpolated)
p_value_interpolated
```

```python
from scipy.stats import shapiro

thalach_data = data['thalach'][0:24]
shapiro_test = shapiro(thalach_data)
print("Shapiro-Wilk Test:")
print("W:", shapiro_test.statistic)
print("P-Value : ", shapiro_test.pvalue)
```

```python
from scipy.stats import probplot

# Creating the Q-Q plot
plt.figure(figsize=(10, 6))
probplot(thalach_data, dist="norm", plot=plt)
plt.title('Q-Q Plot of Thalach Data')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("cause_death.csv")
df.head(4)
```

```python
# Ambil kolom 'Total Deaths'
total_deaths = df['Total Deaths']

# Uji Shapiro-Wilk
stat, p_value = shapiro(total_deaths)
print(f"Statistik: {stat:.4f}, p-value: {p_value:.4f}")
```

```python
from scipy.stats import probplot
import matplotlib.pyplot as plt

# Q-Q Plot
plt.figure(figsize=(7, 3))
probplot(total_deaths, dist="norm", plot=plt)
plt.title('Q-Q Plot of Total Deaths')  # Sesuaikan judul jika perlu
plt.xlabel('')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
import numpy as np

total_deaths = total_deaths.sort_values().reset_index(drop=True)

q1 = np.percentile(total_deaths, 25)
q3 = np.percentile(total_deaths, 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

total_deaths = total_deaths[(total_deaths >= lower_bound) & (total_deaths <= upper_bound)]
total_deaths
```

```python
# Q-Q Plot
plt.figure(figsize=(7, 3))
probplot(total_deaths, dist="norm", plot=plt)
plt.title('Q-Q Plot of Total Deaths')  # Sesuaikan judul jika perlu
plt.xlabel('')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
