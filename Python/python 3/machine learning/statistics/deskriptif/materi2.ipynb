```python
import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
data = pd.read_csv("penyakit_jantung.csv")
thalach = data["thalach"]
thalach.head()
```

```python
mean_val = np.mean(thalach)
print("Mean:", mean_val)
```

```python
median_val = np.median(thalach)
print("Median:", median_val)
```

```python
mode_val = stats.mode(thalach)[0]
print("Mode:", mode_val)
```

```python
variance_val = np.var(thalach)
std_dev_val = np.std(thalach)

print("Variance:", variance_val)
print("Standard Deviation:", std_dev_val)

plt.figure(figsize=(10, 6))
sns.scatterplot(x=np.arange(len(thalach)), y=thalach, s=30, label='Data Points')
plt.axhline(mean_val, color='red', linestyle='dashed', label=f'Mean: {mean_val:.2f}')
plt.fill_between(np.arange(len(thalach)), mean_val - std_dev_val, mean_val + std_dev_val, color='gray', alpha=0.2, label=f'1 Std Dev')
plt.title('Scatter Plot Standard Deviation')
plt.legend()
plt.grid(True)
plt.show()
```
