# Normalitas Kelembaban

```python
# x

sort = df['Kelembaban (%)'].sort_values()

hasil_sort = sort.reset_index(drop = True)

hasil_sort
```

```python
# x - 𝑥̄

𝑥̄ = hasil_sort.mean() # rata-rata

hasil_kurang = hasil_sort - 𝑥̄

print(f"mean = {𝑥̄}")

print(hasil_kurang)
```

```python
# (x - 𝑥̄)²

hasil_pangkat = pow(hasil_sort - 𝑥̄, 2)
hasil_pangkat
```

```python
# ai

koefisien_24 = koefisien.loc["a1":"a12", 24]
hasil_koefisien = koefisien_24.reset_index(drop = True)
hasil_koefisien
```

```python
# X n+1-i

iklim = hasil_sort[::-1]

iklim_descending = iklim.reset_index(drop = True)

descending = iklim_descending[0:12]

descending
```

```python
# Xi

data_x = hasil_sort[0:12]

data_x
```

```python
# (X n+1-i - Xi)

hasil_pengurangan = descending - data_x

hasil_pengurangan
```

```python
# ai(X n+1-i - Xi)

hasil_akhir = hasil_koefisien * hasil_pengurangan
hasil_akhir
```

```python
# kelembaban

normalitas_iklim = {
    "x": hasil_sort,
    "x - 𝑥̄": hasil_kurang,
    "(x - 𝑥̄)²": hasil_pangkat,
    "ai" : hasil_koefisien,
    "X n+1-i" : descending,
    "Xi" : data_x,
    "(X n+1-i - Xi)" : hasil_pengurangan,
    "ai(X n+1-i - Xi)" : hasil_akhir
}

data_iklim = pd.DataFrame(normalitas_iklim)

data_iklim.fillna("-", inplace = True)

data_iklim
```

```python
# W_pembilang b2

W_pembilang = pow(hasil_akhir.sum(), 2)

print(f"b2 = {W_pembilang}")
```

```python
# W_penyebut SS

W_penyebut = pow(hasil_sort - 𝑥̄, 2).sum()

print(f"SS = {W_penyebut}")
```

```python
# W_kelembaban

w_kelembaban = W_pembilang / W_penyebut

w_kelembaban
```

```python
# W_penyebut SS

W_penyebut = pow(hasil_sort - 𝑥̄, 2).sum()

print(f"SS = {W_penyebut}")

# W_kelembaban

w_kelembaban = W_pembilang / W_penyebut

w_kelembaban

p_value_24 = p_value[0.05][24]

# Buat Q-Q plot
fig, ax = plt.subplots(figsize=(5, 3))
sm.qqplot(hasil_akhir, line='s', ax=ax)  # line='s' untuk menampilkan garis diagonal

# Tambahkan garis referensi distribusi normal dari hasil uji normalitas
ax.set_title(f'Q-Q Plot Normalitas kelembaban: {w_kelembaban:.4f}, p-value: {p_value_24:.4f}')

plt.show()
```

Notes :

metode Shapiro-Wilk:

*   Jika nilai statistik uji mendekati 1, ini mengindikasikan bahwa distribusi data cenderung normal.
*   Semakin besar nilainya mendekati 1, semakin mendukung hipotesis bahwa data terdistribusi normal.
*   Jika normalitas > signifikansi p-value (5% / 0.05), maka data terdistribusi normal.
*   Jika normalitas < signifikansi p-value (5% / 0.05), maka data tidak terdistribusi normal.

```python
# x
df_kelembaban = df["Kelembaban (%)"]

# y
df_diare = df["Jumlah Kasus Diare"]
```

```python
print(df_kelembaban)
```

```python
print(df_diare)
```

```python
# x^2

x_pangkat = pow(df_kelembaban, 2)

print(x_pangkat)
```

```python
# y^2

y_pangkat = pow(df_diare, 2)

print(y_pangkat)
```

```python
# xy

xy = df_kelembaban * df_diare

print(xy)
```

```python
data_korelasi = {
    "x" : df_kelembaban,
    "y" : df_diare,
    "x^2" : x_pangkat,
    "y^2" : y_pangkat,
    "xy" : xy,
}

korelasi = pd.DataFrame(data_korelasi)

korelasi
```

```python
n = len(korelasi['x'])
Exy = korelasi['xy'].sum()
ExEy = (korelasi['x'].sum() * korelasi['y'].sum()) / n
r1 = Exy - ExEy
print(r1)
```

```python
Exx = korelasi['x^2'].sum() - ((korelasi['x'].sum() ** 2) / n)
Eyy = korelasi['y^2'].sum() - ((korelasi['y'].sum() ** 2) / n)
r2 = math.sqrt(Exx * Eyy)
print(r2)
```

```python
r = r1 / r2

print(r)
```
