```python
from bs4 import BeautifulSoup

html_doc = """
<html>
<head>
<title>Contoh script html</title>
</head>
<body>
<h1 id="title"><b>Title 1</b></h1>
<h1 id="title"><b>Title 2</b></h2>
<p class="nama1">Agik</p>
<p class="nama2">Anang</p>
<p class="nama3">Wahyu</p>
</body>
</html>
"""
```

PARSE HTML

```python
soup = BeautifulSoup(html_doc, 'html')
soup
```

FIND

```python
data_find = soup.find("p")
data_find
```

```python
data_find = soup.find("p").text
data_find
```

```python
data_find = soup.find("h1", id = "title").text
data_find
```

```python
data_find = soup.find("p", class_ = "nama2").text
data_find
```

FIND ALL

```python
data_find_all = soup.find_all("p")
data_find_all
```

```python
data_find_all = soup.find_all("p")
for i in data_find_all:
    print(i.text)
```

```python
data_find_all = soup.find_all("h1", id = "title")
data_find_all
```

```python
data_find_all = soup.find_all("p", class_ = "nama2")
data_find_all
```

SELECT

```python
data_select = soup.select("h1#title")
data_select
```

```python
data_select = soup.select("#title")
data_select
```

```python
data_select = soup.select("p.nama3")
data_select
```

```python
data_select = soup.select(".nama3")
data_select
```
