Selenium

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service

chrome_driver_path = "/Applications/MAMP/htdocs/local/kodingakademi/Python/python 3/machine learning/scrapping/selenium/chromedriver-mac-arm64/chromedriver"

service = Service(chrome_driver_path)
driver = webdriver.Chrome(service=service)

driver.get("https://www.wikipedia.org/")

print("Judul Halaman:", driver.title)
```

```python
from selenium.webdriver.common.by import By

# Menemukan elemen berdasarkan CLASS
element = driver.find_element(By.CLASS_NAME, "central-textlogo")
print(element.text)
```

```python
# Menemukan elemen berdasarkan ID
element = driver.find_element(By.ID, "search-form")
print(element.text)
```

```python
# Menemukan elemen berdasarkan TAG NAME
element = driver.find_elements(By.TAG_NAME, "strong")
print(element.text)
```

```python
# Menemukan elemen berdasarkan Name dan input teks
element = driver.find_element(By.NAME, "search")
element.send_keys("Selenium Python")  # Memasukkan teks
```

```python
# Menemukan elemen berdasarkan Selector CSS
element = driver.find_element(By.CSS_SELECTOR, "footer.footer")
print(element.text)
```

```python
# Menemukan elemen berdasarkan XPath
element = driver.find_element(By.XPATH, "//p[@class='site-license']")
print(element.text)
```

Selenium & Beautiful soup

```python
from bs4 import BeautifulSoup

# Menemukan elemen berdasarkan ID
element = driver.find_element(By.CLASS_NAME, "central-textlogo")
soup = BeautifulSoup(element.get_attribute('outerHTML'), "html.parser")
print(soup)
```
