```python
import os
import sqlalchemy
import pandas as pd

# Buat koneksi ke database vior6 dan vior9
vior6 = sqlalchemy.create_engine('mysql+pymysql://root:root@localhost/vior6')
vior9 = sqlalchemy.create_engine('mysql+pymysql://root:root@localhost/vior9')

# Query untuk mendapatkan nama tabel dari kedua database
query_vior6 = "SELECT table_name FROM information_schema.tables WHERE table_schema = 'vior6';"
query_vior9 = "SELECT table_name FROM information_schema.tables WHERE table_schema = 'vior9';"

# Ambil nama tabel dari vior6 dan vior9
data_vior6 = pd.read_sql(query_vior6, vior6)
data_vior9 = pd.read_sql(query_vior9, vior9)

# Mengubah nama tabel menjadi list untuk perbandingan
list_vior6 = data_vior6['TABLE_NAME'].tolist()
list_vior9 = data_vior9['TABLE_NAME'].tolist()

# Folder untuk menyimpan CSV dan Excel
output_folder = "/Applications/MAMP/htdocs/local/kodingakademi/Python/wwwfiordb/diff_data"
os.makedirs(output_folder, exist_ok=True)

# Bandingkan tabel yang ada di vior6 dan vior9
for table in list_vior6:
    if table in list_vior9:
        # Jika tabel ada di kedua database, ambil data dari keduanya
        query_data_vior6 = f"SELECT * FROM {table};"
        query_data_vior9 = f"SELECT * FROM {table};"
        
        data_vior6 = pd.read_sql(query_data_vior6, vior6)
        data_vior9 = pd.read_sql(query_data_vior9, vior9)
        
        # Cari data yang ada di vior6 tapi tidak ada di vior9
        diff_data = pd.merge(data_vior6, data_vior9, indicator=True, how='left')
        diff_data = diff_data[diff_data['_merge'] == 'left_only'].drop(columns=['_merge'])
        
        # Jika ada data di vior6 yang tidak ada di vior9, simpan ke CSV dan kemudian ke Excel
        if not diff_data.empty:
            csv_output = os.path.join(output_folder, f"{table}_diff_data.csv")
            excel_output = os.path.join(output_folder, f"{table}_diff_data.xlsx")
            
            # Simpan ke CSV
            diff_data.to_csv(csv_output, index=False)
            print(f"Data yang berbeda dari tabel {table} telah disimpan di {csv_output}")
            
            # Simpan ke Excel
            diff_data.to_excel(excel_output, index=False)
            print(f"Data yang berbeda dari tabel {table} juga telah disimpan di {excel_output}")
    else:
        # Jika tabel hanya ada di vior6
        print(f"Tabel {table} hanya ada di vior6.")
```
