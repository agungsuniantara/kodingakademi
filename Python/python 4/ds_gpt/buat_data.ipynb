```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Settings
num_clients = 100
num_transactions = 10000

client_ids = [f'C{str(i).zfill(3)}' for i in range(1, num_clients + 1)]
data = []

for _ in range(num_transactions):
    client = random.choice(client_ids)
    txn_date = datetime(2024, 1, 1) + timedelta(days=random.randint(0, 180))
    amount = round(random.uniform(10, 1000), 2)
    txn_id = f"T{random.randint(1000,9999)}"
    data.append([client, txn_id, txn_date, amount])

df = pd.DataFrame(data, columns=['client_id', 'transaction_id', 'transaction_date', 'amount'])

# --- Add Duplicates ---
# Duplicate 5 random rows based on 3 columns: client_id, transaction_id, transaction_date
duplicates = df.sample(5, random_state=42)[['client_id', 'transaction_id', 'transaction_date']]
dupe_rows = df.merge(duplicates, on=['client_id', 'transaction_id', 'transaction_date'], how='inner')
df = pd.concat([df, dupe_rows], ignore_index=True)

# --- Add Missing Values ---
# Randomly set some entries to NaN in various columns
for col in ['client_id', 'transaction_id', 'transaction_date', 'amount']:
    idx = df.sample(frac=0.05, random_state=random.randint(0,1000)).index
    df.loc[idx, col] = np.nan

# Optional: shuffle
df = df.sample(frac=1, random_state=1).reset_index(drop=True)

df.to_csv("client_database.csv")
```
