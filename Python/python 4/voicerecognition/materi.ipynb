```python
import speech_recognition as sr

def rekam_suara():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Silakan berbicara...")
        audio = recognizer.listen(source)
        try:
            teks = recognizer.recognize_google(audio, language="id-ID")
            print("Anda mengatakan:", teks)
        except sr.UnknownValueError:
            print("Maaf, saya tidak memahami apa yang Anda ucapkan.")
        except sr.RequestError:
            print("Permintaan gagal, periksa koneksi internet Anda.")

rekam_suara()
```
