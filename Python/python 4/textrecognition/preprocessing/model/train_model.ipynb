```python
import pickle
import spacy
import random
from spacy.util import minibatch, compounding
from spacy.training import Example

# Memuat dataset
with open('ner_spacy_fmt_datasets.pickle', 'rb') as f:
    ner_spacy_fmt_datasets = pickle.load(f)

# Inisialisasi model kosong untuk bahasa Indonesia
nlp = spacy.blank("id")

# Menambahkan pipeline NER
if "ner" not in nlp.pipe_names:
    ner = nlp.add_pipe("ner")
else:
    ner = nlp.get_pipe("ner")

# Menambahkan label entitas ke model
for _, annotations in ner_spacy_fmt_datasets:
    for ent in annotations.get("entities"):
        ner.add_label(ent[2])

# Mulai pelatihan
nlp.begin_training()

# Pengecualian pipeline selain NER
pipe_exceptions = ["ner", "trf_wordpiecer", "trf_tok2vec"]
unaffected_pipes = [pipe for pipe in nlp.pipe_names if pipe not in pipe_exceptions]

# Melatih model NER
with nlp.disable_pipes(*unaffected_pipes):
    for iteration in range(50):
        print(f"Iterasi ke-{iteration + 1}")
        random.shuffle(ner_spacy_fmt_datasets)  # Acak dataset di setiap epoch
        losses = {}

        # Membuat batch dengan fungsi minibatch
        batches = minibatch(ner_spacy_fmt_datasets, size=compounding(4.0, 32.0, 1.001))
        
        for batch in batches:
            texts, annotations = zip(*batch)
            
            # Membuat objek Example dari teks dan anotasi
            examples = [Example.from_dict(nlp.make_doc(text), annotation) for text, annotation in zip(texts, annotations)]
            
            # Melatih model menggunakan Example
            nlp.update(
                examples,     # Batch dalam bentuk Example
                drop=0.5,     # Dropout untuk mencegah overfitting
                losses=losses # Menyimpan nilai kerugian
            )

        print(f"Losses: {losses}")

# Simpan model yang telah dilatih
nlp.to_disk("ner_model_id")
print("Model berhasil disimpan!")
```
