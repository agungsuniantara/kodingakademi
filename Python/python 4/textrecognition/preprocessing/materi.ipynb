TOKENISASI 

Tokenisasi adalah proses memecah teks menjadi unit yang lebih kecil (token), seperti kata atau kalimat.

```python
from nltk.tokenize import word_tokenize

teks = "Saya belajar NLP."
tokens = word_tokenize(teks)
print(tokens)
```

STOPWORD REMOVAL

Stopwords adalah kata-kata umum yang tidak memiliki banyak makna dalam pemrosesan teks.

```python
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('indonesian'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

STEMMING AND LEMMATIZATION

- Stemming: Menghapus akhiran untuk mendapatkan bentuk dasar kata.
- Lemmatization: Mengubah kata menjadi bentuk lemat (bentuk dasar yang diakui secara linguistik).

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Membuat stemmer
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Proses stemming
text = "Saya berlari ke sekolah dengan berjalan kaki."
stemmed_text = stemmer.stem(text)
print(stemmed_text)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import string
import re

nltk.download('punkt')
nltk.download('stopwords')

factory = StemmerFactory()
stemmer = factory.create_stemmer()
stop_words = set(stopwords.words('indonesian'))

teks = "AKU ADALAH ANAK GEMBALA YANG SELALU RIANG 20! ^&&&"

# Case folding
teks = teks.lower()  

# Hapus karakter non-huruf (termasuk angka)
teks = re.sub(r'[^a-zA-Z\s]', '', teks)

# Tokenization
tokens = word_tokenize(teks)

# Hapus tanda baca
tokens = [kata for kata in tokens if kata not in string.punctuation]

# Stopword removal
tokens = [kata for kata in tokens if kata not in stop_words]

# Stemming
tokens = [stemmer.stem(kata) for kata in tokens]

print(tokens)
```

```python
# Inisialisasi Stemmer Sastrawi
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Stopwords Bahasa Indonesia
stop_words = set(stopwords.words('indonesian'))

# Kamus Normalisasi (contoh, bisa diperluas)
kamus_normalisasi = {
    'gapapa' : 'tidak apa apa',
    'gpp': 'tidak apa apa',
    'bgt': 'banget',
    'yg': 'yang',
    'aku': 'saya',
    'kamu': 'anda',
    'dgn': 'dengan',
    'tdk': 'tidak',
    'sdh': 'sudah',
    'blm': 'belum',
}

# Teks yang akan diolah
teks = "AKU ADALAH ANAK GEMBALA YG SELALU RIANG 20, GAPAPA BGT!"

# 1. Case Folding (Mengubah ke huruf kecil)
teks = teks.lower()

# 2. Cleaning (Menghapus karakter non-alphabet & number)
teks = re.sub(r'[^a-zA-Z\s]', '', teks)

# 3. Tokenization (Memecah menjadi kata-kata)
tokens = word_tokenize(teks)

# 4. Normalization (Mengganti kata tidak baku)
tokens = [kamus_normalisasi.get(kata, kata) for kata in tokens]

# 5. Menghapus Tanda Baca (sudah dihapus oleh regex, ini hanya untuk tambahan)
tokens = [kata for kata in tokens if kata not in string.punctuation]

# 6. Stopword Removal (Menghapus kata umum)
tokens = [kata for kata in tokens if kata not in stop_words]

# 7. Stemming (Mengembalikan ke bentuk dasar menggunakan Sastrawi)
tokens_stemmed = [stemmer.stem(kata) for kata in tokens]

print("Teks Asli           :", teks)
print("Token Setelah Stemming :", tokens_stemmed)
```
