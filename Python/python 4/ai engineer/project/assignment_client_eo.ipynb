```python
import pandas as pd
import numpy as np

np.random.seed(42)

def buat_segmen(usia_a, usia_b, jarak_a, jarak_b, budget_a, budget_b,
                tamu_a, tamu_b, masakan, gaya, paket, tambah_prob, deal_prob, n):
    df = pd.DataFrame({
        'usia': np.random.randint(usia_a, usia_b, n),
        'jarak_km': np.random.randint(jarak_a, jarak_b, n),
        'budget_juta': np.random.randint(budget_a, budget_b, n),
        'jumlah_tamu': np.random.randint(tamu_a, tamu_b, n),
        'jenis_masakan': np.random.choice(masakan, n),
        'gaya_resepsi': np.random.choice(gaya, n),
        'level_paket': np.random.choice(paket, n),
        'tambah_layanan': np.random.choice([0, 1], n, p=[1 - tambah_prob, tambah_prob]),
    })
    df['deal_jadi'] = np.random.choice([0, 1], n, p=[1 - deal_prob, deal_prob])
    return df

# Segmen 1: Klien Premium, budget besar, tamu banyak
s1 = buat_segmen(28, 40, 3, 15, 180, 300, 1000, 2500,
                 ['Internasional', 'Modern'], ['Plating', 'Prasmanan'],
                 ['Platinum', 'Gold'], 0.8, 0.90, 250)

# Segmen 2: Klien Hemat, budget kecil, tamu sedikit
s2 = buat_segmen(22, 32, 3, 25, 30, 100, 200, 600,
                 ['Jawa', 'Sunda'], ['Prasmanan'],
                 ['Basic', 'Silver'], 0.2, 0.35, 250)

# Segmen 3: Klien Adat Tradisional, usia lebih tua, buffet adat
s3 = buat_segmen(38, 50, 5, 35, 80, 180, 500, 1200,
                 ['Jawa', 'Sunda'], ['Buffet_Adat'],
                 ['Silver', 'Gold'], 0.8, 0.75, 250)

# Segmen 4: Klien Modern Urban, jarak jauh, masakan modern
s4 = buat_segmen(24, 36, 15, 50, 100, 200, 500, 1200,
                 ['Modern'], ['Food_Station', 'Plating'],
                 ['Gold', 'Silver'], 0.6, 0.60, 250)

df = pd.concat([s1, s2, s3, s4], ignore_index=True)
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
df.to_csv('client_katering.csv', index=False)

print(f"Dataset dibuat: {len(df)} baris, {len(df.columns)} kolom")
print(df.head())
```
