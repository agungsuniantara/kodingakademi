```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 1500

jumlah_jam = np.random.randint(4, 13, n)
jumlah_foto = np.random.randint(100, 1001, n)
jenis_paket = np.random.choice(
    ['Basic', 'Standard', 'Premium', 'Exclusive'], n,
    p=[0.25, 0.35, 0.25, 0.15]
)
jumlah_videografer = np.random.randint(0, 4, n)
termasuk_album = np.random.choice([0, 1], n, p=[0.6, 0.4])
termasuk_drone = np.random.choice([0, 1], n, p=[0.7, 0.3])
lokasi_venue = np.random.choice(
    ['Satu_Lokasi', 'Berbeda_Lokasi', 'Luar_Kota'], n,
    p=[0.4, 0.35, 0.25]
)

# Logika biaya dokumentasi
biaya_dokumentasi = []
for i in range(n):
    biaya = 0
    # Biaya dasar paket
    if jenis_paket[i] == 'Basic':
        biaya += 3_000_000
    elif jenis_paket[i] == 'Standard':
        biaya += 5_000_000
    elif jenis_paket[i] == 'Premium':
        biaya += 10_000_000
    else:  # Exclusive
        biaya += 18_000_000

    # Biaya per jam (overtime)
    biaya += 500_000 * max(0, jumlah_jam[i] - 8)

    # Biaya per foto tambahan (>500)
    biaya += 15_000 * max(0, jumlah_foto[i] - 500)

    # Videografer
    biaya += 3_000_000 * jumlah_videografer[i]

    # Album
    if termasuk_album[i] == 1:
        biaya += 1_500_000

    # Drone
    if termasuk_drone[i] == 1:
        biaya += 2_500_000

    # Lokasi
    if lokasi_venue[i] == 'Berbeda_Lokasi':
        biaya += 2_000_000
    elif lokasi_venue[i] == 'Luar_Kota':
        biaya += 5_000_000

    # Noise
    biaya *= np.random.normal(1, 0.04)

    biaya_dokumentasi.append(int(biaya))

df = pd.DataFrame({
    'jumlah_jam': jumlah_jam,
    'jumlah_foto': jumlah_foto,
    'jenis_paket': jenis_paket,
    'jumlah_videografer': jumlah_videografer,
    'termasuk_album': termasuk_album,
    'termasuk_drone': termasuk_drone,
    'lokasi_venue': lokasi_venue,
    'biaya_dokumentasi': biaya_dokumentasi
})

df.to_csv('estimasi_biaya_dokumentasi.csv', index=False)
print(f"Dataset: {len(df)} baris, {len(df.columns)} kolom")
print(f"Rentang biaya: Rp{df['biaya_dokumentasi'].min():,} - Rp{df['biaya_dokumentasi'].max():,}")
print(df.head())
```
