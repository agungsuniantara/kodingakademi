Data Tabular

```python
import pandas as pd
from autogluon.tabular import TabularPredictor

data_tabular = pd.DataFrame({
    'pendapatan_bulanan': [5000, 7000, 4500, 8000, 12000, 4000, 6500, 9000, 5500, 11000, 7200, 8100],
    'skor_kredit': [600, 750, 580, 800, 850, 550, 700, 820, 620, 830, 710, 790],
    'pekerjaan': ['Kontrak', 'Tetap', 'Kontrak', 'Tetap', 'Tetap', 
                  'Kontrak', 'Tetap', 'Tetap', 'Kontrak', 'Tetap', 
                  'Kontrak', 'Tetap'],
    'status_pinjaman': ['Ditolak', 'Disetujui', 'Ditolak', 'Disetujui', 
                        'Disetujui', 'Ditolak', 'Disetujui', 'Disetujui', 
                        'Ditolak', 'Disetujui', 'Ditolak', 'Disetujui'] # Target
})

data_tabular.head()
```

```python
from sklearn.metrics import mean_absolute_percentage_error

nama_kolom_target = 'status_pinjaman'

# Pemisahan Data: 9 baris untuk pelatihan, 3 baris untuk pengujian
train_data = data_tabular.iloc[:-3].copy()
test_data = data_tabular.iloc[-3:].copy()

print("--- Memulai Pelatihan Fase Evaluasi ---")
predictor_eval = TabularPredictor(
    label=nama_kolom_target, 
    eval_metric='accuracy'
).fit(
    train_data=train_data, 
    time_limit=60, 
    presets='medium_quality'
)

hasil_benchmark = predictor_eval.leaderboard(test_data, silent=True)
print("\n--- TABEL BENCHMARKING (EVALUASI SKOR MODEL) ---")
print(hasil_benchmark[['model', 'score_test', 'score_val']])
```

```python
# Pelatihan ulang menggunakan SELURUH data (12 baris) agar model memiliki wawasan maksimal
print("\n--- Memulai Pelatihan Penuh (Full Data) ---")
predictor_final = TabularPredictor(
    label=nama_kolom_target, 
    eval_metric='accuracy'
).fit(
    train_data=data_tabular, # Menggunakan DataFrame utuh
    time_limit=60, 
    presets='medium_quality'
)

# Pembuatan Data Baru (Data masa depan yang belum memiliki status pinjaman)
data_baru = pd.DataFrame({
    'pendapatan_bulanan': [15000, 3000],
    'skor_kredit': [880, 500],
    'pekerjaan': ['Tetap', 'Kontrak']
})

# Prediksi menggunakan model final
prediksi_final = predictor_final.predict(data_baru)
data_baru['prediksi_status'] = prediksi_final

print("\n--- HASIL PREDIKSI DATA BARU ---")
print(data_baru)
```

Data Time Series

```python
import numpy as np
from autogluon.timeseries import TimeSeriesDataFrame, TimeSeriesPredictor
import warnings
warnings.filterwarnings('ignore')

periode_hari = 30
df_mentah = pd.DataFrame({
    'id_toko': ['Cabang_A'] * periode_hari,
    'tanggal': pd.date_range(start='2026-01-01', periods=periode_hari, freq='D'),
    'penjualan': np.random.randint(100, 500, size=periode_hari)
})

df_mentah.head()
```

```python
ts_full = TimeSeriesDataFrame.from_data_frame(
    df_mentah,
    id_column="id_toko",
    timestamp_column="tanggal"
)
ts_full.head()
```

```python
pred_length = 5 

# Memotong 5 hari terakhir untuk keperluan Evaluasi (Train: 25 hari pertama)
ts_train = ts_full.slice_by_timestep(None, -pred_length)
ts_test = ts_full # AutoGluon secara otomatis akan menguji performa pada bagian akhir data ini

print("--- Memulai Pelatihan Time Series (Fase Evaluasi) ---")
ts_predictor_eval = TimeSeriesPredictor(
    prediction_length=pred_length,
    target="penjualan",
    eval_metric="MAE"
).fit(
    ts_train,
    presets="fast_training",
    time_limit=60
)

# Mendapatkan skor akurasi model
ts_benchmark = ts_predictor_eval.leaderboard(ts_test, silent=True)
print("\n--- TABEL BENCHMARKING TIME SERIES ---")
print(ts_benchmark[['model', 'score_test', 'score_val']])
```

```python
# Pelatihan diulang menggunakan 100% data historis (30 hari utuh)
print("\n--- Memulai Pelatihan Time Series (Full Data) ---")
ts_predictor_final = TimeSeriesPredictor(
    prediction_length=pred_length,
    target="penjualan",
    eval_metric="MAE"
).fit(
    ts_full, # Menggunakan data historis lengkap
    presets="fast_training",
    time_limit=60
)

# Melakukan peramalan (Forecasting) murni 5 hari ke masa depan
hasil_forecast = ts_predictor_final.predict(ts_full)

print("\n--- HASIL FORECASTING MASA DEPAN (5 Hari Kedepan) ---")
print(hasil_forecast[['mean', '0.1', '0.9']])
```
