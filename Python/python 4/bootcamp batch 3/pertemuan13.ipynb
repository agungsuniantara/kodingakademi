```python
import pandas as pd

data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'Alice', 'Eve'],
    'Total': [100, 150, 100, 250, 300],
    "City": ['New York', 'Los Angeles', 'New York', 'Chicago', 'Houston']
})

print(data)
print(data.duplicated()) # cek duplikat dataframe
print(data.duplicated(subset=['Name'])) # cek duplikat hanya kolom name
print(data["Name"].duplicated()) # cek duplikat hanya kolom name
```

```python
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'Alice', 'Eve'],
    'Total': [100, 150, 100, 250, 300],
    "City": ['New York', 'Los Angeles', 'New York', 'Chicago', 'Houston']
})

print(data)
print(data.drop_duplicates()) # hapus duplikat dataframe
print(data.drop_duplicates(subset=['Name'])) # hapus duplikat hanya kolom name
print(data.drop_duplicates(subset=['Name'], keep='last')) # hapus duplikat hanya kolom name, keep last
print(data.drop_duplicates(subset=['Name'], keep=False)) # hapus duplikat hanya kolom name, keep false (hapus semua duplikat)
```

```python
# combining data 

data1 = pd.read_csv('user_age_2017.csv')
data2 = pd.read_csv('user_age_2018.csv')

print(data1)
print(data2)

gabung = pd.concat([data1, data2], ignore_index=True) # gabungkan data dengan concat, ignore index untuk reset index
print(gabung)
```

```python
print(gabung)
print(gabung.info())
gabung["date"] = pd.to_datetime(gabung["date"])
print(gabung)
print(gabung.info())
```

```python
import datetime as dt

waktu = dt.datetime.now() # cek waktu sekarang

gabung["age"] = gabung["date"].map(lambda x: waktu.year - x.year) # hitung umur dengan apply dan lambda
print(gabung)
```

```python
# binning data 

gabung["age_group"] = pd.cut(gabung["age"], 
                             bins=[0, 18, 35, 60, 100], 
                             labels=["Child", "Youth", "Adult", "Senior"])
print(gabung)
```

```python
# matplotlib 

import matplotlib.pyplot as plt

plt.figure(figsize=(5,3))
plt.plot([1,2,3,4], [10,5,30,20], label = "Plot 1") # (x, y)
plt.plot([1,2,3,4], [20,10,40,30], label = "Plot 2") # (x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple Plot")
plt.xticks([1,2,3,4]) # ubah nilai x-axis
plt.yticks([10,20,30,40]) # ubah nilai y-axis
plt.legend(loc = "lower right") # ubah posisi legend
plt.show()
```

```python
data = pd.read_csv("jantung.csv")[["age", "sex", "trestbps", "chol"]]
age = data.groupby("age")["age"].count()
print(age.head())
```

```python
# line plot 

plt.figure(figsize=(5,3))
plt.plot(age.index, age.values, color = "red", marker='o', ls = '--') # (x, y)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()

# cara cepet pandas

plt.figure(figsize=(5,3))
age.plot(kind='line', color = "red", marker='o', ls = '--')
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
# bar plot 

plt.figure(figsize=(5,3))
plt.bar(age.index, age.values, color = "red") # (x, y)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
# scatter plot 

plt.figure(figsize=(5,3))
plt.scatter(age.index, age.values, color = "red") # (x, y)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
# histogram

plt.hist(age.values, bins=10) # (y, bins)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
# pie plot

gender = data.groupby("sex")["sex"].count()
print(gender)

plt.pie(gender.values, labels=gender.index, autopct='%1.1f%%') # (y, labels, autopct untuk menampilkan persentase)
plt.title("Gender Distribution")
plt.show()
```

```python
# subplot (1 baris 2 kolom)

plt.figure(figsize=(10,5))
plt.subplot(121) # (baris, kolom, letak plot)
plt.plot(age.index, age.values, color = "red", marker='o', ls = '--') # (x, y)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")

plt.subplot(122) # (baris, kolom, letak plot)
plt.bar(gender.index, gender.values, color = "blue") # (x, y)
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Gender Distribution")

plt.tight_layout() # agar tidak overlap
plt.show()
```
