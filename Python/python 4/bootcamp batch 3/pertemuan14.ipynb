```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt

csv = pd.read_csv("jantung.csv")
source = csv[["sex", "age", "cp", "trestbps", "chol"]]
source.head()
```

```python
# histplot

sns.histplot(x = "trestbps", data = source, bins = 10)
plt.title("Distribution of Resting Blood Pressure")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Frequency")
```

```python
# kdeplot

sns.kdeplot(x = "trestbps", data = source, fill = True)
plt.title("Density of Resting Blood Pressure")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Density")
```

```python
# scatterplot

sns.scatterplot(x = "age", y = "trestbps", data = source)
plt.title("Age vs Resting Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Resting Blood Pressure (mm Hg)")
```

```python
# pairplot

test = source[["age", "trestbps", "chol"]]
sns.pairplot(test)
plt.show()
```

```python
# barplot

plt.figure(figsize=(20,5))
sns.barplot(x = "trestbps", y = "chol", data = source, errorbar = None)
plt.title("Average Cholesterol by Resting Blood Pressure")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Cholesterol (mg/dl)")
```

```python
group = source.groupby("sex")["sex"].count().reset_index(name = "count")
print(group)

sns.barplot(x = "sex", y = "count", data = group, errorbar = None)
```

```python
# countplot

sns.countplot(x = "cp", data = source)
plt.title("Distribution of Chest Pain Types")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")
```

```python
# boxplot

sns.boxplot(x = "cp", y = "trestbps", data = source)
plt.title("Resting Blood Pressure by Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Resting Blood Pressure (mm Hg)")
```

```python
# strip plot

sns.stripplot(x = "cp", y = "trestbps", data = source)
plt.title("Resting Blood Pressure by Chest Pain Type")
plt.xlabel("Chest Pain Type")
plt.ylabel("Resting Blood Pressure (mm Hg)")
```

```python
# pivot table

pivot = source.pivot_table(index = "cp", columns = "age", values = "chol", aggfunc="mean")
pivot = pivot.fillna(0)
pivot
```

```python
plt.figure(figsize=(20,5))
sns.heatmap(pivot, cmap = "YlGnBu")
plt.title("Cholesterol Levels by Chest Pain Type and Age")
plt.xlabel("Age")
plt.ylabel("Chest Pain Type")
```

```python
# hue seaborn

sns.barplot(x = "cp", y = "trestbps", hue = "sex", data = source, errorbar = None)
plt.title("Resting Blood Pressure by Chest Pain Type and Sex")
plt.xlabel("Chest Pain Type")
plt.ylabel("Resting Blood Pressure (mm Hg)")
```

```python
sns.lineplot(x = "age", y = "trestbps", hue = "sex", data = source, errorbar=None)
plt.title("Resting Blood Pressure by Age and Sex")
plt.xlabel("Age")
plt.ylabel("Resting Blood Pressure (mm Hg)")
```

```python
# seaborn subplot

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.histplot(x = "trestbps", data = source, bins = 10)
plt.title("Distribution of Resting Blood Pressure")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
sns.kdeplot(x = "trestbps", data = source, fill = True)
plt.title("Density of Resting Blood Pressure")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```
