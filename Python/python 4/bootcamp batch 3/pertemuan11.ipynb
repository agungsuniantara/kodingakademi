```python
# hapus dataframe (baris)

data_df = pd.DataFrame({
    "populasi" : [35, 63, 80, 60],
    "GDP" : [100000, 200000, 300000, 400000],
    "Continent" : ["America", "Europe", "Asia", "Europe"]
})

print(data_df)
# data_df = data_df.drop(2) menghapus 1 data
data_df = data_df.drop([2,3]) # menghapus lebih dari 1 data
print(data_df)
```

```python
# hapus dataframe (kolom)

data_df = pd.DataFrame({
    "populasi" : [35, 63, 80, 60],
    "GDP" : [100000, 200000, 300000, 400000],
    "Continent" : ["America", "Europe", "Asia", "Europe"]
})

data_df = data_df.drop(columns = ["populasi"])
print(data_df)
data_df = data_df.drop(1, axis = 0) # axis = 0 (baris)
print(data_df)
data_df = data_df.drop(["Continent"], axis = 1) # axis = 1 (kolom)
print(data_df)
```

```python
# broadcasting

data_df = pd.DataFrame({
    "populasi" : [35, 63, 80, 60],
    "GDP" : [100000, 200000, 300000, 400000],
    "Continent" : ["America", "Europe", "Asia", "Europe"]
})

baru = pd.DataFrame({
    "populasi" : [10, 20, 30, 40]
})

print(data_df)
data_df["populasi"] = data_df["populasi"] + 50
print(data_df)
data_df["populasi"] = data_df["populasi"] + baru["populasi"]
print(data_df)
data_df[["populasi", "GDP"]] = data_df[["populasi", "GDP"]] + 1000
print(data_df)
```

```python
# tambah kolom

data_df = pd.DataFrame({
    "populasi" : [35, 63, 80, 60],
    "GDP" : [100000, 200000, 300000, 400000],
    "Continent" : ["America", "Europe", "Asia", "Europe"]
})

data_df["harga"] = [10000, 20000, 30000, 40000]
print(data_df)
data_df["total"] = data_df["populasi"] * data_df["GDP"]
print(data_df)
```
