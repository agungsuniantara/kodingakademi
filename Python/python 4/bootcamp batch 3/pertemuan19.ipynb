```python
import requests 

url = "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue"
header = {
    "User-Agent": "testing"
}
data = requests.get(url, headers=header)
print(data.text)
```

```python
scrap = BeautifulSoup(data.text, "html")
print(scrap.prettify())
```

```python
table = scrap.find("table", class_ = "wikitable sortable sticky-header-multi sort-under")
title = table.find("tr") 
title_th = title.find_all("th")
tmp_title = [i.text.strip() for i in title_th]
tmp_title.remove("State-owned")
tmp_title.remove("Ref.")
title_finish = [i.replace("[note 1]", "") for i in tmp_title]
print(title_finish)
```

```python
import pandas as pd 

data = pd.DataFrame(columns = title_finish)
tr_row = table.find_all("tr")[2:]

for i in tr_row:
    rank = i.find_all("th")
    row = rank + i.find_all("td")
    row_data = [z.text.strip() for z in row if z.text.strip() != ""]
    print(row_data)
```

```python
import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue"
headers = {"User-Agent": "testing"}

response = requests.get(url, headers=headers)
tabel_web = pd.read_html(response.text)
df = tabel_web[0] 
df.columns = df.columns.droplevel(1)
df.columns = [col.replace("[note 1]", "").strip() for col in df.columns]
df = df.drop(columns=['State-owned', 'Ref.'])
df
```

```python
import pandas as pd
import requests
from io import StringIO

url = "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue"
headers = {"User-Agent": "MyScraper/1.0 (your_email@example.com)"}

response = requests.get(url, headers=headers)
html_data = StringIO(response.text)
tabel_web = pd.read_html(html_data)
df = tabel_web[0]
df.columns = df.columns.droplevel(1)
df.columns = [col.replace("[note 1]", "").strip() for col in df.columns]
df = df.drop(columns=['State-owned', 'Ref.'])
df
```
