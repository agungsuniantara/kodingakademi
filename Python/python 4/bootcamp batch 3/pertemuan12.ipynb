```python
import pandas as pd 

data = pd.read_csv("bali_tourist.csv")
backup = data.copy()
data
```

```python
data_2018 = data.loc[data["year"] == 2018]
data_2019 = data.loc[data["year"] == 2019]
```

```python
# total

total_2018 = data_2018.loc[0:, "January":"December"].sum().sum()
total_2019 = data_2019.loc[0:, "January":"December"].sum().sum()

print(total_2018)
print(total_2019)
```

```python
# rata - rata 

rata_2018 = data_2018.loc[0:, "January":"December"].sum().mean()
rata_2019 = data_2019.loc[0:, "January":"December"].sum().mean()

print(rata_2018)
print(rata_2019)
```

```python
# terbanyak & tersedikit

terbanyak_2018 = data_2018.loc[0:, "January":"December"].sum().idxmax()
terbanyak_2019 = data_2019.loc[0:, "January":"December"].sum().idxmax()

tersedikit_2018 = data_2018.loc[0:, "January":"December"].sum().idxmin()
tersedikit_2019 = data_2019.loc[0:, "January":"December"].sum().idxmin()

print(terbanyak_2018)
print(terbanyak_2019)
print(tersedikit_2018)
print(tersedikit_2019)
```

```python
# area mana yang paling banyak dikunjungi

data_area2018 = data_2018.drop(columns=["year", "country"])
group = data_area2018.groupby("area").sum() 
total_area2018 = group.sum(axis=1)
total_area2018.idxmax()
```

```python
data_area2019 = data_2019.drop(columns=["year", "country"])
group = data_area2019.groupby("area").sum()
total_area2019 = group.sum(axis=1)
total_area2019.idxmax()
```

```python
# buat dataframe baru 

new = data.drop(columns=["area", "country"])
group = new.groupby("year").sum()
group["max"] = group.idxmax(axis=1)
group["min"] = group.drop(columns=["max"]).idxmin(axis=1)
group["total"] = group.drop(columns=["max", "min"]).sum(axis=1)
group["average"] = group.drop(columns=["max", "min", "total"]).mean(axis=1)
group.to_excel("hasil.xlsx", sheet_name="hasil")
```

```python
data = pd.DataFrame({
    "name": ["John", "Jane", "Doe", "Smith"],
    "age": [25, 30, None, 40],
    "city": ["New York", "Los Angeles", "Chicago", None]
})

# data = data.fillna(1) # ganti seluruh data yang kosong dengan 1
# data.fillna(1, inplace=True) inplace = True (langsung simpan ke data asli)
data["age"] = data["age"].fillna(0) # ganti data yang kosong pada kolom age dengan 0
data["city"] = data["city"].fillna("-") # ganti data yang kosong pada kolom city dengan "-"
print(data)
```
