```python
import pandas as pd

driver.get("https://en.wikipedia.org/wiki/List_of_countries_by_population_(United_Nations)")

table = driver.find_element(By.CLASS_NAME, "wikitable")
rows = table.find_elements(By.TAG_NAME, "tr")

data = []
for row in rows:
    cols = row.find_elements(By.TAG_NAME, "td")
    if cols:
        country = cols[0].text
        population = cols[1].text
        data.append([country, population])

df = pd.DataFrame(data, columns=["Country", "Population"])
df.head()
```

```python
import pandas as pd

driver.get("https://finance.yahoo.com/quote/AAPL/history/?p=AAPL")

table = driver.find_element(By.CLASS_NAME, "table-container")
rows = table.find_elements(By.TAG_NAME, "tr")

data = []
for row in rows:
    cols = row.find_elements(By.TAG_NAME, "td")
    if cols and len(cols) == 7:
        date = cols[0].text
        open = cols[1].text
        high = cols[2].text
        low = cols[3].text
        close = cols[4].text
        adj_close = cols[5].text
        volume = cols[6].text
        data.append([date, open, high, low, close, adj_close, volume])

df = pd.DataFrame(data, columns=["date", "open", "high", "low", "close", "adj_close", "volume"])
df.to_excel("AAPL_stock_data.xlsx", index=False)
```

```python
import numpy as np 
import pandas as pd 
from scipy import stats 
```

```python
data = pd.read_csv("jantung.csv")
thalach = data["thalach"]
thalach
```

```python
mean = np.mean(thalach)
median = np.median(thalach)
mode = stats.mode(thalach)
variance = np.var(thalach)
std = np.std(thalach)
skewness = stats.skew(thalach)

print("Mean:", mean)
print("Median:", median)
print(f"nilai = {mode[0]}, jumlah = {mode[1]}")
print("Variance:", variance)
print("Standard Deviation:", std)
print("Skewness:", skewness)

# rata - rata = 180
# batas atas = 180 + 22 = 202
# batas bawah = 180 - 22 = 158
```
