```python
import pandas as pd

data_series = pd.Series(["Januari",
                         "Februari",
                         "Maret",
                         "April"], index = [1, 2, 3, 4])
print(data_series)

# loc

print(data_series.loc[2:3])

# iloc

print(data_series.iloc[1:3])
```

```python
print(data_series.values)
print(data_series.index)
```

```python
# conditional & operation series

data1 = pd.Series([1000, 2000, 3000, 4000, 5000], index = ["a", "b", "c", "d", "e"])

print(data1.loc["a"])
print(data1 > 1000)
print(data1.loc[data1 > 1000])
print(data1.loc[(data1 > 1000) & (data1 < 5000)]) # operator AND (&), OR (||), NEGASI (!)

# fungsi math

print(f"rata - rata = {data1.mean()}")
print(f"jumlah = {data1.sum()}")

# conditional + operasi math

print(data1.loc[data1 > data1.mean()])
```

```python
data1 = data1 + 10
print(data1)

data2 = pd.Series([1000, 2000, 3000, 4000, 5000], index = ["a", "b", "c", "d", "e"])
data1 = data1 + data2 
print(data1)
```

```python
# konversi data 

print(data1)
data1 = data1.astype(float)
print(data1)
```

```python
# edit data

print(data1)
data1.loc["a"] = 300
print(data1)
data1.loc["d":"e"] = [500, 600]
print(data1)

data1 = data1.astype(float)
data1.loc[data1 > data1.mean()] = data1.mean()
print(data1)
```

```python
# dataframe

data1 = [
    [10,20],
    [30,40]
]

data2 = {
    "nama" : ["agung","sandhy"],
    "alamat" : ["denpasar", "ubud"]
}

df_list = pd.DataFrame(data1)
df_dict = pd.DataFrame(data2)

print(df_list)
print(df_dict)

# series
print(df_dict["nama"])
```

```python
data1 = [
    [10,20],
    [30,40]
]

df_list = pd.DataFrame(data1, columns=["angka1", "angka2"], index = ["a", "b"])

print(df_list)

df_list.columns = ["angka_1", "angka_2"]
df_list.index = [1, 2]

print(df_list)
```

```python
dataset = pd.DataFrame({
    "Populasi" : [35.6, 40.5, 80.9],
    "GDP" : [1000, 1500, 2000],
    "Surface Area" : [10, 20, 30]
}, index = ["Canada", "France", "Germany"])

print(dataset)
print(dataset["Populasi"]) # hanya memilih 1 kolom dan menjadi data series, tidak perlu pilih barus.
print(dataset[["Populasi", "GDP"]]) # memilih lebih dari 1 kolom, data tetap dataframe, tidak perlu pilih baris.
print(dataset.loc["Canada": , "Populasi" : "GDP"]) # [baris, kolom], loc bisa atur baris dan kolom.
print(dataset.iloc[0:, 0:2]) # [baris, kolom], iloc bisa atur baris dan kolom.
```

```python
print(dataset)
dataset.loc["Canada", "Populasi" : "GDP"] = [40.9, 9000]
print(dataset)
```
