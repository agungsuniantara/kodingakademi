```python
# ==============================================================================
# PENGATURAN TELEGRAM (SILAKAN ISI BAGIAN INI)
# ==============================================================================
TOKEN_BOT = "MASUKKAN_TOKEN_DARI_BOTFATHER_DISINI"
CHAT_ID = "MASUKKAN_CHAT_ID_DISINI" 
NAMA_FILE_GAMBAR = "prediksi.png"
CAPTION = "*Prediksi Sales*"
# ==============================================================================

def kirim_ke_telegram(token, chat_id, file_path, caption):
    url = f"https://api.telegram.org/bot{token}/sendPhoto"
    print("\n--- Mengirim Gambar ke Telegram ---")
    try:
        with open(file_path, 'rb') as photo:
            response = requests.post(
                url, 
                files={'photo': photo}, 
                data={'chat_id': chat_id, 'caption': caption, 'parse_mode': 'Markdown'}
            )
            if response.status_code == 200:
                print(f"[SUKSES] Gambar berhasil dikirim ke Telegram!")
            else:
                print(f"[GAGAL] Respon Telegram: {response.text}")
    except Exception as e:
        print(f"[ERROR] Gagal mengirim: {e}")

if __name__ == "__main__":
    import requests
    import os

    if os.path.exists(NAMA_FILE_GAMBAR):
        kirim_ke_telegram(TOKEN_BOT, CHAT_ID, NAMA_FILE_GAMBAR, CAPTION)
    else:
        print(f"[ERROR] File gambar '{NAMA_FILE_GAMBAR}' tidak ditemukan")
```
