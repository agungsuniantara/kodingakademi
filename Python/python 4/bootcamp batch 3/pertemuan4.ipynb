```python
# collection (variabel yang dapat menampung banyak value)

# list = datanya itu dapat di ubah, akses, hapus
# tuple = datanya dapat di akses dan hapus, tidak bisa di ubah.
# set = datanya "acak", tidak bisa di akses menggunakan index
# dictionary = data yang memiliki key & value
```

```python
# list 

data = ["sapi", "ayam", "kambing"] # [0, 1, 2] = index
data_acak = [10, "agung", True, 20.4] # [0, 1, 2, 3] = index

# akses list (index / looping)

print(data[0])
print(data_acak[0])

print("\n")

for i in range(len(data)):
    if i != 0:
        print(data[i])

print("\n")

for i in data:
    print(i)
        
```

```python
# slicing

# forward index = ["sapi", "ayam", "kambing"] , index = [0,1,2]
# backward index = ["sapi", "ayam", "kambing"] , index = [-3, -2, -1]

# -3, -2, -1, 0, 1, 2, 3

# slicing positive index / forward index
print(data)
print(data[0 : 2]) # [start : stop]
print(data[1 : 3]) # [start : stop]
print(data[::2]) # [start : stop : step]
# start kosong (default = 0), stop kosong (default = paling akhir), step kosong (default = 1)

print("\n")

# slicing negative index / backward index
print(data)
print(data[-3 : -1]) # [start : stop]
print(data[::-1]) # [start : stop : step]
print(data[-2:-4:-1]) # [start : stop : step]
```

```python
# mengolah list (tambah, edit, hapus)

# edit
data = ["merah", "kuning", "putih"]
data[0] = "hijau"
print(data[0])
```

```python
# tambah (append, insert)

data = ["merah", "kuning", "putih"]

# append (menambah data diakhir list)
data.append("biru")
print(data)

# insert (menambah data sesuai index)
data.insert(1, "hijau")
print(data)
```

```python
# hapus (pop, del, clear)

data = ["merah", "kuning", "putih", "hijau"]

# del

del data[0] # del harus deklarasi index
print(data)

# pop

# data.pop() - pop tidak pakai index (hapus paling belakang)
data.pop(0)
print(data)

simpan = data.pop(1) # pop, value yang dihapus bisa disimpan
print(data)
print(f"value yang terhapus = {simpan}")

# clear 
data.clear()
print(data)
```

```python
order = []

kondisi = True
while kondisi:
    print("Silahkan pilih menu makanan : ")
    print("1. nasi goreng")
    print("2. mie goreng")
    print("3. es jeruk")
    print("4. jus mangga")
    
    pesan = input("ketikan menu yang akan dipesan : ")
    order.append(pesan)
    
    tanya = input("masih ingin memesan ? ")
    if tanya == "tidak":
        print("pesanan : ", order)
        print("total harga : ", len(order) * 20000)
        kondisi = False
```

```python
# studi kasus looping : 

# buatkan sistem yang menerima input bilangan bulat berulang kali 
# sampai pengguna memasukkan angka 0, buat dengan versi while & for
# Tampilkan angka yang dimasukkan (tidak termasuk 0)

# step 1 : buat list untuk menyimpan angka inputan.
# step 1 : buat perulangan untuk inputan.
# step 2 : setelah perulangan input selesai, buat perulangan untuk menampilkan angka
# step 2 : di dalam looping buat kondisi jika angkanya > 0 tampilkan.
```

```python
angka = [] # [1,2,3,4]

for i in range(5):
    inputan = input("masukan angka : ")
    angka.append(inputan)

for i in angka: 
    if int(i) > 0:
        print(i)
```

```python
menu = ["nasi goreng", "mie goreng"]
order = []

while True:
    print("menu : ")
    for i in menu:
        print(i)
        
    pesanan = input("masukan pesanan : ")
    if pesanan in menu :
        order.append(pesanan)
    
    tanya = input("masih ingin memesan ? ")
    if tanya == "tidak":
        print("pesanan : ", order)
        print("total harga : ", len(order) * 20000)
        break
    
    
```

```python
# tugas 1 

harga = [15000, 32000, 8000, 45000, 12000] 
jumlah = 0

for i in harga:
    jumlah += i 

print(jumlah)
```

```python
# tugas 4 

# Diberikan sebuah list data_mentah = [1, 2, 2, 3, 4, 4, 4, 5, 1, 6]. 
# Buatlah sebuah list baru bernama data_bersih. 
# Gunakan perulangan untuk memasukkan angka dari data_mentah ke data_bersih 
# hanya jika angka tersebut belum ada di data_bersih.
```

```python
data_mentah = [1, 2, 2, 3, 4, 4, 4, 5, 1, 6]
data_bersih = []

for i in data_mentah:
    ada = i in data_bersih
    if ada == False:
        data_bersih.append(i)

print(data_bersih)
```
