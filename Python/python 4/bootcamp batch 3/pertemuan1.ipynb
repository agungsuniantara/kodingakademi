```python
# tipe data (string, integer, float, boolean)

nama = "sandhi" # string (kalimat)
angka_int = 10 # integer (bilangan bulat)
angka_float = 3.94 # float (bilangan desimal)
boolean_benar = True # boolean (true / false)
boolean_salah = False # boolean (true / false)

print(type(nama))
print(type(angka_int))
print(type(angka_float))
print(type(boolean_benar))
print(type(boolean_salah))
```

```python
# operator (aritmatika, perbandingan, logika, penugasan)

# aritmatika (+, -, *, /, //, %, **)

a = 20
b = 3

print("a + b =", a + b) # pertambahan
print("a - b =", a - b) # pengurangan
print("a * b =", a * b) # perkalian
print("a / b =", a / b) # pembagian (hasilnya float)
print("a // b =", a // b) # floor division (hasilnya integer / pembulatan ke bawah)
print("a % b =", a % b)  # modulus (sisa bagi)
print("a ** b =", a ** b) # pangkat (a pangkat b)
```

```python
# penugasan (=, +=, -=, *=, /=, //=, %=, **=)

a = 10
print("nilai a =", a)

a += 5 # a = a + 5
print("nilai a setelah ditambah 5 =", a)

a -= 3 # a = a - 3
print("nilai a setelah dikurangi 3 =", a)
```

```python
# perbandingan (==, !=, >, <, >=, <=)

a = 10
b = 5

print("a == b =", a == b) # sama dengan
print("a != b =", a != b) # tidak sama dengan
print("a > b =", a > b) # lebih besar dari
print("a < b =", a < b) # lebih kecil dari
print("a >= b =", a >= b) # lebih besar atau sama dengan
print("a <= b =", a <= b) # lebih kecil atau sama dengan
```

```python
# logika (and, or, not)

# and = jika salah satu kondisi bernilai false, maka hasilnya false
# or = jika salah satu kondisi bernilai true, maka hasilnya true
# not = kebalikan dari kondisi

a = 10
b = 5
c = 15

print("a > b and a > c =", a > b and a > c)
print("a > b or a > c =", a > b or a > c) 
print("!(a > b) =", not(a > b) and a < b)

print("a > b and a > c and b > c =", a > b and a > c and b > c) 
print("a > b or a > c and a < b =", a > b or a > c and a < b)
```
