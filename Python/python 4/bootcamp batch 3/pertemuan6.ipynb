```python
# dictionary = tidak bisa di akses menggunakan index

data_list = [10,20,30,40] # list [0,1,2,3]
data_dictionary = {"nama":"agung","alamat":"denpasar"} # {key : value}
data_dict = dict(nama = "mamat", alamat = "denpasar", umur = 20)

print(data_list)
print(data_list[0])

print(data_dictionary)
print(data_dictionary["nama"])

print(data_dict["nama"])
print(data_dict["alamat"])
```

```python
# dictionary (copy, clear, fromkey, items, keys, values)

data = {
    "nama" : "agung",
    "alamat" : "denpasar"
}

# data_copy = data  -- jika tidak menggunakan copy, data asli dan data copy nyambung
data_copy = data.copy()
data_copy["nama"] = "wahyu"

print(data)
print(data_copy)
```

```python
# clear 

data.clear()
print(data)
```

```python
# fromkey

key_baru = {"sapi", "ayam", "kambing"}
hewan = dict.fromkeys(key_baru, "hewan")
print(hewan)
```

```python
# keys & values

data = {"nama" : "agung", "alamat" : "denpasar"}
print(data.keys())
print(data.values())

for i in data.keys():
    print(i)

for i in data.values():
    print(i)
```

```python
# items

data = {"nama" : "agung", "alamat" : "denpasar"}
print(data.items())

data_backup = data.copy()

for i , b in data_backup.items():
    if b == "denpasar":
        data.pop(i)

print(data)
```

```python
data = {
    1 : "ayam",
    2 : "sapi",
    3 : "kambing"
}
backup = data.copy()

for i in backup.keys():
    if i % 2 != 0:
        data[i] = "hewan"
```
