```python
# if , elif, else 
# if = jika kondisi terpenuhi maka akan menjalankan perintah
# elif = jika kondisi if tidak terpenuhi maka akan mengecek kondisi elif
# else = jika kondisi if dan elif tidak terpenuhi maka akan menjalankan perintah else
# jika if kemudian if maka akan mengecek kondisi if kedua meskipun kondisi if pertama terpenuhi

if True:
    print("ini adalah if")

if False:
    print("ini adalah if")
elif False:
    print("ini adalah elif 1")
elif False:
    print("ini adalah elif 2")
else:
    print("ini adalah else")
```

```python
# looping 

# for = untuk melakukan perulangan dengan jumlah yang sudah ditentukan
# while = untuk melakukan perulangan dengan kondisi tertentu

# for 

angka = 5

for i in range(angka): # i = 0
    print(i)
    print("perulangan")

print("\n")

for i in range(1, 10): # i = 1, (start, stop)
    print(i)
    print("perulangan")

print("\n")

for i in range(1, 10, 2): # i = 1, (start, stop, step)
    print(i)
    print("perulangan")
```

```python
# while 

angka = 0
end_score = 10

while angka < end_score:
    print("perulangan")
    angka += 1
    
    if angka == 5:
        break # untuk menghentikan perulangan
```

```python
# No 5

jumlah = 0
count = 0

while count < 5:
    angka = int(input("masukan angka : "))
    print(f"Masukkan angka {angka}")
    jumlah += angka 
    count += 1

print("Rata - rata = ", jumlah / 5)
    
```
