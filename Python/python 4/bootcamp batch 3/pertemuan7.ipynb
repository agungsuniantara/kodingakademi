```python
# function = blok kode yang berisikan perintah
# yang dapat dipanggil berulang kali

# function (void, parameter, return)

# function void (fungsi yang hasil dari pemanggilan langsung ditampilkan)

def tambah():
    angka1 = 10
    angka2 = 20
    hasil = angka1 + angka2
    print(hasil)

tambah()
tambah()
```

```python
score = 0

def ubah():
    global score
    score += 1
    print(score)

ubah()
```

```python
# isinstance = untuk memeriksa suatu object dari tipe data tertentu

x = "hello"

print(isinstance(x, str))
print(isinstance(x, int))

data = [10, "hello", 3.14]

jumlah = 0

for i in data:
    if isinstance(i, (int, float)):
        print(f"{i} adalah angka")
        jumlah += i
    else:
        print(f"{i} bukan angka")

print(f"total = {jumlah}")
```

```python
def menu():
    print("pilihlah menu : ")
    print("1. pertambahan")
    print("2. pengurangan")
    print("3. pembagian")

def pertambahan():
    angka1 = int(input("masukan angka1 : "))
    angka2 = int(input("masukan angka2 : "))
    hasil = angka1 + angka2
    print(f"hasil pertambahan : {hasil}")

def pengurangan():
    angka1 = int(input("masukan angka1 : "))
    angka2 = int(input("masukan angka2 : "))
    hasil = angka1 - angka2
    print(f"hasil pengurangan : {hasil}")
    
while True:
    menu()
    
    try:
        inputan = int(input("masukan pilihan : "))
        if inputan == 1:
            tambah()
        elif inputan == 2:
            pengurangan()
        else:
            print("pilihan tidak tersedia")
        
        kondisi = input("apakah ingin memilih lagi ?").lower()
        if kondisi == "tidak":
            break
    except:
        print("inputan harus angka / terdapat kesalahan")
```

```python
# function parameter : fungsi yang dapat menerima inputan

def nama(a, b): # fungsi(parameter, parameter)
    print(f"Halo {a}")
    print(f"Alamat : {b}")

nama("agung", "denpasar")
nama("mas mamat", "pemogan")
```

```python
# fungsi return : fungsi yang pemanggilannya menghasilkan nilai

def pertambahan(a,b):
    hasil = a + b
    print(hasil)
    return hasil

jawaban = pertambahan(10,10) # hasil masuk ke variabel jawaban
print(jawaban)
```

```python
# Map

def kali(x): 
    return x * 2

hasil = list(map(kali, [10,20,30,40]))
print(hasil)
```

```python
# nested function

def data_diri(x, z):
    print(f"nama = {x}")
    print(f"alamat = {z}")

def luas_pp(a, b, c, d):
    data_diri(a, b)
    print(f"luas = {c * d}")

data_diri("agung", "denpasar")
luas_pp("agung", "denpasar", 20, 40)
```

```python
# function rekursif

def faktorial(n): # n = 1
    if n == 0:
        return 1
    else:
        return n * faktorial(n - 1)

# 5 * (4 * (3 * (2 * (1 * 1))))

hasil = faktorial(5)
print(hasil)
```
