```python
# list = variabel yang dapat menyimpan value lebih dari satu.
# dictionary = variabel yang dapat menyimpan key & value

data_list = [1,2,3,4] # list memakai index
data_dictionary = {"nama" : ["agung", "sandhy"], "alamat" : "denpasar"} # dict memakai key

print(data_list[0])
print(data_dictionary["nama"])
print(data_dictionary["nama"][0])
print(data_dictionary["alamat"])
```

```python
# cara lain untuk membuat dictionary

data_dict = dict(nama = "agung", alamat = "denpasar")
print(data_dict["nama"])
```

```python
# edit dictionary

data_dict = dict(nama = "agung", alamat = "denpasar")
print(data_dict)

data_dict["nama"] = "wahyu"
print(data_dict)
```

```python
# terdapat beberapa fungsi pada dictionary

# copy (salin data) || bisa digunakan pada list & dictionary

data_mentah = {1:"hello", 2:"world"}
data_baru = data_mentah # tidak menggunakan fungsi copy

data_baru[1] = "halo"
print(data_baru)
print(data_mentah) # data_baru dirubah maka data_mentah akan berubah juga

data_baru2 = data_mentah.copy() # menggunakan fungsi copy
data_baru2[1] = "halo baru"
print(data_baru2)
print(data_mentah) # jika menggunakan copy, data lama tidak berubah.

test = [1,2,3,4]
test_baru = test.copy()
print(test_baru)
```

```python
# del, pop, clear ( meghapus dictionary )

data_dict = dict(nama = "agung", alamat = "denpasar", hoby = "coding")
del data_dict["nama"]
print(data_dict)

simpan = data_dict.pop("alamat")
print(data_dict)
print(simpan)

data_dict.clear()
print(data_dict)
```

```python
# update (menambah data)

data_dict = dict(nama = "agung", alamat = "denpasar")

data_dict.update({"hoby" : "coding", "status" : "belum menikah"})
print(data_dict)
```

```python
# keys = "fungsi untuk mengembalikan semua key"
# values = "fungsi untuk mengembalikan semua value"
# items = "fungsi untuk memisahkan key & value"

data_dict = dict(nama = "agung", alamat = "denpasar")

# keys 
data_keys = data_dict.keys()
print(data_keys)

for i in data_keys:
    print(i)

# values
data_values = data_dict.values()
print(data_values)

for i in data_values:
    print(i)

# items
data_items = data_dict.copy().items() # gunakan copy jika ingin menghapus data dictionary.
print(data_items)

for i, z in data_items:
    print(f"keys : {i}, values : {z}")

for i, z in data_items: # i = "keys", z = "values"
    if z == "denpasar":
        del data_dict[i]
```

```python
# studi kasus

data_siswa = {
    "Andi" : {"Matematika": 85, "IPA": 90},
    "Budi" : {"Matematika": 70, "IPA": 75},
    "Citra" : {"Matematika": 95, "IPA": 92}
}

print(data_siswa)
print(data_siswa["Andi"])
print(data_siswa["Andi"]["Matematika"])

# update 

data_siswa.update({"Dina": {"Matematika": 88, "IPA": 80}})
print(data_siswa)

# ubah nilai

data_siswa["Andi"]["IPA"] = 60
print(data_siswa)

# hapus nilai

siswa_keluar = data_siswa.pop("Andi", "Siswa tidak ada")
print(data_siswa)
print(siswa_keluar)
data_siswa.pop("Andi", "Siswa tidak ada") # jika data tidak ada output = Siswa tidak ada"

# rata - rata nilai setiap siswa

for a, b in data_siswa.items():
    nilai_matematika = b["Matematika"]
    nilai_ipa = b["IPA"]
    
    rata = (nilai_matematika + nilai_ipa) / 2
    
    print(f"Nama : {a}")
    print(f"Nilai matematika : {nilai_matematika}")
    print(f"Nilai ipa : {nilai_ipa}")
    print(f"rata - rata : {rata}")
```

```python
# tugas 1

# buatkan sistem kelola buku.
# buat dictionary untuk menyimpan buku, datanya : 
# action : "marvel", "spidermaen"
# romance : "cinta kasih"

# buatkan chatbot untuk tambah, edit, hapus, tampil data.

data = {
    "action" : ["marvel", "spidermaen"],
    "romance" : ["cinta kasih"]
}

while True:
    print(data)
    print("menu : ")
    print("1. tambah data")
    print("2. edit data")
    print("3. hapus data")
    print("4. tampilkan data")
    print("5. keluar")
    
    pilih = input("pilih menu : ")
    if pilih == "1":
        genre = input("masukan genre : ")
        if genre in data.keys():
            buku = input("masukan nama buku : ")
            data[genre].append(buku)
        else:
            buku = input("masukan nama buku : ")
            data.update({genre : [buku]})
    elif pilih == "2":
        genre = input("masukan genre : ")
        if genre in data.keys():
            buku = input("masukan nama buku di edit : ")
            if buku in data[genre]:
                index_buku = data[genre].index(buku) # fungsi index untuk cari index dari value
                buku_baru = input("Masukan nama buku baru : ")
                data[genre][index_buku] = buku_baru
                print("Buku berhasil diubah!")
        else:
            print("genre buku tidak tersedia !!")
    elif pilih == "3":
        genre = input("masukan genre yang ingin dihapus : ")
        if genre in data.keys():
            del data[genre]
    elif pilih == "4":
        print(data)
    elif pilih == "5":
        print()
        break
```
