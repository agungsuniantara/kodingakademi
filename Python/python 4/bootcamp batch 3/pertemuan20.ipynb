```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 

chrome_driver_path = "/Applications/MAMP/htdocs/local/kodingakademi/Python/python 4/bootcamp batch 3/chromedriver-mac-arm64/chromedriver"

service = Service(chrome_driver_path)
driver = webdriver.Chrome(service=service)

driver.get("https://www.wikipedia.org/")

print("Judul Halaman:", driver.title)
```

```python
from selenium.webdriver.common.by import By 

driver.implicitly_wait(10)

element = driver.find_element(By.CLASS_NAME, "central-textlogo")
print(element.text)
```

```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

wait = WebDriverWait(driver, 10)
element = wait.until(EC.text_to_be_present_in_element((By.CLASS_NAME, "central-textlogo"),"Wikipedia"))

if element :
    print("Teks ditemukan dalam elemen.")
else:
    print("Teks tidak ditemukan dalam elemen.")
```

```python
import time 

last_height = driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height
```
