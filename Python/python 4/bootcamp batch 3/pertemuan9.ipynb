```python
# numpy = sebuah library yang fokus untuk mengolah angka

import numpy as np

data_2d = np.array([
    [1,2,3],
    [4,5,6]
])

print(data_2d)
```

```python
data_3d = np.array([
    [
        [1,2,3],
        [4,5,6]
    ],
    [
        [1,2,3],
        [4,5,6]
    ]
])

print(data_3d)
```

```python
# fungsi numpy (shape, ndim, size)

print(f"shape 2d : {data_2d.shape}, 3d : {data_3d.shape}")
print(f"ndim 2d : {data_2d.ndim}, 3d : {data_3d.ndim}")
print(f"size 2d : {data_2d.size}, 3d : {data_3d.size}")
```

```python
dataset = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

# indexing numpy

print(dataset)
print(dataset[0][0])
print(dataset[1][1])

# slicing numpy

print(dataset[0:2]) # [baris]
print(dataset[0:2, 0:2]) # [baris, kolom]

# ubah nilai numpy

dataset[2][1] = 100
print(dataset)

dataset[0] = [1000,2000,3000]
print(dataset)
```

```python
# pandas (series & dataframe)
# series = data yang tidak memiliki kolom.
# dataframe = data yang memiliki baris dan kolom

import pandas as pd
import numpy as np

data = np.array([10,22,30,40])
series = pd.Series(data)
dataframe = pd.DataFrame(data)
print(series)
print(dataframe)
```

```python
# series operation

# deklarasi 

data_list = ["data1", "data2"]
data_dict = {"data1" : 90, "data2" : 100}

series_list = pd.Series(data_list, index = ["satu", "dua"])
series_dict = pd.Series(data_dict)
series_scalar = pd.Series(99, index = [i for i in range(100)])

print(series_list)
print(series_dict)
print(series_scalar)
```

```python
data1 = pd.Series([10,20,30,40], index = [1,2,3,4])
data2 = pd.Series([10,20,30,40], index = ["satu","dua","tiga","empat"])

# indexing
print(data1[1])
print(data2["satu"])

# slicing
print(data1[0:3]) # harus menggunakan index asli
print(data2["satu":"tiga"])

# loc (index edit) & iloc (index asli)
print(data1.loc[1:3])
print(data1.iloc[0:3])

print(data2.loc["satu":"tiga"])
print(data2.iloc[0:3])
```
