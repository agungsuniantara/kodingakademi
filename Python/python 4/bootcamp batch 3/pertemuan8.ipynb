```python
def min_max(x):
    min_x = x[0]
    max_x = x[0]
    
    for i in x:
        if i > max_x:
            max_x = i
        elif i < min_x:
            min_x = i
    
    return min_x, max_x

terkecil, terbesar = min_max([10,5,4,2,100])
print(f"min = {terkecil}, max = {terbesar}")
```

```python
# function args & kwargs

# args : parameter yang dapat menampung nilai lebih dari 1
# kwargs : parameter yang dapat menampung key : value

# args

def data_diri(a, b):
    print(a)
    print(b)

def data_diri2(*a): # a = ("mamat", "denpasar")
    print(a)
    print(a[0])
    print(a[1])

data_diri("agung", "denpasar")
data_diri2("mamat", "denpasar")
```

```python
# kwargs

def test1(x):
    print(x)
    
def test2(**x): # parameter kwargs dapat memproses value dict lebih dari 1
    print(x)

# test(nama = "wahyu") # parameter biasa = tidak bisa memproses hasil dict
test2(nama = "wahyu", alamat = "denpasar")
```

```python
def min_max(*a): # a = ([.....])
    min_a = a[0][0]
    max_a = a[0][0]
    
    for i in a: # i = [.......]
        for z in i:
            if z < min_a:
                min_a = z
            elif z > max_a:
                max_a = z
    
    return min_a, max_a

angka = []
jumlah = int(input("masukan jumlah : "))

for i in range(jumlah):
    angka_input = int(input(f"masukan angka ke-{i+1} : "))
    angka.append(angka_input)

terkecil, terbesar = min_max(angka)
print(f"Terkecil = {terkecil}")
print(f"Terbesar = {terbesar}")
```
