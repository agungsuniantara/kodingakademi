```python
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize

kalimat = "Halo, selamat datang di dunia NLP!"
tokens = word_tokenize(kalimat)
print("Word tokens:", tokens)
# Output: ['Halo', ',', 'selamat', 'datang', 'di', 'dunia', 'NLP', '!']
```

```python
from nltk.tokenize import sent_tokenize

teks = "Halo! Selamat datang. NLP sangat menarik."
kalimat_tokens = sent_tokenize(teks)
print("Kalimat tokens:", kalimat_tokens)
# Output: ['Halo!', 'Selamat datang.', 'NLP sangat menarik.']
```

```python
import re, nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')

text = "AKU SUKA banget makan baksoooo 😍😍!!! #BaksoEnak www.bakso123.com"

# 1. Lowercasing
text = text.lower()

# 2. Cleaning dengan regex (hapus url dan simbol)
text = re.sub(r"http\S+|www\S+|https\S+", "", text)
text = re.sub(r"[^a-zA-Z\s]", "", text)

# 3. Tokenization
tokens = word_tokenize(text)
print("Hasil tokenisasi:", tokens)
# Output: ['aku', 'suka', 'banget', 'makan', 'baksoooo', 'baksoenak']
```
