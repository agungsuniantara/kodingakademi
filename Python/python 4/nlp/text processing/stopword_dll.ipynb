```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
nltk.download('stopwords')
nltk.download('punkt')

kalimat = "Saya ingin belajar Natural Language Processing di tahun 2024."
stop_words = set(stopwords.words('indonesian'))  # Bisa 'english' untuk Bahasa Inggris

tokens = word_tokenize(kalimat)
filtered = [w for w in tokens if w.lower() not in stop_words]
print("Tanpa stopword:", filtered)
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

factory = StemmerFactory()
stemmer = factory.create_stemmer()

kalimat = "Mahasiswa-mahasiswa itu sedang berlarian dan membaca buku-buku baru."
hasil_stem = stemmer.stem(kalimat)
print("Setelah stemming:", hasil_stem)
# Output: 'mahasiswa itu sedang lari dan baca buku baru'
```

```python
from nltk.stem.snowball import SnowballStemmer

stemmer = SnowballStemmer("english")
kata_kata = ["running", "flies", "easily", "fairly"]
hasil = [stemmer.stem(kata) for kata in kata_kata]
print("Setelah stemming:", hasil)
# Output: ['run', 'fli', 'easili', 'fair']
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

kalimat = "He is running and eats apples."
doc = nlp(kalimat)
lemmas = [token.lemma_ for token in doc]
print("Lemmatized:", lemmas)
# Output: ['he', 'be', 'run', 'and', 'eat', 'apple', '.']
```

```python
import re, nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

nltk.download('punkt')
nltk.download('stopwords')

text = "AKU SUKA banget makan baksoooo 😍😍!!! #BaksoEnak www.bakso123.com"

# Lowercasing
text = text.lower()

# Cleaning (hapus url & simbol)
text = re.sub(r"http\S+|www\S+|https\S+", "", text)
text = re.sub(r"[^a-zA-Z\s]", "", text)

# Tokenization
tokens = word_tokenize(text)

# Stopword removal
stop_words = set(stopwords.words('indonesian'))
tokens = [t for t in tokens if t not in stop_words]

# Stemming
stemmer = StemmerFactory().create_stemmer()
hasil = stemmer.stem(' '.join(tokens))

print("Hasil akhir (Stopword Removal + Stemming):", hasil)
```

```python
import spacy
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
nltk.download('stopwords')
nltk.download('punkt')

nlp = spacy.load("en_core_web_sm")

text = "He is running and eats apples in the park."
tokens = word_tokenize(text)
stop_words = set(stopwords.words('english'))
tokens = [w for w in tokens if w.lower() not in stop_words]
filtered_text = ' '.join(tokens)

doc = nlp(filtered_text)
lemmas = [token.lemma_ for token in doc]
print("Lemmatized:", lemmas)
# Output: ['run', 'eat', 'apple', 'park', '.']
```
