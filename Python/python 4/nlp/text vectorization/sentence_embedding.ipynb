```python
from gensim.models import Word2Vec
import numpy as np

def sentence_embedding(sentence, model):
    vectors = [model.wv[w] for w in sentence if w in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

corpus = [
    ['aku', 'suka', 'minum', 'kopi', 'pagi', 'hari'],
    ['kopi', 'panas', 'lebih', 'nikmat'],
    ['setiap', 'hari', 'minum', 'kopi'],
    ['kopi', 'hitam', 'tanpa', 'gula'],
    ['banyak', 'orang', 'tidak', 'bisa', 'lepas', 'dari', 'kopi'],
    ['teh', 'juga', 'minuman', 'favorit'],
    ['ada', 'yang', 'lebih', 'suka', 'teh', 'dibanding', 'kopi'],
    ['kopi', 'dan', 'teh', 'punya', 'penggemar', 'masing-masing'],
    ['kopi', 'hangat', 'bikin', 'semangat'],
    ['kopi', 'dingin', 'enak', 'saat', 'siang', 'hari']
]

model = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, workers=4)

kalimat1 = ['minum', 'kopi', 'pagi', 'hari']
kalimat2 = ['kopi', 'hangat', 'pagi', 'hari']

emb1 = sentence_embedding(kalimat1, model)
emb2 = sentence_embedding(kalimat2, model)

similarity = cosine_similarity(emb1, emb2)

print("Vektor kalimat 1:", emb1)
print("Vektor kalimat 2:", emb2)
print("Cosine similarity:", similarity)
```

```python
# USE

import tensorflow_hub as hub
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
    
# Load model dari TF Hub
embed = hub.load("https://tfhub.dev/google/universal-sentence-encoder/4")

kalimat1 = "Saya suka minum kopi pagi hari."
kalimat2 = "Minum kopi di pagi hari sangat menyenangkan."

# Embedding (output: [512])
vecs = embed([kalimat1, kalimat2])

print("USE Embedding kalimat 1:\n", np.array(vecs[0]))
print("USE Embedding kalimat 2:\n", np.array(vecs[1]))

# Penggunaan: Kemiripan dua kalimat
print("Similarity (USE):", cosine_similarity(vecs[0], vecs[1]))
```
