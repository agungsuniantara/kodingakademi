```python
# manual corpus

import gensim
from gensim.models import Word2Vec

# Contoh data (bisa berupa kalimat bahasa Indonesia)
corpus = [
    ['aku', 'suka', 'minum', 'kopi', 'pagi', 'hari'],
    ['kopi', 'panas', 'lebih', 'nikmat'],
    ['setiap', 'hari', 'minum', 'kopi'],
    ['kopi', 'hitam', 'tanpa', 'gula'],
    ['banyak', 'orang', 'tidak', 'bisa', 'lepas', 'dari', 'kopi'],
    ['teh', 'juga', 'minuman', 'favorit'],
    ['ada', 'yang', 'lebih', 'suka', 'teh', 'dibanding', 'kopi'],
    ['kopi', 'dan', 'teh', 'punya', 'penggemar', 'masing-masing'],
    ['kopi', 'hangat', 'bikin', 'semangat'],
    ['kopi', 'dingin', 'enak', 'saat', 'siang', 'hari']
]

model = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, workers=4)
```

```python
print("Vektor kata 'kopi':\n", model.wv['kopi'])
```

```python
print("Kata paling mirip 'kopi':")
print(model.wv.most_similar('kopi'))
```

```python
print("Kemiripan 'kopi' dan 'teh':", model.wv.similarity('kopi', 'teh'))
```

```python
print("Analogi: kopi + hangat - dingin ≈ ?", model.wv.most_similar(positive=['kopi', 'hangat'], negative=['dingin']))
```

```python
# corpus nltk 

import nltk
nltk.download('brown')
from nltk.corpus import brown

# Setiap kalimat adalah list kata (sudah tokenized)
corpus = brown.sents()
print(corpus[:2])  # Contoh output: [['The', 'Fulton', 'County', ...], ['The', 'jury', ...]]
```

```python
from gensim.models import Word2Vec

# Training Word2Vec pada data nyata
model = Word2Vec(sentences=corpus, vector_size=100, window=5, min_count=2, workers=4)
```

```python
# Vektor untuk kata
print("Vektor 'coffee' (jika ada di corpus):", model.wv['coffee'])

# Kata paling mirip
print("Kata paling mirip 'money':", model.wv.most_similar('coffee'))

# Cek kemiripan dua kata
print("Kemiripan 'love' dan 'like':", model.wv.similarity('love', 'like'))
```

```python
import gensim.downloader as api

# Download GloVe (dimensi 50, bahasa Inggris, ringan)
model_glove = api.load('glove-wiki-gigaword-50')
```

```python
# Vektor kata 'coffee' (pakai bahasa Inggris karena pre-trained)
print(model_glove['coffee'])

# Kata paling mirip 'coffee'
print(model_glove.most_similar('coffee'))

# Cek kemiripan dua kata
print(model_glove.similarity('coffee', 'tea'))
```

```python
from gensim.models import FastText

corpus = [
    ['aku', 'suka', 'minum', 'kopi', 'pagi', 'hari'],
    ['kopi', 'panas', 'lebih', 'nikmat'],
    ['setiap', 'hari', 'minum', 'kopi'],
    ['kopi', 'hitam', 'tanpa', 'gula'],
    ['banyak', 'orang', 'tidak', 'bisa', 'lepas', 'dari', 'kopi'],
    ['teh', 'juga', 'minuman', 'favorit'],
    ['ada', 'yang', 'lebih', 'suka', 'teh', 'dibanding', 'kopi'],
    ['kopi', 'dan', 'teh', 'punya', 'penggemar', 'masing-masing'],
    ['kopi', 'hangat', 'bikin', 'semangat'],
    ['kopi', 'dingin', 'enak', 'saat', 'siang', 'hari']
]

# Training
model_ft = FastText(sentences=corpus, vector_size=50, window=3, min_count=1, workers=4)
```

```python
print("Vektor untuk 'kopi':\n", model_ft.wv['kopi'])
print("Kata paling mirip 'kopi':\n", model_ft.wv.most_similar('kopiii'))
print("Kemiripan 'kopi' dan 'teh':", model_ft.wv.similarity('kopi', 'teh'))
print("Vektor untuk 'kopiii':\n", model_ft.wv['kopiii'])  # FastText masih bisa kasih vektor
```
