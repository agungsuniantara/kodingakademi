```python
import tensorflow as tf
import numpy as np
import tensorflow as tf
import random
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

# --- 1. Data Pipeline ---

def label_map(status):
    return tf.cast(tf.equal(status, b'success'), tf.float32)

def preprocess(features, label):
    x = tf.stack([features['amount']], axis=1)
    return x, label_map(label)

ds = tf.data.experimental.make_csv_dataset(
    'transaksi.csv',
    batch_size=4,
    label_name='status',
    num_epochs=1,
    shuffle=True,
    select_columns=['amount', 'status']
)

ds = ds.map(preprocess)
ds = ds.prefetch(tf.data.AUTOTUNE)

# --- 2. Model Sequential API ---
model = tf.keras.Sequential([
    tf.keras.layers.Dense(8, activation='relu', input_shape=(1,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# --- 3. Training ---
model.fit(ds, epochs=30)

# --- 4. Prediksi (Opsional) ---
# Contoh prediksi: amount = 900
import numpy as np
X_new = np.array([[900.0]])
y_pred = model.predict(X_new)
print(f"Prediksi probabilitas sukses untuk amount=900: {y_pred[0,0]:.3f}")
```
