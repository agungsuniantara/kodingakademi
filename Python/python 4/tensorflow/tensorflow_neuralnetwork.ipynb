NEURAL NETWORK

Kasus: Prediksi Siswa Lulus / Tidak Lulus

Fitur Data :

x1: Nilai Matematika

x2: Nilai Bahasa Inggris

Label (Target) :

1 = Lulus

0 = Tidak lulus

Struktur jaringan : 

Input layer : 2 neuron (x1, x2)

Hidden layer : 2 neuron (h1, h2)

Output layer : 1 neuron (prediksi: lulus/tidak)


HIDDEN LAYER

Neuron h1 : 

z1 = (x1 * w11) + (x2 * w12) + b1

   = (85 * 0.6) + (80 * 0.4) + (-0.3)
   
   = 51 + 32 - 0.3
   
   = 82.7


h1 = ReLU(z1) 
= max(0, 82.7) = 82.7

Neuron h2 : 

z2 = (x1 * w21) + (x2 * w22) + b2
   
   = (85 * 0.5) + (80 * 0.7) + 0.1
   
   = 42.5  + 56     + 0.1

   = 98.6

h2 = ReLU(z2) = max(0, 98.6) = 98.6


OUTPUT LAYER

zo = (h1 * v1) + (h2 * v2) + bo

   = (82.7 * 0.8) + (98.6 * 0.6) + (0.2)

   = 66.16 + 59.16 + 0.2
   
   = 125.52


SIGMOID

output = 1 / (1 + exp(-125.52))

output ≈ 1 / (1 + 0.000...0001)

output ≈ 1 / 1

output ≈ 1



KESIMPULAN SIMULASI

Input: Nilai Matematika 85, Bahasa Inggris 80

Prediksi jaringan: LULUS (output hampir 1)

Proses:

Input × bobot + bias → hidden layer (pakai ReLU)

Hidden layer × bobot + bias → output layer (pakai sigmoid)

Hasil antara 0–1 (semakin dekat 1 = lulus, 0 = tidak lulus)

```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Input data
x1 = 70  # Nilai Matematika
x2 = 80  # Nilai Bahasa Inggris

# Bobot dan bias untuk hidden layer
bobot1, bobot2 = 0.2, 0.4
bobot3, bobot4 = 0.3, 0.5
bias1, bias2 = 0.1, -0.2

# Bobot dan bias untuk output layer
bobot5, bobot6 = 0.6, 0.8
bias3 = 0.05

# --- Hidden Layer ---

# neuron 1 (h1)
z1 = (x1 * bobot1) + (x2 * bobot2) + bias1
h1 = relu(z1)

# neuron 2 (h2)
z2 = (x1 * bobot3) + (x2 * bobot4) + bias2
h2 = relu(z2)

# --- Output Layer ---

z0 = (h1 * bobot5) + (h2 * bobot6) + bias3
output = sigmoid(z0)

print("Input:", [x1, x2])
print("Neuron 1 (h1):", h1)
print("Neuron 2 (h2):", h2)
print("Output (Sigmoid):", output)
```

```python
# Input data
x1 = 70  # Nilai Matematika
x2 = 80  # Nilai Bahasa Inggris

# Bobot dan bias hidden layer 1
bobot1, bobot2 = 0.2, 0.4
bias1 = 0.1
bobot3, bobot4 = 0.3, 0.5
bias2 = -0.2

# Bobot dan bias hidden layer 2
bobot5, bobot6 = 0.5, 0.7
bias3 = 0.2
bobot7, bobot8 = 0.6, 0.8
bias4 = -0.1

# Bobot dan bias output layer
bobot9, bobot10 = 0.9, 1.1
bias5 = 0.05

# --- Hidden Layer 1 ---

# Neuron h1
z1 = (x1 * bobot1) + (x2 * bobot2) + bias1
h1 = relu(z1)

# Neuron h2
z2 = (x1 * bobot3) + (x2 * bobot4) + bias2
h2 = relu(z2)

# --- Hidden Layer 2 ---

# Neuron h3
z3 = (h1 * bobot5) + (h2 * bobot6) + bias3
h3 = relu(z3)

# Neuron h4
z4 = (h1 * bobot7) + (h2 * bobot8) + bias4
h4 = relu(z4)

# --- Output Layer ---

z0 = (h3 * bobot9) + (h4 * bobot10) + bias5
output = sigmoid(z0)

print("Input:", [x1, x2])
print("Hidden Layer 1 neuron h1:", h1)
print("Hidden Layer 1 neuron h2:", h2)
print("Hidden Layer 2 neuron h3:", h3)
print("Hidden Layer 2 neuron h4:", h4)
print("Output (Sigmoid):", output)
```
