```python
import tensorflow as tf

scalar = tf.constant(7)  
vector = tf.constant([1.0, 2.0, 3.0])  
matrix = tf.constant([[1, 2], [3, 4]])  
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print(scalar)
print(vector)
print(matrix)
print(tensor_3d)
```

SIFAT TENSOR

```python
print("Shape:", matrix.shape)  # Output: (2, 2)
print("Tipe data:", matrix.dtype)  # Output: <dtype: 'int32'>
print("Nilai: \n", matrix.numpy())  # Konversi ke numpy array
```

```python
zeros = tf.zeros([2, 3]) 
ones = tf.ones([3])  
eye = tf.eye(3)  
random = tf.random.normal([2, 2], mean=0, stddev=1)

print(f"zeros : \n {zeros.numpy()}")
print(f"ones : \n {ones.numpy()}")
print(f"eye : \n {eye.numpy()}")
print(f"random : \n {random.numpy()}")
```

MANIPULASI TENSOR

indexing

```python
# indexing

tensor_1d = tf.constant([1, 2])
tensor_2d = tf.constant([[1, 2], [3, 4], [5, 6]])
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

elem_1d = tensor_1d[0]
elem_2d = tensor_2d[1, 0]
elem_3d = tensor_3d[1, 0, 1]

print(f"indexing 1d : {elem_1d.numpy()}")
print(f"indexing 2d : {elem_2d.numpy()}")
print(f"indexing 3d : {elem_3d.numpy()}")
```

```python
tensor = tf.constant([[1, 2, 3], [4, 5, 6], [7, 8, 9]])

# Indexing
elem = tensor[1, 2] 

# Slicing
row = tensor[1, :]  
col = tensor[:, 1]  
sub = tensor[0:2, 1:3]  # [[2, 3], [5, 6]]
```

Slicing

```python
tensor = tf.constant([0, 1, 2, 3, 4, 5])
matrix = tf.constant([[1, 2, 3], [4, 5, 6], [7, 8, 9]])

slice1 = tensor[1:4]
slice2 = tensor[::2]

slice_matrix = matrix[:2, -1]
slice_col = matrix[:, 1:] 

print(f"slice 1 : {slice1.numpy()}")
print(f"slice 2 : {slice2.numpy()}")
print(f"slice matrix : {slice_matrix.numpy()}")
print(f"slice column : \n {slice_col.numpy()}")
```

reshape

```python
tensor = tf.constant([[1, 2], [3, 4], [5, 6]]) 
reshaped = tf.reshape(tensor, [6])
new_shape = tf.reshape(tensor, [2, 3])

print(f"tensor : \n {tensor.numpy()}")
print(f"reshaped : {reshaped.numpy()}")
print(f"new shape : \n {new_shape.numpy()}")
```

```python
matrix = tf.constant([[1, 2], [3, 4]]) 
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

transpose_matrix = tf.transpose(matrix) 
transpose_3d = tf.transpose(tensor_3d) 

print(f"matrix : \n {matrix.numpy()}")
print(f"tensor_3d : \n {tensor_3d.numpy()}")
print(f"transpose_matrix : \n {transpose_matrix.numpy()}")
print(f"transpose_3d : \n {transpose_3d.numpy()}")
```

```python
import tensorflow as tf

a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

concat_axis0 = tf.concat([a, b], axis=0)  # Gabung vertikal (baris)
concat_axis1 = tf.concat([a, b], axis=1)  # Gabung horizontal (kolom)

print(f"concat_axis0 : \n {concat_axis0.numpy()}")
print(f"concat_axis1 : \n {concat_axis1.numpy()}")
```

```python
a = tf.constant([1, 2])
b = tf.constant([3, 4])

stack0 = tf.stack([a, b], axis=0)
stack1 = tf.stack([a, b], axis=1)  
unstack = tf.unstack(stack0, axis=1)

print(f"stack0 : \n {stack0.numpy()}")
print(f"stack1 : \n {stack1.numpy()}")
print("Unstack :")
for tensor in unstack:
    print(tensor.numpy())
```

```python
tensor = tf.constant([[1, 2], [3, 4], [5, 6]])
result = tf.split(tensor, num_or_size_splits=3, axis=0)

print(f"tensor : \n {tensor.numpy()}")
print(f"split :")
for i in result:
    print(i.numpy())
```

```python
# NO 1

import tensorflow as tf

a = tf.constant([[10, 20]])
b = tf.constant([[30, 40]])

# Gabung Vertikal (axis=0)
concat_axis0 = tf.concat([a, b], axis=0)
print("Vertikal (axis=0):")
print(concat_axis0.numpy())

# Gabung Horizontal (axis=1)
concat_axis1 = tf.concat([a, b], axis=1)
print("Horizontal (axis=1):")
print(concat_axis1.numpy())
```

```python
# NO 2

tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

# Reshape menjadi (3,2)
reshaped = tf.reshape(tensor, (3, 2))
print("Reshape (3,2):")
print(reshaped.numpy())

# Transpose
transposed = tf.transpose(tensor)
print("Transpose:")
print(transposed.numpy())
```

```python
# NO 3

a = tf.constant([1, 2])
b = tf.constant([3, 4])
c = tf.constant([5, 6])

# Stack
stacked = tf.stack([a, b, c], axis=0)
print("Stacked:")
print(stacked.numpy())

# Unstack
unstacked = tf.unstack(stacked, axis=0)
print("Unstacked:")
for tensor in unstacked:
    print(tensor.numpy())
```
