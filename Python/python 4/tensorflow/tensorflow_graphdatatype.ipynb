Data type

```python
import tensorflow as tf

a = tf.constant(5, dtype=tf.int32)
b = tf.constant(10, dtype=tf.uint16)
c = tf.constant(3.14, dtype=tf.float64)
d = tf.constant(True, dtype=tf.bool)
e = tf.constant("Kalimat", dtype=tf.string)

print(f"a : {a.numpy()}, tipe data : {a.dtype}")
print(f"b : {b.numpy()}, tipe data : {b.dtype}")
print(f"c : {c.numpy()}, tipe data : {c.dtype}")
print(f"d : {d.numpy()}, tipe data : {d.dtype}")
print(f"e : {e.numpy()}, tipe data : {e.dtype}")
```

```python
# Secara eksplisit
a = tf.constant(5, dtype=tf.int32)
b = tf.constant(3.14, dtype=tf.float64)

# Implisit
c = tf.constant([1, 2, 3])  # otomatis tf.int32
d = tf.constant([1.0, 2.0]) # otomatis tf.float32

print(f"a : {a.numpy()}, tipe data : {a.dtype}")
print(f"b : {b.numpy()}, tipe data : {b.dtype}")
print(f"c : {c.numpy()}, tipe data : {c.dtype}")
print(f"d : {d.numpy()}, tipe data : {d.dtype}")
```

```python
data = tf.constant("Wahyu", dtype=tf.string)
unicode_tensor = tf.strings.unicode_decode(data, 'UTF-8')
print(unicode_tensor.numpy())
```

```python
re_encoded = tf.strings.unicode_encode(unicode_tensor, output_encoding='UTF-8')
print(re_encoded.numpy())
```

```python
x = tf.constant(5, dtype=tf.int32)

x_float = tf.cast(x, tf.float64)      
x_string = tf.strings.as_string(x)     
string_num = tf.strings.to_number(x_string, tf.int32)

print(f"x : {x.numpy()} | {x.dtype}")
print(f"x_float : {x_float.numpy()} | {x_float.dtype}")
print(f"x_string : {x_string.numpy().decode('utf-8')} | {x_string.dtype}")
print(f"string_num : {string_num.numpy()} | {string_num.dtype}")
```

```python
x_bool = tf.constant(True, tf.bool)

bool_str = tf.strings.as_string(x_bool)   
str_bool = tf.equal(bool_str, "true")  # "true" → True, lainnya → False

print(f"x_bool : {x_bool.numpy()} | {x_bool.dtype}")
print(f"bool_str : {bool_str.numpy()} | {bool_str.dtype}")
print(f"str_bool : {str_bool.numpy()} | {str_bool.dtype}")
```

Static graph ( tensorflow 2.x ) ( auto graph tf.function() )

```python
@tf.function
def abs_value_tf(a, b):
    return a + b

result = abs_value_tf(tf.constant(3), tf.constant(5))
print(result)
```

```python
# Define matrices
matrix1 = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)  # Must be float for inverse
matrix2 = tf.constant([[5, 6], [7, 8]], dtype=tf.float32)

dot = tf.tensordot(matrix1, matrix2, axes=1)  # Standard matrix multiplication
matmul = tf.matmul(matrix1, matrix2)  # [[19, 22], [43, 50]]
transpose = tf.transpose(matrix1)     # [[1, 3], [2, 4]]
determinant = tf.linalg.det(matrix1)  # -2.0
inverse = tf.linalg.inv(matrix1)      # [[-2., 1.], [1.5, -0.5]]

print("Dot product:\n", dot.numpy())
print("\nMatrix multiplication:\n", matmul.numpy())
print("\nTranspose:\n", transpose.numpy())
print("\nDeterminant:", determinant.numpy())
print("\nInverse:\n", inverse.numpy())
```

```python
angka = tf.constant(20)
array = tf.constant([10,20,30])

angka_float =  tf.cast(angka, tf.float64) 
array_bool = tf.equal(array, 30)

print("Data : ")
print(f"angka = {angka.numpy()}, dtype = {angka.dtype}")
print(f"array = {array.numpy()}, dtype = {array.dtype}")

print("Konversi : ")
print(f"angka_float = {angka_float.numpy()}, dtype = {angka_float.dtype}")
print(f"array_bool = {array_bool.numpy()}, dtype = {array_bool.dtype}")
```
