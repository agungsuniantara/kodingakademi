```python
import tensorflow as tf
import numpy as np

data = np.array([10, 20, 30, 40, 50])
ds = tf.data.Dataset.from_tensor_slices(data)
for item in ds:
    print(item.numpy(), end = " ")
```

```python
shuffled_ds = ds.shuffle(buffer_size=5)
for item in shuffled_ds:
    print(item.numpy(), end = " ")
```

```python
batched_ds = ds.batch(2)
for batch in batched_ds:
    print(batch.numpy(), end = " ")
```

```python
def tambah_satu(x):
    return x + 1

mapped_ds = ds.map(tambah_satu)
for item in mapped_ds:
    print(item.numpy(), end = " ")
```

```python
prefetch_ds = ds.batch(2).prefetch(tf.data.AUTOTUNE)
```

```python
filtered_ds = ds.filter(lambda x: x < 40)

for item in filtered_ds:
    print(item.numpy(), end = " ")
```

```python
import tensorflow as tf

# Membuat pipeline dataset dari CSV
ds = tf.data.experimental.make_csv_dataset(
    'transaksi.csv',
    batch_size=5,
    label_name='status',
    shuffle=True,
    num_epochs=2,
    column_defaults=[0, 0.0, '', '', ''],
    select_columns=['user_id', 'amount', 'status', 'payment_type', 'region']
)

# Cetak satu batch pertama
print("Batch Pertama (A):")
for batch in ds.take(1):
    features, label = batch
    features_combined = dict(features)
    features_combined['status'] = label
    for key in features_combined:
        print(f"{key}: {features_combined[key].numpy()}")
    print()
```

```python
import tensorflow as tf

# Buat pipeline batch 1
ds_b = tf.data.experimental.make_csv_dataset(
    'transaksi.csv',
    batch_size=1,   # <-- batch_size=1 supaya filter bisa per baris!
    label_name='status',
    shuffle=True,
    num_epochs=2,
    column_defaults=[0, 0.0, '', '', ''],
    select_columns=['user_id', 'amount', 'status', 'payment_type', 'region']
)

# Filter per baris
def filter_func(features, label):
    is_credit = tf.equal(features['payment_type'][0], tf.constant('credit_card'))
    is_amount = tf.greater(features['amount'][0], 500.0)
    return tf.logical_and(is_credit, is_amount)

def map_normalize(features, label):
    features = dict(features)
    features['amount'] = features['amount'] / 1000.0
    return features, label

# Terapkan filter dan map
ds_b = ds_b.filter(filter_func).map(map_normalize).batch(5)  # batch lagi jika mau

# Cetak satu batch
print("Batch Pertama Setelah Filter dan Normalisasi (B):")
for batch in ds_b.take(1):
    features, label = batch
    features_combined = dict(features)
    features_combined['status'] = label
    for key in features_combined:
        print(f"{key}: {features_combined[key].numpy()}")
    print()
```

```python
import collections

# Pipeline batch 1 (karena reduce per data)
ds_c = tf.data.experimental.make_csv_dataset(
    'data/transaksi.csv',
    batch_size=1,
    label_name='status',
    shuffle=False,
    num_epochs=1,
    column_defaults=[0, 0.0, '', '', ''],
    select_columns=['user_id', 'amount', 'status', 'payment_type', 'region']
)

# Kumpulkan semua region dalam list
regions = []
for batch in ds_c:
    features, label = batch
    # batch_size = 1
    regions.append(features['region'].numpy()[0].decode())

# Hitung jumlah transaksi per region menggunakan Counter
region_count = collections.Counter(regions)
print("Jumlah transaksi per region (C):")
print(dict(region_count))
```
