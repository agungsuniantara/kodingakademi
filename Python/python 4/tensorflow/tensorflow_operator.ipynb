```python
import tensorflow as tf
```

Aritmatika

```python
a = tf.constant(5)
b = tf.constant(3)

add = tf.add(a, b)       
sub = tf.subtract(a, b)  
mul = tf.multiply(a, b) 
div = tf.divide(a, b)    
pow = tf.pow(a, b)       

print(f"add = {add.numpy()}")
print(f"sub = {sub.numpy()}")
print(f"mul = {mul.numpy()}")
print(f"div = {div.numpy()}")
print(f"pow = {pow.numpy()}")
```

Perbandingan

```python
vector1 = tf.constant([1,2])
vector2 = tf.constant([10,5])

greater = tf.greater(vector1, vector2)        
less = tf.less(vector1, vector2)   
equal = tf.equal(vector1, vector2)      
not_equal = tf.not_equal(vector1, vector2)    

print(f"greater = {greater.numpy()}")
print(f"less = {less.numpy()}")
print(f"equal = {equal.numpy()}")
print(f"not_equal = {not_equal.numpy()}")
```

Reduksi

```python
matrix1 = tf.constant([[1, 2], [3, 4]])
matrix2 = tf.constant([1,2,3])
matrix3 = tf.constant([True, False])

sum = tf.reduce_sum(matrix1) 
mean = tf.reduce_mean(matrix1) 
max = tf.reduce_max(matrix1)  
min = tf.reduce_min(matrix1)
prod = tf.reduce_prod(matrix1)
argmax = tf.argmax(matrix2)
argmin = tf.argmin(matrix2)
all = tf.reduce_all(matrix3)
any = tf.reduce_any(matrix3)

print(f"sum = {sum.numpy()}")
print(f"mean = {mean.numpy()}")
print(f"max = {max.numpy()}")
print(f"min = {min.numpy()}")
print(f"prod = {prod.numpy()}")
print(f"argmax = {argmax.numpy()}")
print(f"argmin = {argmin.numpy()}")
print(f"all = {all.numpy()}")
print(f"any = {any.numpy()}")
```

Logika

```python
a = tf.constant([True, False, True])
b = tf.constant([False, False, True])

logic_and = tf.logical_and(a, b)
logic_or = tf.logical_or(a, b)
logic_xor = tf.math.logical_xor(a, b)
logic_not = tf.logical_not(a)

print("Logical AND:", logic_and.numpy())
print("Logical OR:", logic_or.numpy())
print("Logical XOR:", logic_xor.numpy())
print("Logical NOT:", logic_not.numpy())
```
