```python
from datetime import datetime

date = datetime.now()
print(date.strftime("%d %m %y"))
print(date.strftime("%c"))
print(date.strftime("%A %d %B %Y"))
```

```python
tanggal = date.strftime("%A %d %B %Y")
convert = date.strptime(tanggal, "%A %d %B %Y")

print(tanggal)
print(convert)
```
