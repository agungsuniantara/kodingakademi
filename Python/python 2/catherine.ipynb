```python
def express_checkout(product_to_quantity):
    total_quantity = 0
    for product in product_to_quantity:
        total_quantity += int(len(product_to_quantity.values()))
    return total_quantity <= 8

express_checkout({'banana': 3, 'soy milk': 1, 'peanut butter': 1})
```

```python
def build_placements(shoes):
    """Return a dictionary where each key is a shoe company from shoes and each
    value is a list of placements by people wearing footwear made by that
    company. First place is at index 1 of shoes, 2nd place is at index 2, and
    so on.

    >>> build_placements(['Saucony', 'Asics', 'Asics', 'NB', 'Saucony', 'Nike', 'Asics', 'Adidas', 'Saucony', 'Asics'])
    {'Saucony': [1, 5, 9], 'Asics': [2, 3, 7, 10], 'NB': [4], 'Nike': [6], 'Adidas': [8]}
    """
    result = {}
    for i, shoe in enumerate(shoes):
        if shoe not in result:
            result[shoe] = []
        result[shoe].append(i + 1) 
    
    return result

build_placements(['Saucony', 'Asics', 'Asics', 'NB', 'Saucony', 'Nike', 'Asics', 'Adidas', 'Saucony', 'Asics'])
```
