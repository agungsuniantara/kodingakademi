```python
def jumlah(*data): # [ [10,20,30] ]
    total = 0
    for i in data:
        for z in i:
            total += z
    return total

data = []

for i in range(5):
    angka = int(input("masukan angka : "))
    data.append(angka)

hasil = jumlah(data)

rata = lambda a,b: print("rata - rata = ", a / b)
rata(hasil, len(data))
```
