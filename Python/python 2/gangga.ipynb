```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import ScalarFormatter

data = pd.read_csv("/content/data transaksi.csv")
data["jumlah_investasi"] = data["jumlah_investasi"].astype(float)

data_kabupaten = data[["nama_kabupaten_kota", "jumlah_investasi", "tahun"]]

group = data_kabupaten.groupby("nama_kabupaten_kota")["jumlah_investasi"].sum()

fig, ax = plt.subplots(figsize=(10, 8))

# Creating the horizontal bar plot
bars = plt.barh(group.index, group.values)

plt.title("Investasi Penanaman Modal Kabupaten/Kota Provinsi Jawa Barat")

# Format y-axis ticks to avoid scientific notation
ax.xaxis.set_major_formatter(ScalarFormatter(useMathText=True))
plt.ticklabel_format(style='plain', axis='x')

# Display the plot
plt.show()
```
