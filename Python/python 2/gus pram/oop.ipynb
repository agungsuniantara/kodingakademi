```python
# class default

class siswa():
    nama = "gus pram",
    alamat = "denpasar"
    
    def tampil_nama(self):
        print("halo", self.nama)

test1 = siswa()
test1.tampil_nama()
print(test1.alamat)
```

```python
# class constructor

class siswa():
    def __init__(self, a, b):
        self.nama = a
        self.alamat = b
        
    def tampil_nama(self):
        print("halo", self.nama)

test1 = siswa("gus pram", "denpasar")
test1.tampil_nama()
print(test1.alamat)
```

```python
class tunas_daud():
    def __init__(self, a, b):
        self.nama = a
        self.alamat = b
    
    def tampildata(self):
        print("nama murid : ", self.nama)
        print("alamat murid : ", self.alamat)
    
class siswa(tunas_daud):
    def __init__(self, nama, alamat, jalur):
        super().__init__(nama,alamat)
        self.jalur = jalur
    
    def tampil_jalur(self):
        print("jalur : ", self.jalur)

siswa1 = siswa("gus pram", "denpasar", "prestasi")
siswa2 = siswa("wahyu", "canggu", "tes")

siswa1.tampildata()
print(siswa1.jalur)

siswa2.tampildata()
print(siswa2.jalur)
```

```python
# class default

class karyawan:
    nama = "yani"
    alamat = "denpasar"
    
    def tampil_data(self):
        print(f"nama : {self.nama}, alamat : {self.alamat}")

karyawan1 = karyawan()
karyawan1.tampil_data()
print(karyawan1.alamat)
```

```python
# class constructor 

class karyawan:
    def __init__(self, a, b): # constructor
        self.nama = a
        self.alamat = b
    
    def tampil_data(self):
        print(f"nama : {self.nama}, alamat : {self.alamat}")
    
karyawan1 = karyawan("yani", "denpasar")
karyawan1.tampil_data()
print(karyawan1.alamat)
```
