```python
point = 0

def hello():
    print("hello world")


def data_diri(nama, umur):
    print(f"Nama : {nama}")
    print(f"Umur : {umur}")


def tambah_point():
    global point # mengubah value variabel point
    point += 1
    print(f"Point = {point}")


hello()  # function biasa
data_diri("mang agus", 50)  # function parameter
data_diri("wah asta", 1000)  # function parameter
tambah_point()
tambah_point()
```
