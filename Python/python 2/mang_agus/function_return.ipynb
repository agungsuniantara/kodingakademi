```python
point = 0


def hello(): # function void
    print("hello world")


def data_diri(nama, umur): # function parameter void
    print(f"Nama : {nama}")
    print(f"Umur : {umur}")


def tambah_point(): # function void
    global point  # mengubah value variabel point
    point += 1
    print(f"Point = {point}")


hello()  # function void
data_diri("mang agus", 50)  # function parameter void
data_diri("wah asta", 1000)  # function parameter void
tambah_point()
tambah_point()
```

```python
def hello():  # function void
    print("hello world")


def jumlah(angka1, angka2):  # function return
    hasil = angka1 + angka2
    return hasil


hello()
hasil_jumlah = jumlah(20, 30)
jumlah2 = hasil_jumlah * 2

print(hasil_jumlah)
print(jumlah2)
```
