```python
def angka(*a): # args (tuple)
    hasil = 0
    for i in a:
        for z in i:
            hasil += z 
    return hasil

list_angka = []

for i in range(5):
    input_angka = int(input("angka : "))
    list_angka.append(input_angka)

sum = angka(list_angka)

mean = lambda a,b : print("rata - rata = ", a / b)
mean(sum, len(list_angka))
```

```python
class sekolah: # class parent
    def __init__(self, a, b):
        self.nama = a
        self.alamat = b
    
    def tampil_nama(self):
        print(f"nama = {self.nama}")
    
    def tampil_alamat(self):
        print(f"alamat = {self.alamat}")

class murid(sekolah): # class child
    def __init__(self, a, b, c):
        super().__init__(a, b)
        self.kelas = c
    
    def tampil_kelas(self):
        print(f"kelas = {self.kelas}")

yani = murid("yani", "denpasar", "ipa")
yani.tampil_nama()
yani.tampil_alamat()
yani.tampil_kelas()
```
