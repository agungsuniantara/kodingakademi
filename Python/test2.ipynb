```python
COLUMN_ID = 0
COLUMN_NAME = 1
COLUMN_HIGHWAY = 2
COLUMN_LAT = 3
COLUMN_LON = 4
COLUMN_YEAR_BUILT = 5
COLUMN_LAST_MAJOR_REHAB = 6
COLUMN_LAST_MINOR_REHAB = 7
COLUMN_NUM_SPANS = 8
COLUMN_SPAN_DETAILS = 9
COLUMN_DECK_LENGTH = 10
COLUMN_LAST_INSPECTED = 11
COLUMN_BCI = 12

INDEX_BCI_YEARS = 0
INDEX_BCI_SCORES = 1
MISSING_BCI = -1.0

EARTH_RADIUS = 6371

EXAMPLE_BRIDGES = \
    [[1, 'Highway 24 Underpass at Highway 403',
      '403', 43.167233, -80.275567, '1965', '2014', '2009', 4,
      [12.0, 19.0, 21.0, 12.0], 65.0, '04/13/2012',
      [['2013', '2012', '2011', '2010', '2009', '2008', '2007',
        '2006', '2005', '2004', '2003', '2002', '2001', '2000'],
       [MISSING_BCI, 72.3, MISSING_BCI, 69.5, MISSING_BCI, 70.0, MISSING_BCI,
        70.3, MISSING_BCI, 70.5, MISSING_BCI, 70.7, 72.9, MISSING_BCI]]],

     [2, 'WEST STREET UNDERPASS',
      '403', 43.164531, -80.251582, '1963', '2014', '2007', 4,
      [12.2, 18.0, 18.0, 12.2], 61.0, '04/13/2012',
      [['2013', '2012', '2011', '2010', '2009', '2008', '2007',
        '2006', '2005', '2004', '2003', '2002', '2001', '2000'],
       [MISSING_BCI, 71.5, MISSING_BCI, 68.1, MISSING_BCI, 69.0, MISSING_BCI,
        69.4, MISSING_BCI, 69.4, MISSING_BCI, 70.3, 73.3, MISSING_BCI]]
      ],

     [3, 'STOKES RIVER BRIDGE', '6',
      45.036739, -81.33579, '1958', '2013', '', 1,
      [16.0], 18.4, '08/28/2013',
      [['2013', '2012', '2011', '2010', '2009', '2008', '2007',
        '2006', '2005', '2004', '2003', '2002', '2001', '2000'],
       [85.1, MISSING_BCI, 67.8, MISSING_BCI, 67.4, MISSING_BCI, 69.2,
        70.0, 70.5, MISSING_BCI, 75.1, MISSING_BCI, 90.1, MISSING_BCI]]
      ]
     ]
    
# example_bridges = EXAMPLE_BRIDGES
# bridge_id = [1,2]

# for bridge in example_bridges:
#     id = bridge[0]
#     if id in bridge_id:
#       bci = bridge[12][1]
#       print(bci)

# Function to find the bridge ID with the lowest recent valid BCI score
def find_worst_bci(bridges: list[list], bridge_ids: list[int]) -> int:
    worst_bridge_id = None
    worst_bci = float('inf')  # Start with the highest possible BCI for comparison

    for bridge in bridges:
        id = bridge[COLUMN_ID]
        if id in bridge_ids:
            bci_scores = bridge[COLUMN_BCI][INDEX_BCI_SCORES]

            # Loop through BCI scores in reverse to get the most recent valid one
            for bci in reversed(bci_scores):
                if bci != MISSING_BCI:
                    if bci < worst_bci:
                        worst_bci = bci
                        worst_bridge_id = id
                    break  # Stop after finding the most recent valid BCI

    return worst_bridge_id

# Example usage
example_bridges = EXAMPLE_BRIDGES
bridge_ids = [1, 2]

# Finding the bridge with the worst BCI
worst_bridge = find_worst_bci(example_bridges, bridge_ids)
print("Bridge with the worst BCI:", worst_bridge)


      
```
