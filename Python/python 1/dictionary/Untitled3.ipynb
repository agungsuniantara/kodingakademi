```python
data = {"action" : "avenger", "drama" : "mirna"}

while(True):
  print("daftar jenis buku : ")
  for i in data:
    print(f"{i} = {data[i]}")

  pilihan = input("tambah / hapus : ")

  if pilihan == 'tambah':
    jenis = input("masukan jenis buku : ")
    nama = input("masukan nama buku : ")
    data.update({jenis : nama})
  elif pilihan == 'hapus':
    jenis = input("masukan jenis buku : ")
    del data[jenis]
  else:
    print("pilihan tidak tersedia !")

  kondisi = input("apakah ingin menambahkan atau menghapus lagi ? ")

  if kondisi == "tidak":
    break
```
