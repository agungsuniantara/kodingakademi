```python
# no 1

masukan = input("masukan kata kata")
a = []
i = []
u = []
e = []
o = []

for z in masukan:
    if z == "a" or z == "A":
        a.append(z)
    elif z == "i" or z == "I":
        i.append(z)
    elif z == "u" or z == "U":
        u.append(z)
    elif z == "e" or z == "E":
        e.append(z)
    elif z == "o" or z == "O":
        o.append(z)

print("total a = ", len(a))
print("total i = ", len(i))
print("total u = ", len(u))
print("total e = ", len(e))
print("total o = ", len(o))

kata = list(masukan)

index = 0

for i in kata:
    if i == "a":
        kata[index] = "i"
    index += 1

print("".join(kata))
```

```python
# no 2

masukan = input("masukan kata = ") # aku kamu dia
kata = masukan.split() # [aku, kamu, dia] # [kamu, dia]
urutkan = [] # [aku]

while kata:
    terkecil = min(kata)
    urutkan.append(terkecil)
    kata.remove(terkecil)

print("kata sebelum diurut : ", masukan)
print("kata setelah diurut tanpa sort : ", urutkan)
```
