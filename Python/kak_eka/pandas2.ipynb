```python
import pandas as pd 

covid = pd.read_csv("covid.csv", header=None) # header membuat kolom menjadi row yang pertama
covid = covid.drop(0) # hapus row index 0
covid.columns = ["Tanggal", "ODP", "PDP", "Kasus Positif", "Kasus Sembuh", "Kasus Meninggal"] # ubah kolom
covid
```

```python
covid = covid.rename(columns={
    "Kasus Positif" : "Positif", 
    "Kasus Sembuh" : "Sembuh", 
    "Kasus Meninggal" : "Meninggal"})

covid
```

```python
covid["Positif"] = covid["Positif"].replace({
    "17" : "100", 
    "28" : "1000"})

covid
```

```python
import numpy as np

data = pd.DataFrame({
    "I" : [100, 90, np.nan, 95],
    "II" : [30, 45, 56, np.nan],
    "III" : [np.nan, 40, 80, 98]
})

data
```

```python
data.isnull().sum()
```

```python
data = data.fillna(data.mean())
data
```
