```python
import pandas as pd

df = pd.read_csv("death.csv")
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
q1 = df["Total Deaths"].quantile(0.25)
q3 = df["Total Deaths"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df_clean = df[ (df["Total Deaths"] >= lower_bound) & (df["Total Deaths"] <= upper_bound) ]
df_clean.to_csv("death_clean.csv", index = False)
```

Encoding

encoding adalah proses mengubah data kategori menjadi format numerik. 

beberapa metode encoding : 

1. one hot encoding : metode yang mengubah setiap kategori menjadi true (1) dan false (0).
2. label encoding : metode yang memberikan setiap kategori nilai integer yang unik (0,1,2,3,4).

Perbedaan one hot dengan label encoding : 
1. one hot digunakan jika datanya tidak terurut.
2. label encoding digunakan jika datanya terurut.

```python
# one hot encoding 

df_hot = pd.get_dummies(df_clean, columns=["Cause", "Type"])
df_hot
```

```python
# label encoding 

from sklearn.preprocessing import LabelEncoder

encode = LabelEncoder()
df_label = df_clean.copy()
df_label["Type"] = encode.fit_transform(df_label["Type"])
df_label
```

Data transform (transformasi data)

Data transform adalah proses mengubah data numerik ke dalam skala yang berbeda.

Tipe data transform : 

1. Normalisasi : proses merubah skala data ke rentang 0 sampai 1

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_normal = df_clean.copy()
df_normal[["Total Deaths"]] = scaler.fit_transform(df_normal[["Total Deaths"]])
df_normal
```

Data splitting 

Data splitting adalah proses membagi dataset menjadi data training dan data testing.

Sebelum melakukan data splitting, harus memilih data fitur dan data target.

```python
x = df_clean[["Cause", "Type"]] # data fitur
y = df_clean[["Total Deaths"]] # data target
```

```python
from sklearn.model_selection import train_test_split

x_encode = pd.get_dummies(x) # one hot encoding

scaler_y = MinMaxScaler()
y_norm = scaler_y.fit_transform(y).ravel() # normalisasi

# ravel mengubah data 2 dimensi menjadi 1 dimensi

x_train, x_test, y_train, y_test = train_test_split(x_encode, y_norm, test_size=0.2)

# test size = 0.2, data training 80% data testing 20%

print(f"x_train = {len(x_train)}, x_test = {len(x_test)}")
print(f"y_train = {len(y_train)}, y_test = {len(y_test)}")
```
