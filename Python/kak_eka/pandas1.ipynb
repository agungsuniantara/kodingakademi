Deklarasi series

```python
import pandas as pd 

# data series 

data = [10,20,30,40]

series1 = pd.Series(data)
series2 = pd.Series([100,200,300,400])
series3 = pd.Series([80,90,70,60], index = ["satu", "dua", "tiga", "empat"])
series4 = pd.Series({"data1" : 40, 
                     "data2" : 50, 
                     "data3" : 60})

print(series1)
print(series2)
print(series3)
print(series4)
```

Mengakses series

```python
print(series2)
print(series2[0])
print(series2[0 : 3]) # (start : batas) // slicing
```

```python
print(series3)
print(series3["satu"])
print(series3["satu" : "dua"]) # slicing 
print(series3.loc["satu" : "dua"]) # loc -- batasnya ikut
print(series3.iloc[0 : 3]) # iloc -- batasnya tidak ikut
```

Conditional series

```python
data = pd.Series([1500, 1250, 2000, 3000, 4000])
print(data)
print(data > 1500)
print(data[data > 1500])
print(data[(data > 1500) & (data < 4000)])
```

```python
hasil = data[data > 1500]
hasil = hasil + 100
hasil
```

```python
rata = data.mean()
total = data.sum()
print(rata)
print(total)
print(data[data > rata])
```

```python
data1 = pd.Series([100,200,300,400])
print(data1)

data1 = data1.astype(float) # astype mengganti tipe data
print(data1)
```

Mengubah series

```python
data3 = pd.Series([200,300,400])
print(data3)
data3[0] = 1000
print(data3)
```

```python
data3[data3 < 1000] = 600
print(data3)
```

Dataframe

```python
data = [
    [10, 20],
    [30, 40],
    [50, 60]
]

df = pd.DataFrame(data)
df.columns = ["kolom 1", "kolom 2"]
df.index = ["satu", "dua", "tiga"]
df
```

```python
data = {
    "kolom 1" : [10, 20, 30, 100, 200],
    "kolom 2" : [20, 30, 90, 300, 400],
    "kolom 3" : [20, 20, 30, 20, 20],
    "kolom 4" : [20, 20, 30, 20, 20]
}

df = pd.DataFrame(data)
df
```

```python
print(df.iloc[0])
print(df.iloc[0 : 3])
print(df.iloc[0 : 3, 0 : 2])
```
