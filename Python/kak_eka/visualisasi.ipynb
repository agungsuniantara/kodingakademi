```python
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np
```

```python
plt.plot([1,2,3,4], [10,5,30,20], label = "Plot 1")
plt.plot([1,2,3,4], [15,20,5,10], label = "Plot 2")
plt.title("Title")
plt.xlabel("Sumbu x")
plt.ylabel("Sumbu y")
plt.xlim(0,5)
plt.ylim(0,50)
plt.xticks([1, 2, 3, 4])
plt.yticks([0 ,10, 20, 30, 40])
plt.legend(loc = "upper left")
plt.show()
```

```python
data = pd.read_csv("penyakit_jantung.csv")[["age", "sex", "trestbps", "chol"]]
age = data.groupby("age")["age"].count()
age.head()
```

Line plot

```python
plt.plot(age.index, age.values, color = "red", marker = "o", ls = "--")
plt.title("line plot")
plt.xlabel("age")
plt.ylabel("count")
plt.ylim(0,20)
plt.yticks([0,5,10,15,20])
plt.show()
```

Bar plot

```python
plt.bar(age.index, age.values)
plt.title("Bar plot")
plt.xlabel("age")
plt.ylabel("count")
plt.ylim(0,20)
plt.yticks([0,5,10,15,20])
plt.show()
```

Scatter plot

```python
plt.scatter(age.index, age.values)
plt.title("Scatter plot")
plt.xlabel("age")
plt.ylabel("count")
plt.ylim(0,20)
plt.yticks([0,5,10,15,20])
plt.show()
```

Histogram

```python
plt.hist(age, bins = 10)
plt.title("Histogram")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```

Pie plot

```python
gender = data.groupby("sex")["sex"].count()

plt.pie(gender.values, labels=gender.index, autopct='%1.1f%%')
plt.title("Pie plot")
plt.legend()
plt.show()
```

Sub plot

```python
plt.subplot(221) # (baris, kolom, letak plot)
plt.plot(age.index, age.values)
plt.title("line plot")
plt.xlabel("age")
plt.ylabel("count")

plt.subplot(222) # (baris, kolom, letak plot)
plt.bar(age.index, age.values)
plt.title("bar plot")
plt.xlabel("age")
plt.ylabel("count")

plt.subplot(223) # (baris, kolom, letak plot)
plt.bar(age.index, age.values)
plt.title("bar plot")
plt.xlabel("age")
plt.ylabel("count")

plt.subplot(224) # (baris, kolom, letak plot)
plt.bar(age.index, age.values)
plt.title("bar plot")
plt.xlabel("age")
plt.ylabel("count")

plt.tight_layout()
plt.show()
```

Berbasis objek

```python
fig = plt.figure()
axes = fig.add_axes([0,0,0.4,0.9]) # (left, bottom, width, height)
axes.plot(age.index, age.values, color = "orange")
axes.set_xlabel("Age")
axes.set_ylabel("count")
plt.show()
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.7,0,1.3,1.2]) # (left, bottom, width, height)
axes1.plot(age.index, age.values, color = "red")
axes1.set_xlabel("Age")
axes1.set_ylabel("count")
axes1.set_title("Age axes 1")

axes2 = fig.add_axes([0.2, 0.8, 0.3, 0.3]) # (left, bottom, width, height)
axes2.plot(age.index, age.values, color = "blue")
axes2.set_xlabel("Age")
axes2.set_ylabel("count")
axes2.set_title("Age axes 2")

plt.show()
```

```python
fig, axes = plt.subplots(1,2) # (baris, kolom)

# kolom 1
axes[0].plot(age.index, age.values, color = "red")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("count")
axes[0].set_title("age kolom 1")

# kolom 2
axes[1].plot(age.index, age.values, color = "red")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("count")
axes[1].set_title("age kolom 1")

fig.tight_layout()
plt.show()
```
