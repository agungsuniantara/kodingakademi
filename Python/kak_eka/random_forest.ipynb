Random forest 

- random forest regressor 
    digunakan ketika outputnya adalah nilai kontinu

- random forest classifier
    digunakan ketika outputnya berupa kategori

kelebihan random forest : 
    1. akurasi lebih bagus 
    2. mengurangi resiko overfitting 
    3. Robust terhadap data hilang atau outlier

Random forest regressor

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler 
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv("death_clean.csv")

df.head()
```

```python
x = df[["Cause", "Type"]]
y = df[["Total Deaths"]]

x = pd.get_dummies(x, columns=["Cause", "Type"])

scaler = MinMaxScaler()
y = scaler.fit_transform(y).ravel()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
y_pred
```

```python
y_test_orig = scaler.inverse_transform(y_test.reshape(-1, 1)).ravel()
y_pred_orig = scaler.inverse_transform(y_pred.reshape(-1, 1)).ravel() 

results_df = pd.DataFrame({
    "Actual" : y_test_orig,
    "Predict" : y_pred_orig
})

results_df
```

```python
model.score(x_test, y_test)
```

Random forest classifier

```python
from sklearn.ensemble import RandomForestClassifier 
from sklearn.preprocessing import LabelEncoder 

df = pd.read_csv("penyakit_jantung.csv")

df.head()
```

```python
X = df.drop(columns=["target"])
Y = df[["target"]] 

encoder = LabelEncoder()
X["sex"] = encoder.fit_transform(X["sex"])

kolom = df.drop(columns=["target", "sex"]).columns
print(kolom)

scaler_jantung = MinMaxScaler()
X[['age', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 
   'thalach', 'exang','oldpeak', 'slope', 'ca', 'thal']] = scaler_jantung.fit_transform(X[['age', 'cp', 'trestbps', 'chol', 'fbs', 
                                                                                           'restecg', 'thalach', 'exang','oldpeak', 'slope', 'ca', 'thal']])

X
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)

model.score(x_test, y_test)
```

```python
results = pd.DataFrame({
    "Actual" : y_test["target"].values, 
    "Predict" : y_pred
})

results
```
