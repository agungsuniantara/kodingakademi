```python
import pandas as pd 

df = pd.read_csv("death.csv")
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
q1 = df["Total Deaths"].quantile(0.25)
q3 = df["Total Deaths"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr 
upper_bound = q3 + 1.5 * iqr

print(f"q1 = {q1}")
print(f"q3 = {q3}")
print(f"lower bound = {lower_bound}")
print(f"upper bound = {upper_bound}")

outlier = df[(df["Total Deaths"] < lower_bound) | (df["Total Deaths"] > upper_bound)]
outlier
```

```python
df_clean = df[(df["Total Deaths"] >= lower_bound) & (df["Total Deaths"] <= upper_bound)]
df_clean.to_csv("death_clean.csv", index=False)
```

Encoding

```python
df_clean
```

```python
# One hot encoding

df_hot = pd.get_dummies(df_clean.copy(), columns=["Cause", "Type"])
df_hot
```

```python
# label encoding

from sklearn.preprocessing import LabelEncoder

Label_encoder = LabelEncoder()
df_label = df_clean.copy() 
df_label["Type"] = Label_encoder.fit_transform(df_label["Type"])
df_label
```

Data transform

```python
# normalisasi 

from sklearn.preprocessing import MinMaxScaler 

scaler = MinMaxScaler()
df_normal = df_clean.copy()
df_normal[["Total Deaths"]] = scaler.fit_transform(df_normal[["Total Deaths"]])
df_normal
```

Splitting data

```python
x = df_clean[["Cause", "Type"]] # fitur
y = df_clean[["Total Deaths"]] # target
```

```python
from sklearn.model_selection import train_test_split

scale = MinMaxScaler()

x_encoded = pd.get_dummies(x)
y_norm = scale.fit_transform(y).ravel()

x_train, x_test, y_train, y_test = train_test_split(x_encoded, y_norm, test_size=0.2)
print(f"x_train = {len(x_train)}, x_test = {len(x_test)}")
print(f"y_train = {len(y_train)}, y_test = {len(y_test)}")
```
