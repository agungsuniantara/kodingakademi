```python
import numpy as np

a = np.array([1,2,3,4]) # 1 dimensi
print(a)
print(a[0])
print(a[1])
```

```python
b = np.array([  # 2 dimensi
    [1,2,3],
    [5,6,7],
    [8,9,10]
])
print(b)
print(b[0])
print(b[0][1])
```

```python
c = np.array([ # array 3 dimensi
    [
        [1,2,3],
        [4,5,6]
    ],
    [
        [4,5,6],
        [7,8,9]
    ]
])

print(c)
print(c[0])
print(c[0][0])
print(c[0][0][2])
```

```python
b = np.array([  # 2 dimensi
    [1,2,3],
    [5,6,7],
    [8,9,10]
])

print(b.shape)
print(b.ndim)
print(b.size)
```
