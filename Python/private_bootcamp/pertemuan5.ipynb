SLICING

```python
data = [10,20,30,40,50]

print(data[0 : 3]) # [awal : batas]
print(data[0 : 4 : 2]) # [awal : batas : langkah]
print(data[0 : ]) # [awal, batas, langkah]
print(data[-1])
print(data[-5 : -1]) # [awal : batas]
print(data[-5 : ]) # [awal : sampai habis]
print(data[-5 : -1 : 2]) # [awal : batas : langkah]
print(data[::-1])
```
