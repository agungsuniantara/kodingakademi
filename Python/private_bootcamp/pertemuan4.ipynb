WHILE LOOPS

```python
boolean = True 

while boolean == True:
    print("putaran while")
    kondisi = False
    boolean = kondisi
```

```python
nama = "agung" # wayan

while nama == "agung" :
    kondisi = input("masukan nama : ")
    nama = kondisi
    
```

```python
while True:
    print("hello world")
    break
    
```

```python
while True:
    print("menu : ")
    print("1. tampil hello world")
    print("2. keluar")
    
    kondisi = input("pilih menu : ")
    
    if kondisi == "1":
        print("hello world")
    elif kondisi == "2":
        print("Keluar sistem")
        break
```

```python
angka = int(input("angka : "))
hasil = 0

for i in range(angka, 31, 2):
    hasil += i

print(hasil)
```

FOR DI DALAM FOR

```python
for i in range(3): 
    for z in range(5): 
        if i >= z : 
            print("* ", end = "")
    print("")
```
