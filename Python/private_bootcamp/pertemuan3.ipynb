OPERATOR ARITMATIKA

```python
num1 = int(input("masukan angka 1 : "))
num2 = int(input("masukan angka 2 : "))

tambah = num1 + num2 
kurang = num1 - num2
bagi = num1 / num2 
kali = num1 * num2 
sisa_bagi = num1 % num2 
pangkat = num1 ** num2

print(f"Tambah : {tambah}")
print(f"Kurang : {kurang}")
print(f"Bagi : {bagi}")
print(f"Kali : {kali}")
print(f"Sisa bagi : {sisa_bagi}")
print(f"Pangkat : {pangkat}")
```

OPERATOR PENUGASAN

```python
angka = 0
print(angka)
angka += 10 # angka = angka + 10
print(angka)
angka += 2 # angka = angka + 2
print(angka)
angka -= 1
print(angka)
angka *= 2
print(angka)
angka /= 2
print(angka)
```

OPERATOR PERBANDINGAN

```python
angka1 = 50
angka2 = 100

kondisi1 = angka1 > angka2
kondisi2 = angka1 < angka2 
kondisi3 = angka1 == angka2
kondisi4 = angka1 >= angka2
kondisi5 = angka1 <= angka2
kondisi6 = angka1 != angka2

print(kondisi1)
print(kondisi2)
print(kondisi3)
print(kondisi4)
print(kondisi5)
print(kondisi6)
```

OPERATOR LOGIKA

```python
angka1 = 50
angka2 = 100
angka3 = 40 
angka4 = 90

# opertator and : 
# true and true = true
# false and true = false
# false and false = false

# opertator and : 
# true or true = true
# false or true = true
# false or false = false

kondisi1 = angka1 > angka2 and angka3 < angka4
kondisi2 = angka1 > angka2 or angka3 < angka4
print(kondisi1)
print(kondisi2)
```

IN , NOT IN, IS

```python
data = [1,2,3,4,5]
angka = 1
a = angka 

print(angka in data)
print(angka not in data)
print(angka is a)
```

```python
angka1 = 10
angka2 = 2

print(angka1 % angka2)
print(angka1 % angka2 == 0)
print(angka1 % angka2 == 1)
```

ELIF STATEMENT

```python
username = "agung"

if username == "admin":
    print(f"halo {username}")
elif username == "wahyu":
    print(f"halo {username}")
elif username == "agung":
    print(f"halo {username}")
else:
    print("bukan admin")
```

MULTIPLE STATEMENT

```python
username = "wahyu"
password = 222

if username == "admin" and password == 123:
    print(f"halo {username}")
elif username == "wahyu" or password == 123:
    print(f"halo {username}")
```
