```python
data = []

for i in json["restaurants"]:
    if isinstance(i , list):
        for lapisan1 in i :
            for lapisan2 in lapisan1.values():
                for key,value in lapisan2.items():
                    if key == "name":
                        data.append(value)

df = pd.DataFrame(data, columns=["Restaurant"])
df.to_excel("restoran.xlsx")
```

STATISTIK

```python
stat = pd.read_csv("penyakit_jantung.csv")
thalach = stat["thalach"]
thalach.head()
```

```python
import numpy as np
from scipy import stats 

print(thalach.mean())
print(thalach.median())
print(stats.mode(thalach))
print(np.var(thalach))
print(np.std(thalach))
print(stats.skew(thalach))
```

```python
data = thalach.sort_values()

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df_clean = data[(data >= lower) & (data <= upper)]
df_clean
```
