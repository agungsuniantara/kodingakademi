TIPE DATA

```python
nama = "ivan" # tipe data string
huruf = 'c' # char (bisa menyimpan 1 karakter)
angka = 10 # tipe data int / integer (bilangan bulat)
koma = 10.5 # tipe data float (bilangan koma)
kondisi1 = True # tipe data boolean
kondisi2 = False # tipe data boolean

print("Tipe data : ")
print(nama)
print(huruf)
print(angka)
print(koma)
print(kondisi1)
print(kondisi2)
```

```python
print(type(nama))
print(type(huruf))
print(type(angka))
print(type(koma))
print(type(kondisi1))
```

Konversi

```python
angka_int = 10
print(angka_int)

angka_str = "100"
print(type(angka_str))

angka_float = float(angka_int)
print(angka_float)

angka_baru = int(angka_str)
print(type(angka_baru))
```
