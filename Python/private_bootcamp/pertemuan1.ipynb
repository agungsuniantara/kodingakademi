```python
nama = "ivan" # tipe data string
angka = 10 # tipe data int / integer (bilangan bulat)
koma = 10.5 # tipe data float (bilangan koma)
kondisi1 = True # tipe data boolean
kondisi2 = False # tipe data boolean

print("Tipe data : ")
print(nama)
print(angka)
print(koma)
print(kondisi1)
print(kondisi2)
```
