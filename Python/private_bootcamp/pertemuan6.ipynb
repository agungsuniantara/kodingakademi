DICTIONARY

```python
data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}
data2 = dict(nama = "agung", alamat = "denpasar")

print(data1["nama"])
print(data2["nama"])
```

```python
# ubah data dictionary

data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}

data1["nama"] = "wahyu"
print(data1)
```

```python
# hapus data dictionary

data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}

del data1["nama"]
print(data1)
```

```python
data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}

hapus = data1.pop("nama")
print(data1)
print(hapus)
```

```python
# menambahkan data ke dictionary

data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}

data1.update({"password" : "admin"})
print(data1)
```

```python
data = [1,2,3,4]
data1 = {
    "nama" : "agung",
    "alamat" : "denpasar"
}
print(len(data))
print(len(data1))
```

```python
data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

kondisi = input("tambah / hapus : ")

if kondisi == "tambah":
    jenis = input("masukan jenis buku :")
    buku = input("masukan buku : ")
    data.update({jenis : buku})
    print(data)
elif kondisi == "hapus":
    jenis = input("masukan jenis buku :")
    del data[jenis]
    print(data)
```

DICTIONARY FUNCTION

```python
# copy

data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

data2 = data.copy()
print(data2)
```

```python
# clear

data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

data.clear()
print(data)
```

```python
# fromkey

nama_hewan = {"Sapi", "Ayam", "Kambing"}
hewan = dict.fromkeys(nama_hewan, "hewan")
print(hewan)
```

```python
# get

data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

action = data.get("action")
print(action)
```

```python
# items, keys, values

data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

print(data.keys())
print(data.values())
print(data.items())
```

ASSIGNMENT

```python
data = {
    "1" : "Ayam", # hewan
    "2" : "Sapi",
    "3" : "Mawar",
    "4" : "Kambing",
    "5" : "Lily"
}

for i in data.keys(): 
    angka = int(i) 
    if angka % 2 != 0: 
        data[i] = "hewan"
        
print(data)
```
