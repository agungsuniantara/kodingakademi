```python
# Program menghitung biaya perjalanan berdasarkan jarak tempuh dan konsumsi bahan bakar

# Input dari pengguna
jarak = int(input("Masukkan jarak perjalanan (km): "))
konsumsi_bahan_bakar = int(input("Masukkan konsumsi bahan bakar (liter/km): "))
harga_per_liter = int(input("Masukkan harga bahan bakar per liter: "))

# Menghitung total biaya bahan bakar
total_biaya = jarak * konsumsi_bahan_bakar * harga_per_liter

# Output hasil
print("Jarak perjalanan :", jarak)
print("Konsumsi bahan bakar :", konsumsi_bahan_bakar)
print("Harga per liter :", harga_per_liter)
print("Total biaya bahan bakar:", total_biaya)
```
