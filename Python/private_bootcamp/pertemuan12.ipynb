```python
data = {
    "GDP" : [100,200,300,400],
    "Population" : [1000,2000,3000,4000],
    "HDI" : [10,20,30,40],
    "Surface" : [20,30,40,50]
}

df = pd.DataFrame(data)
df
```

```python
tambah = [30,90,40,60]

df["Index"] = tambah
df
```

```python
df["Total"] = df["HDI"] * df["Surface"]
df
```

```python
df = df.rename(
    columns={
        "Population" : "Populasi",
        "Index" : "Harga"
    }
)

df
```

```python
df["GDP"] = df["GDP"].replace(400, 1000)
df["HDI"] = df["HDI"].replace([10,20], [100,200])
df
```
