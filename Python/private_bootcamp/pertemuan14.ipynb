```python
import pandas as pd 
import seaborn as sns 

data = pd.read_csv("bali_tourist.csv")
data
```

```python
# no 1 & 2

bulan = data.loc[0:, "January" : "December"].columns
tourist = data.groupby("year")[bulan].sum() 
tourist
```

```python
total_tourist = tourist.sum(axis=1)
mean_tourist = tourist.mean(axis=1)
max_2018 = tourist.loc[2018].idxmax()
max_2019 = tourist.loc[2019].idxmax()
min_2018 = tourist.loc[2018].idxmin()
min_2019 = tourist.loc[2019].idxmin()
```

```python
# no 5

area = data.groupby(["year", "area"])[bulan].sum()
area_max_2018 = area.loc[2018].sum(axis = 1).idxmax()
area_max_2019 = area.loc[2019].sum(axis = 1).idxmax()

print(area_max_2019)
print(area_max_2018)
```
