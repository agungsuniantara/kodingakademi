FUNCTION RECURSIVE

```python
def faktorial(n): # n = 0
    if n == 0:
        return 1
    else:
        return n * faktorial(n-1) # 4 3 2 1 0

nilai_faktorial = faktorial(5) # 5 * 4 * 3 * 2 * 1 = 120
print("Nilai faktorial = ", nilai_faktorial)
```

FUNCTION ARGS & KWARGS

```python
def tampil(*a): # args
    print(a)
    print(a[0])
    print(a[1])
    print(a[2])

tampil("agung", "wahyu", "fitri")
```

```python
def tampil2(**a): # kwargs
    print(a)
    print(a["nama"])
    print(a["alamat"])

tampil2(nama = "agung", alamat = "denpasar")
```

```python
def penjumlahan(a,b): # angka1, angka2
    hasil = a + b
    return hasil

def pengurangan(a,b):
    hasil = a - b
    return hasil

def perkalian(a,b):
    hasil = a * b
    return hasil

def pembagian(a,b):
    hasil = a / b
    return hasil

while True:
    print("1. Penjumlahan")
    print("2. Pengurangan")
    print("3. Perkalian")
    print("4. Pembagian")
    
    pilihan = int(input("masukan pilihan : "))
    
    if pilihan == 1:
        angka1 = int(input("masukan angka 1 : "))
        angka2 = int(input("masukan angka 2 : "))
        jumlah = penjumlahan(angka1, angka2)
        print(jumlah)
    elif pilihan == 2:
        angka1 = int(input("masukan angka 1 : "))
        angka2 = int(input("masukan angka 2 : "))
        jumlah = pengurangan(angka1, angka2)
        print(jumlah)
    elif pilihan == 3:
        angka1 = int(input("masukan angka 1 : "))
        angka2 = int(input("masukan angka 2 : "))
        jumlah = perkalian(angka1, angka2)
        print(jumlah)
    elif pilihan == 4:
        angka1 = int(input("masukan angka 1 : "))
        angka2 = int(input("masukan angka 2 : "))
        jumlah = pembagian(angka1, angka2)
        print(jumlah)
    else:
        print("pilihan tidak tepat")
        
    
    kondisi = input("masih ingin memilih ? ")
    
    if kondisi == "tidak":
        break
```
