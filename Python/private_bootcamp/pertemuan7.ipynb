FUNCTION

```python
def tambah(): # function void
    a = 20
    b = 30
    print(a + b)

def persegi_panjang(a, b): # function parameter
    hasil = a + b
    print(hasil)

def kali(a, b): # function return
    hasil = a * b 
    return hasil

tambah()
tambah()

persegi_panjang(10,20)
persegi_panjang(30,50)

total1 = kali(40,30) + 5000
print(total1)
total2 = kali(60,70) + 10000
print(total2)
```

```python
angka = 100

def tambah():
    global angka
    angka += 2
    print(angka)

tambah()
```

```python
def penjumlahan():
    angka1 = int(input("masukan angka 1 : "))
    angka2 = int(input("masukan angka 2 : "))
    hasil = angka1 + angka2
    print(hasil)

def pengurangan():
    angka1 = int(input("masukan angka 1 : "))
    angka2 = int(input("masukan angka 2 : "))
    hasil = angka1 - angka2
    print(hasil)

while True:
    print("1. Penjumlahan")
    
    pilihan = int(input("masukan pilihan : "))
    
    if pilihan == 1:
        penjumlahan()
    
    kondisi = input("masih ingin memilih ? ")
    
    if kondisi == "tidak":
        break
```
