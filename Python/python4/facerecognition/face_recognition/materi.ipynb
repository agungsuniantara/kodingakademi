```python
import cv2
import face_recognition

# Load gambar
image = face_recognition.load_image_file("smile.jpg")

# Mendeteksi lokasi wajah dalam gambar
face_locations = face_recognition.face_locations(image)

# Menampilkan jumlah wajah yang terdeteksi
print(f"Jumlah wajah terdeteksi: {len(face_locations)}")

# Menampilkan gambar dengan kotak wajah
image_cv = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
for (top, right, bottom, left) in face_locations:
    cv2.rectangle(image_cv, (left, top), (right, bottom), (0, 255, 0), 2)

cv2.namedWindow("image")
cv2.imshow("image", image_cv)
cv2.waitKey(0)
cv2.destroyWindow('image')
cv2.waitKey(1)
```

```python
import cv2
import face_recognition
import numpy as np

# Load gambar wajah yang dikenal (misalnya, foto diri sendiri)
known_image = face_recognition.load_image_file("smile.jpg")
known_encoding = face_recognition.face_encodings(known_image)[0]

# Simpan encoding dan nama
known_encodings = [known_encoding]
known_names = ["Nama Anda"]

# Inisialisasi webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("⚠ Gagal menangkap frame!")
        break

    # Konversi frame ke RGB (karena OpenCV defaultnya BGR)
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Deteksi wajah dalam frame
    face_locations = face_recognition.face_locations(rgb_frame)
    face_encodings = face_recognition.face_encodings(rgb_frame, face_locations)

    # Bandingkan dengan wajah yang dikenal
    for (top, right, bottom, left), face_encoding in zip(face_locations, face_encodings):
        matches = face_recognition.compare_faces(known_encodings, face_encoding)
        name = "Unknown"

        # Hitung jarak wajah (semakin kecil, semakin mirip)
        face_distances = face_recognition.face_distance(known_encodings, face_encoding)
        best_match_index = np.argmin(face_distances)

        if matches[best_match_index] and face_distances[best_match_index] < 0.6:
            name = known_names[best_match_index]

        # Gambar kotak di sekitar wajah
        cv2.rectangle(frame, (left, top), (right, bottom), (0, 255, 0), 2)

        # Tampilkan nama di bawah kotak
        cv2.putText(frame, name, (left, bottom + 20), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

    # Tampilkan frame
    cv2.imshow("Face Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        cv2.destroyAllWindows()
        cv2.waitKey(1)
        break

# Lepaskan kamera dan tutup jendela
cap.release()
```
