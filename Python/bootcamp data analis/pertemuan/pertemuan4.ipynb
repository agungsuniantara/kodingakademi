```python
# no 2 (cara 1)

data = {
    "action" : "",
    "romance" : ""
}

kondisi = input("tambah / hapus : ")

if kondisi == "tambah":
    jenis = input("jenis buku : ")
    if jenis not in data:
        print("jenis buku tidak ada")
    else:
        buku = input("nama buku : ")
        data[jenis] = buku
elif kondisi == "hapus":
    jenis = input("jenis buku : ")
    if jenis not in data:
        print("jenis buku tidak ada")
    else:
        del data[jenis]
else:
    print("konidi salah !")

print(data)
```

```python
# no 2 (cara 2)

data = {
    "action" : [],
    "romance" : []
}

while(True):
    kondisi = input("tambah / hapus : ")

    if kondisi == "tambah":
        jenis = input("jenis buku : ")
        if jenis not in data:
            print("jenis buku tidak ada")
        else:
            buku = input("nama buku : ")
            data[jenis].append(buku)
    elif kondisi == "hapus":
        jenis = input("jenis buku : ")
        if jenis not in data:
            print("jenis buku tidak ada")
        else:
            buku = input("nama buku : ")
            if buku in data[jenis]:
                data[jenis].remove(buku)
            else:
                print("buku tidak ada")
    else:
        print("konidi salah !")
    
    tanya = input("apakah ingin tambah / hapus lagi ? ")
    
    if tanya == "tidak":
        break

print(data)
```

ASSIGNMENT DICTIONARY FUNCTION


```python
# no 1

data = {
    1 : "ayam",
    2 : "sapi", 
    3 : "mawar",
    4 : "kambing", 
    5 : "lily"
}

data2 = data.copy()

del data[3]
del data[5]

print("sebelum : ")
for key, value in data2.items():
    print(f"{key} = {value}")

print("sesudah dihapus : ")
for key, value in data.items():
    print(f"{key} = {value}")
```

```python
# no 2

data = {
    1 : "ayam",
    2 : "sapi", 
    3 : "mawar",
    4 : "kambing", 
    5 : "lily"
}

for a, b in data.items():
    if a % 2 != 0 :
        data[a] = "hewan"

print(data)
```
