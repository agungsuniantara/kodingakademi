LIBRARY

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import silhouette_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
import matplotlib.pyplot as plt 
```

DATA

```python
df = pd.read_csv("/Applications/MAMP/htdocs/local/kodingakademi/Python/bootcamp data analis/pertemuan/ML/country_data.csv")
df
```

PREPROCESSING

```python
x = df.iloc[0:, 1:9]
y = df[["gdpp"]]

scale_x = MinMaxScaler() # scaler fitur
x_norm = scale_x.fit_transform(x)

scale_y = MinMaxScaler() # scaler target
y_norm = scale_y.fit_transform(y).ravel()

x_train, x_test, y_train, y_test = train_test_split(x_norm, y_norm, test_size=0.3, random_state=42)

print(f"x_train = {len(x_train)}, x_test = {len(x_test)}")
print(f"y_train = {len(y_train)}, y_test = {len(y_test)}")
```

DECISION TREE

```python
model = DecisionTreeRegressor()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

score = model.score(x_test, y_test)
score
```

RANDOM FOREST

```python
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

score = model.score(x_test, y_test)
score
```

KNN

```python
knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(x_train, y_train)

y_pred = knn.predict(x_test)

score = model.score(x_test, y_test)
score
```
