```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

csv = pd.read_csv('data/penyakit_jantung.csv')
source = csv[["sex","age","trestbps","chol"]]
source.head()
```

```python
sns.lineplot(x = "age", y = "trestbps", data = source, errorbar=None)
plt.title("Lineplot")
plt.show()
```

```python
sns.histplot(x = "trestbps", data = source, bins = 10)
plt.title("Histplot trestbps")
plt.show()
```

```python
sns.kdeplot(x = "trestbps", data = source, fill=True)
plt.title('KDE plot trestbps')
plt.show()
```

```python
print(source.trestbps.mean())
print(source.trestbps.max())
print(source.trestbps.min())
```

```python
sns.boxplot(x = "trestbps", data = source)
plt.title('Boxplot trestbps')
plt.show()
```

```python
sns.violinplot(x = "trestbps", data = source)
plt.title('Violinplot trestbps')
plt.show()
```

```python
sns.scatterplot(x = "trestbps", y = "chol", data = source)
plt.title('Scatterplot trestbps')
plt.show()
```

```python
sns.pairplot(data = source)
plt.show()
```

```python
viz = sns.PairGrid(data = source)
viz.map_upper(sns.scatterplot)
viz.map_diag(sns.kdeplot, fill = True)
viz.map_lower(sns.scatterplot)
```

```python
fig, axes = plt.subplots(1, 2) # (baris, kolom)

# kolom 1
sns.histplot(x = "trestbps", data = source, ax = axes[0])
axes[0].set_xlabel('Sumbu x')
axes[0].set_title('Histplot')

# kolom 2
sns.kdeplot(x = "trestbps", data = source, ax = axes[1], fill = True)
axes[1].set_xlabel('Sumbu x')
axes[1].set_title('KDE plot')

fig.tight_layout()
plt.show()
```

```python
sns.barplot(x = "sex", y = "chol", data = source)
plt.title('Bar plot')
plt.show()
```

```python
sns.countplot(x="sex", data = source)
plt.title('Count plot')
plt.show()
```

```python
sns.boxplot(x = "sex", y = "trestbps", data = source)
plt.title('Box plot')
plt.show()
```

```python
sns.violinplot(x = "sex", y = "trestbps", data = source)
plt.title('Violinplot')
plt.show()
```

```python
sns.stripplot(x = "sex", y = "trestbps", data = source)
plt.title('Strip plot')
plt.show()
```

```python
pivot = source.pivot_table(index="sex", columns="age", values="chol")
pivot = pivot.fillna(0)
pivot
```

```python
sns.heatmap(pivot)
plt.title('Heatmap Plot')
plt.show()
```

```python
source = csv[["sex","age","trestbps","chol","cp"]]

sns.barplot(x = "sex", y = "chol", hue = "cp", data = source)
plt.title('Barplot hue')
plt.show()
```
