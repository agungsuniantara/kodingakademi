```python
import pandas as pd 
import numpy as np 
from scipy import stats as st 

data = pd.read_csv("data/tourist.csv")
data
```

```python
# NO 1

data.dtypes
```

```python
data2018 = data.loc[data["year"] == 2018]
total_2018 = data2018.iloc[0:, 3:].sum().sum()
total_2018
```

```python
data2019 = data.loc[data["year"] == 2019]
total_2019 = data2019.iloc[0:, 3:].sum().sum()
total_2019
```

```python
# NO 2

rata2018 = data2018.iloc[0:, 3:].sum().mean()
rata2018
```

```python
rata2019 = data2019.iloc[0:, 3:].sum().mean()
rata2019
```

```python
# NO 3

bulan_2018 = data2018.iloc[0:, 3:].sum()
bulan_2019 = data2019.iloc[0:, 3:].sum()

print("max 2018 : ")
print(bulan_2018.idxmax())
print(bulan_2018.max())
print("max 2019 : ")
print(bulan_2019.idxmax())
print(bulan_2019.max())
```

```python
# NO 4

bulan_2018 = data2018.iloc[0:, 3:].sum()
bulan_2019 = data2019.iloc[0:, 3:].sum()

print("min 2018 : ")
print(bulan_2018.idxmin())
print(bulan_2018.min())
print("min 2019 : ")
print(bulan_2019.idxmin())
print(bulan_2019.min())
```

```python
# NO 5

bulan = data.iloc[:, 3:].columns

area2018 = data2018.groupby("area")[bulan].sum()
area2019 = data2019.groupby("area")[bulan].sum()

total2018 = area2018.sum(axis = 1)
total2019 = area2019.sum(axis = 1)

terbesar2018 = total2018.idxmax()
terbesar2019 = total2019.idxmax()

print(terbesar2018)
print(terbesar2019)
```

```python
# NO 6

test = data.groupby("year")[bulan].sum()
test["Max"] = [bulan_2018.idxmax(),bulan_2019.idxmax()]
test["Min"] = [bulan_2018.idxmin(),bulan_2019.idxmin()]
test["Total Setahun"] = [total_2018, total_2019]
test["Rata setahun"] = [rata2018, rata2019]
test
```

```python
data = pd.read_csv("data/employee.csv")
data
```

```python
data.filter(['name', 'department'])
```

```python
data.filter([0,1,2], axis = 0)
```

```python
data[data.department == 'Sales']
```

```python
data[data.department.isin(['Sales', 'Finance'])]
```

```python
data[data.performance_score >= 700]
```

```python
data = data.dropna()
data
```

```python
data["age"] = data.age.fillna(data.age.mean())
data
```

```python
upper = data["performance_score"].mean() + 3 * data["performance_score"].std()
lower = data["performance_score"].mean() - 3 * data["performance_score"].std()
data = data[(data["performance_score"] < upper) & (data["performance_score"] > lower)]
data
```

```python
upper = data["performance_score"].quantile(.99)
lower = data["performance_score"].quantile(.01)
data = data[(data["performance_score"] < upper) & (data["performance_score"] > lower)]
data
```
