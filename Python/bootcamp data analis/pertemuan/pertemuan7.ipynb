LOC


```python
data = ["scratch", "roblox", "unity"]
series = pd.Series(data, index = ["satu", "dua", "tiga"])

print(series.loc["satu"])
print(series.loc["satu":"tiga"])
```

ILOC


```python
data = ["scratch", "roblox", "unity"]
series = pd.Series(data, index = ["satu", "dua", "tiga"])

print(series.iloc[0])
print(series.iloc[0:3])
```

LOC :

- Menggunakan index yang dibuat
- slicing awal dan batas menggunakan index sebenernya

ILOC :

- Menggunakan index default (dimulai dari 0)
- slicing awal dan batas menggunakan default slicing

CONDITIONAL & OPERATION SERIES


```python
data = np.array([1000,1500,2000,3000])
series = pd.Series(data, index = ["a", "b", "c", "d"])
series
```

```python
series > 1500
```

```python
series[series > 1500]
```

```python
rata = series.mean()
print(rata)

series[series > series.mean()]
```

```python
series[(series > 1000) & (series < 3000)] 
```

```python
series + 5000
```

```python
data1 = np.array([1000,1500,2000,3000])
series1 = pd.Series(data1, index = ["a", "b", "c", "d"])

data2 = np.array([200,200,200,200])
series2 = pd.Series(data2, index = ["a", "b", "c", "d"])

print(series1 + series2)
print(series1.add(series2))
```

```python
data1 = np.array([1000,1500,2000,3000])
series1 = pd.Series(data1, index = ["a", "b", "c", "d"])

data1_type = series1.dtypes
data1_newtype = series1.astype(str)
print(data1_type)
print(data1_newtype.dtypes)
```

```python
# baris

print(df.loc["canada"])

# baris slicing

print(df.loc["canada" : "germany"])

# baris & kolom slicing 

print(df.loc["canada" : "germany", "population" : "gdp"])
```
