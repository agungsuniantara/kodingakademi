FUNCTION ARGS

```python
def greet(*a): # (a, b, c)
    data1 = a[0]
    return data1

contoh = greet("a", "b", "c")
print(contoh)
```

```python
def greet(*a): # (a, b, c)
    for i in a:
        print(i)

greet("a", "b", "c")
```

args = data parameter menjadi tuple
kwargs = data parameter menjadi dictionary

FUNCTION KWARGS

```python
def greet(**a): # {"nama" : "agung", "alamat" : "denpasar"}
    for a,b in a.items():
        print(f"key = {a}, value = {b}")

greet(nama="agung", alamat="denpasar")
```

```python
def rata(*data): # (7,8,9,10)
    jumlah_data = sum(data) # 34
    n = len(data) # 4
    return jumlah_data / n # 34 / 4 = 8.5

data1 = rata(7,8,9,10)
data2 = rata(10,5,6,7)

print("rata-rata data1 =", data1)
print("rata-rata data2 =", data2)
```

ASSIGNMENT

```python
def tambah(*data):
    jumlah_data = sum(data)
    n = len(data)
    return jumlah_data, n

hasil, jumlah = tambah(10,9,5,6)
rata = (lambda a,b : a / b)(hasil, jumlah)
print(f"rata - rata = {rata}")
```

NUMPY INDEXING

```python
a = np.array([
    [10,20,30],
    [5,10,20],
    [2,4,6]
])

print(a[0]) # baris 1
print(a[1]) # baris 2
print(a[0, 2]) # baris 1 kolom 3
print(a[1, 1]) # baris 2 kolom 2
print(a[1, 1]) # baris 2 kolom 2
```

NUMPY SLICING

```python
a = np.array([
    [10,20,30],
    [5,10,20],
    [2,4,6]
])

print(a[0:2]) # [awal:batas]
# posisi kiri (baris), posisi kanan (kolom)
print(a[0:2, 0:2]) # [awal:batas, awal:batas]
print(a[:2, :1])
print(a[1:, 1:])
```

NUMPY SET

```python
a = np.array([
    [10,20,30],
    [5,10,20],
    [2,4,6]
])

a[1] = np.array([8,9,10])
print(a)
a[0] = 100
print(a)
a[2, 1] = 1000
print(a)
```
