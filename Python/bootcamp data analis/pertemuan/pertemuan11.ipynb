```python
import pandas as pd

data = pd.read_csv('data/user_age.csv')

print(data)
print(data.dtypes)
```

```python
data['date'] = pd.to_datetime(data['date'])

print(data)
print(data.dtypes)
```

```python
from datetime import datetime

tanggal_sekarang = datetime.now()
data['age'] = (tanggal_sekarang - data['date']).dt.days
data
```

```python
bins = [0, 1400, 2000, 3000, 4000]

labels = ['< 1400 day', '1400 to 2000 day', '2000 to 3000 day', '3000 > day']

data['age_bin'] = pd.cut(data['age'], bins = bins, labels = labels)

data
```

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1.5,0.9])
axes.bar(max_movie.index, max_movie.values)
axes.set_xlabel('Year', labelpad=30)
axes.set_ylabel('Count', labelpad=30)
plt.xticks(max_movie.index, rotation=70)
plt.title("Movie")
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1.5,0.9])
axes.bar(max_tv.index, max_tv.values)
axes.set_xlabel('Year', labelpad=30)
axes.set_ylabel('Count', labelpad=30)
plt.xticks(max_tv.index, rotation=70)
plt.title("TV Show")
plt.show()
```
