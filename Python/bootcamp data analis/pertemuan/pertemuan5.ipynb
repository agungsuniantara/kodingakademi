FUNCTION PARAMETER & RETURN


```python
def tampil(): # function void
    print("hello world")
    
def tampil2(a, b): # function parameter
    print(a)
    print(b)

def tampil3(): # function return
    nama = "agung"
    return nama

def tampil4(a,b): # function return + parameter
    hasil = a + b
    return hasil

def luas_pp(a,b):
    return a * b

def volume_balok(p,l,t): # function di dalam function
    volume = luas_pp(p,l) * t # 50 * 20
    return volume

tampil()
tampil2("hello world", "halo")
data = tampil3()
print(data)
data2 = tampil4(10, 5)
print(data2)

balok = volume_balok(10, 5, 20)
print(balok)
```

FUNCTION MEMANGGIL DIRINYA SENDIRI


```python
def faktor(n): # 5 # 0
    if n == 0:
        return 1 
    else:
        return n * faktor(n - 1) # 5 * 4 * 3 * 2 * 1

nilai_faktor = faktor(5)
print("Nilai faktor = ", nilai_faktor)
```

```python
def menu():
    print("Silahkan memilih menu : ")
    print("1. Penjumlahan")
    print("2. Pengurangan")
    print("3. Perkalian")
    print("4. Pembagian")

def penjumlahan(a,b):
    print("Jumlah = ", a + b)

def pengurangan(a,b):
    print("Jumlah = ", a - b)

def perkalian(a,b):
    print("Jumlah = ", a * b)

def pembagian(a,b):
    print("Jumlah = ", a / b)

while(True):
    menu()
    pilihan = int(input("masukan pilihan : "))
    if pilihan == 1:
        angka1 = int(input("angka1 : "))
        angka2 = int(input("angka2 : "))
        penjumlahan(angka1, angka2)
    elif pilihan == 2:
        angka1 = int(input("angka1 : "))
        angka2 = int(input("angka2 : "))
        pengurangan(angka1, angka2)
    elif pilihan == 3:
        angka1 = int(input("angka1 : "))
        angka2 = int(input("angka2 : "))
        perkalian(angka1, angka2)
    elif pilihan == 4:
        angka1 = int(input("angka1 : "))
        angka2 = int(input("angka2 : "))
        pembagian(angka1, angka2)
    else:
        print("pilihan tidak ada")
    
    kondisi = input("Masih ingin memilih? ")
    if kondisi == "tidak":
        break
```

```python
# NO 2 (cara 2)

data = [90,40,29,78,22]
terbesar = data[0] # 90
terkecil = data[0] # 22

for i in range(1, len(data)): # (1 , 5), i = 5
    if data[i] > terbesar: # 22 > 90
        terbesar = data[i]
    elif data[i] < terkecil: # 22 < 90
        terkecil = data[i]

print("terbesar : ", terbesar)
print("terkecil : ", terkecil)
```
