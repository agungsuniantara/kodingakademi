```python
import random as rd
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

data = np.array([rd.random() for _ in range(1000)])
```

```python
fig, tampil = plt.subplots(1,1,figsize=(8,4))

tampil.hist(data, bins = 100)
tampil.set_title("histogram")
tampil.set_xlabel("rentang")
tampil.set_ylabel("banyak")

fig.savefig('histogram.png')
```

```python
# mean

np.mean(data)
```

mean adalah nilai rata-rata dari sebuah dataset

```python
# median

np.median(data)
```

median adalah nilai tengah dari sebuah dataset

```python
# modus

import statistics as st

st.mode(data)
```

Modus adalah nilai dalam kumpulan data yang muncul paling sering atau memiliki frekuensi tertinggi.

```python
# varian

varian = np.var(data)

varian
```

Varians mengukur sejauh mana setiap nilai dalam dataset berbeda dari rata-rata

```python
# standar deviasi

std = np.std(data)

std
```

deviasi standar adalah akar kuadrat dari varians
