```python
# args data menjadi tuple

def angka(*data):
  print(data)
  data_list = list(data)
  print(data_list)

angka(1,2,3,4)
```

```python
def angka(**data):
  print(data)
  isi = []
  kunci = []
  for key, value in data.items():
    isi.append(value)
    kunci.append(key)
  print(isi)
  print(kunci)

angka(nama = "agung")
```

```python
def rata(*data):
  jumlah = sum(data) / len(data)
  return jumlah

jumlah = rata(1,2,3,4,5,6)

print(jumlah)
```
