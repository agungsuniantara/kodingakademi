```python
def tambah2(x,b):
  hasil = x + b
  hasil2 = x - 1
  return hasil, hasil2

simpan, simpan2 = tambah2(10,20)

print(simpan)
print(simpan2)
```

```python
def luas_pp(a,b):
  luas = a * b
  return luas

def volume_pp(a,b,t):
  volume = luas_pp(a,b) * t
  return volume

print(f"hasil luas = {luas_pp(10,20)}")
print(f"hasil volume = {volume_pp(10,20,30)}")
```

```python
def faktorial(n): # 10
  if n == 0:
    return 1
  else:
    return n * faktorial(n-1) # 5 * 4 * 3 * 2 * 1

nilai_faktorial = faktorial(10)

print(nilai_faktorial)
```
