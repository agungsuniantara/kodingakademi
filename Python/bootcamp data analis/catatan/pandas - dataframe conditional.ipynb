```python
# assignment

import pandas as pd

df = pd.DataFrame({
    "Athlete" : ["Michael phelps","Michael phelps","Michael phelps","Natalie Coughlin","Aleksey Nemov"],
    "Age" : [23,19,27,25,24],
    "Country" : ["United States", "Canada", "United States", "Norway", "China"],
    "Year" : [2008, 2004, 2012,2008,2000],
    "Gold medals" : [8,6,4,1,2]
})

df
```

```python
# assignment 1

united_states = df[(df["Country"] == "United States")]

united_states[["Athlete"]]
```

```python
# assignment 2

tahun = df[(df["Year"] == 2008)]

tahun[["Athlete"]]
```

```python
# assignment 3

hapus = df[(df["Age"] >= 25) & (df["Country"] == "Norway")]

filter = hapus['Athlete']

for key, value in filter.items():
  df = df.drop([key], axis = "index")

df
```
