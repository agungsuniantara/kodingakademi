```python
# assignment

angka = int(input("masukan nilai : "))

if angka > 0 and angka <= 50:
  print("Nilai D")
elif angka > 50 and angka <= 70:
  print("Nilai C")
elif angka > 70 and angka <= 80:
  print("Nilai B")
elif angka > 80 and angka <= 100:
  print("Nilai A")
else:
  print("masukkan nilai dengan benar")
```

```python
kwh = int(input("masukan kwh : "))
harga = 500

if kwh > 0 and kwh <= 100:
  total = harga * kwh
  print("total yang harus dibayar = Rp", total)
elif kwh > 100 and kwh <= 200:
  total = harga * kwh
  pemakaian = total * 5
  print("total yang harus dibayar = Rp", pemakaian)
elif kwh > 200:
  total = harga * kwh
  pemakaian = total * 10
  print("total yang harus dibayar = Rp", pemakaian)
else:
  print("masukkan nilai kwh dengan benar")
```
