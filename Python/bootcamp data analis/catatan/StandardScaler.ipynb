```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Membuat dataframe contoh
data = {
    'I': [100, 90, 20, 95],
    'II': [30, 45, 56, 10],
    'III': [20, 40, 80, 98]
}

df = pd.DataFrame(data)

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Fit StandardScaler pada kolom 'I'
scaler.fit(df[['I']]) # menggunakan [[]] bukan [] karena standard scaler menggunakan data 2D / dataframe

# Tampilkan rata-rata dan deviasi standar yang telah dihitung
print("Rata-rata:", scaler.mean_)
print("Deviasi Standar:", scaler.scale_)
```

Setelah mendapatkan rata-rata dan standar deviasi dari scaler.fit, selanjutnya menggunakan transform untuk mencari nilai skala standar setiap data nya sebagai berikut :

data kolom I pertama = 100

skala standar = ( data - mean ) / standar deviasi

data pertama = ( 100 - 76.25 ) / 32.66783586 = 0.7270148

proses perhitungan dilakukan sampai data terakhir.

```python
# Transformasi data pada kolom 'I' menjadi skala standar
scaled_data = scaler.transform(df[['I']])

# Tampilkan dataframe hasil
print(scaled_data)
```

Dalam konteks skala standar:

Jika hasil transformasi positif, itu berarti nilai tersebut berada diatas mean.

Jika hasil transformasi negatif, itu berarti nilai tersebut berada di bawah mean.

Jika hasil transformasi adalah 0, itu berarti nilai tersebut sama dengan mean.

```python
# ubah hasil transform kedalam kolom dataframe

scaled_df = pd.DataFrame(scaled_data, columns=['scaled'])

scaled_df
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Contoh data harga saham dua perusahaan (harga penutupan harian dalam dolar)
data = {
    'Saham_A': [100, 10,50,39],
    'Saham_B': [50, 70,55,33]
}

df = pd.DataFrame(data)

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Fit StandardScaler pada dataframe
scaler.fit(df)

# Transformasi data dalam skala standar
scaled_data = scaler.transform(df)

# Hasilnya adalah dataframe baru dengan data dalam skala standar
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

# Tampilkan dataframe hasil
print(scaled_df)
```
