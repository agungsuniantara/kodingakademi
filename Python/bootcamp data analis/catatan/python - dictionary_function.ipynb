```python
# perbedaan keys(), items() dan values()

nama_course = {'EXP': 'Coding Explorer', 'python': 'Python Basic Programming', 'web': 'Web Design HTML CSS'}

print(nama_course.values())
print(nama_course.items())
print(nama_course.keys())

print("\n")

# fungsi keys()
for key in nama_course.keys():
  print(f"key = {key}")

print("\n")

# fungsi values()
for values in nama_course.values():
  print(f"value = {values}")

# fungsi items()
for keys, values in nama_course.items():
  print(f"key = {keys}, value = {values}")
```

```python
# assignment 1

data_list = {} # 0 : tumbuhan, 1 : hewan, 2 : tumubuhan, 3 : hewan
data_tumbuhan = [] # tumbuhan
data_hewan = [] # hewan

for i in range(2): # i = 0
    data1 = input("Masukkan data tumbuhan: ")
    data2 = input("Masukkan data hewan: ")
    data_list.update({i * 2: data1}) # i * 2 = 0
    data_list.update({i * 2 + 1: data2}) # i * 2 + 1 = 1
    data_tumbuhan.append(data1)
    data_hewan.append(data2)

data_backup = data_list.copy()

key_dihapus = []

for key, value in data_list.items():
  if value in data_tumbuhan:
    key_dihapus.append(key)

for key in key_dihapus:
  data_list.pop(key)
  # del data_list[key]

print("data sebelum dihapus : ")
print(data_backup)

print("data setelah dihapus : ")
print(data_list)


```

```python
# assignment 2

data_list = {}
data_tumbuhan = []
data_hewan = []

for i in range(2):
    data1 = input("Masukkan data tumbuhan: ")
    data2 = input("Masukkan data hewan: ")
    data_list.update({i * 2: data1})
    data_list.update({i * 2 + 1: data2})
    data_tumbuhan.append(data1)
    data_hewan.append(data2)

data_backup = data_list.copy()

for key, value in data_list.items():
  if key % 2 == 1:
    data_list[key] = "hewan"

print("data sebelum diganti : ")
print(data_backup)

print("data setelah diganti : ")
print(data_list)
```
