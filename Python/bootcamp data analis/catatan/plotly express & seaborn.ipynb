```python
import pandas as pd

data = pd.read_csv('penyakit_ginjal_kronik-1 (1).csv')

data
```

```python
data = data.dropna()
```

```python
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns

px.bar(data, x = 'seldarahmerah')
```

```python
sns.countplot(data, x = 'seldarahmerah')
```

```python
group = data.groupby('seldarahmerah')['umur'].mean()

px.bar(group.reset_index(), x = 'seldarahmerah', y = 'umur')
```

```python
sns.barplot(data, x = 'seldarahmerah', y = 'umur')
```

```python
px.scatter(data, x = 'MCV', y = 'seldarahmerah', color = 'umur')
```

```python
plt.figure(figsize=(15,5))
sns.scatterplot(data, x = 'MCV', y = 'seldarahmerah', hue = 'umur')
```

```python
px.histogram(data, x = 'MCV')
```

```python
plt.figure(figsize=(20,5))
sns.histplot(data, x = 'MCV')
```

```python
sns.boxplot(data, x = 'seldarahmerah', y = 'umur')
```

```python
px.box(data, x = 'seldarahmerah', y = 'umur')
```

```python
plt.figure(figsize=(20,5))

sns.lineplot(data, x = 'MCV', y = 'umur', hue = 'seldarahmerah')
```

```python
px.line(data, x = 'MCV', y = 'umur', color = 'seldarahmerah')
```

```python
pie = data.groupby('seldarahmerah')['umur'].sum()
pie.plot(kind='pie')
plt.legend(data['seldarahmerah'].unique())
```

```python
px.pie(data, names = 'seldarahmerah', values = 'umur')
```
