```python
def dictionary(**contoh):
  for x,y in contoh.items():
    print(f"key = {x} , value = {y}")

dictionary(nama = "agung", alamat = "denpasar")
```

```python
def hitung(*coba):
  hasil = sum(coba)
  n = len(coba)
  return hasil / n

hasil = hitung(1,2,3,4,5,6)

print(hasil)
```

```python
# slicing :

# [1,2] = 1. baris, 2. kolom
# [0:1] = index 0 sampai index 1
# [1:, 0] = baris index 1 ke bawah
# [:1, 0] = batas baris index 1 ke atas
# [0, :2] = batas kolom index 2 ke belakang
# [0, 2:] = kolom index 2 ke depan
```

```python
data_2d = np.array([
    [1,2,3],
    [6,7,8]
])

data_2d[0] = np.array([2,5,2])

print(data_2d)
```
