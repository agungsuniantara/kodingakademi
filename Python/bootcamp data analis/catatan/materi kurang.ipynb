```python
import pandas as pd
import numpy as np

data = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

data
```

```python
# map

def custom_function(row): # 1
  return row + 2

mapp = data['A'].map(custom_function) # fungsi map hanya bisa pada data series
applymapp = data[['A']].applymap(custom_function)

contoh = list(map(lambda x: x + 2, data['A']))
a,b,c = map(lambda x: x + 2, data['A'])

# print(contoh)
# print(a)
print(applymapp)
```

```python
# apply

def custom_function(row):
  return row + 2

applyy = data[['A']].apply(custom_function) # fungsi apply bisa pada data dataframe atau series

applyy
```

```python
# transform

def custom_function(row):
  return row + 2

transformm = data[['A']].transform(custom_function) # fungsi transform mirip seperti apply

transformm
```

```python
df = pd.DataFrame({
    'key': ['a','b','c'] * 3,
    'A': [4,5,6] * 3,
    'B': [1,2,3] * 3,
})

df
```

```python
group_apply = df.groupby('key')['A'].sum()

group_apply
```

```python
def group_sum(x):
    return x.sum()

group_apply = df.groupby('key')['A'].apply(group_sum)

group_transform = df.groupby('key')['A'].transform(group_sum)

group_apply
```

```python
# merge dan concat

df1 = pd.DataFrame({'Nama': ['Alice', 'Bob', 'Charlie']})

df2 = pd.DataFrame({'Nama': ['Alice', 'Agung', 'Charlie'], 'Umur': [25, 30, 35]})

result_merge = pd.merge(df1, df2, on='Nama')

result_merge
```

```python
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Nama': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'ID': [1, 2, 3], 'Umur': [25, 30, 35]})

result_concat1 = pd.concat([df1, df2])

result_concat2 = pd.concat([df1, df2]).set_index('ID')

result_concat3 = pd.concat([df1, df2], ignore_index = True)

display(result_concat1, result_concat2, result_concat3)
```

stack mengubah kolom menjadi indeks, sedangkan unstack mengubah indeks menjadi kolom

```python
import pandas as pd

data = {
    'Tinggi': [175, 180],
    'Berat': [70, 75]
}

df = pd.DataFrame(data, index = ['agung', 'koko'])

df
```

```python
stacked_df = df.stack()

stacked_df
```

```python
stacked_df.loc['agung']
```

```python
unstacked_df = stacked_df.unstack()

unstacked_df
```

```python
unstacked_df = df.unstack()

unstacked_df
```

```python
unstacked_df.loc[:, 'agung']
```

```python
from folium.plugins import HeatMap
import folium

data = {
    'Lokasi': ['Denpasar', 'Ubud', 'Kuta', 'Seminyak', 'Canggu'],
    'Latitude': [-8.671483, -8.509692, -8.721267, -8.691243, -8.654356],
    'Longitude': [115.213236, 115.262370, 115.169668, 115.168586, 115.133534],
    'Banyak': [1000, 900, 800, 700, 600]
}

df = pd.DataFrame(data)

df
```

```python
group = df.groupby(['Latitude','Longitude'])['Banyak'].sum().reset_index()

group
```

```python
basemap = folium.Map()

HeatMap(group, radius = 30).add_to(basemap)

basemap
```
