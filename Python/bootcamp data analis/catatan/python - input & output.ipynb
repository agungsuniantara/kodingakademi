```python
List = list(map(int, input("Enter List elements : ").split()))
Set = set(map(int, input("Enter the Set elements :").split()))

print(List)
print(Set)
```

List: List adalah koleksi berurutan yang memungkinkan Anda menyimpan berbagai jenis elemen, termasuk duplikasi. Anda dapat mengakses elemen list berdasarkan indeks mereka.

Set: Set adalah koleksi tanpa urutan yang hanya menyimpan nilai unik. Tidak ada duplikasi dalam set, dan Anda tidak dapat mengakses elemen set berdasarkan indeks.

```python
# list (bisa duplikasi) dan set (tidak bisa duplikasi)

angka_list = [1,1,2,3,4,5]

print(angka_list)

angka_set = {1,1,2,3,4,5}

print(angka_set)
```

```python
# list (bisa di akses dengan index) dan set (tidak bisa di akses dengan index)

print(angka_list[0])
print(angka_set[0])
```

```python
# list (bisa di rubah) dan set (tidak bisa di rubah)

angka_list[0] = 5

print(angka_list[0])

angka_set[0] = 5
```

```python
# list (data nya beraturan) dan set (data nya tidak beraturan)

nama = input("masukan nama :")

list_nama = list(map(str, nama.split()))
set_nama = set(map(str, nama.split()))

print(list_nama)
print(set_nama)

print("list nama = ")

for x in range(len(list_nama)):
  print(list_nama[x] + " " , end="")

print("\nset nama = ")

for x in range(len(set_nama)):
  print(list_nama[x] + " " , end="")
```

List: Biasanya digunakan saat Anda memerlukan urutan dan duplikasi, serta akses berdasarkan indeks. Namun, untuk operasi pencarian dan penghapusan, kinerja list mungkin tidak secepat set, terutama untuk koleksi besar.

Set: Digunakan ketika Anda ingin menghindari duplikasi dan tidak memerlukan urutan tertentu. Set cenderung lebih cepat dalam operasi pencarian dan penghapusan elemen daripada list, terutama untuk jumlah data yang besar.
