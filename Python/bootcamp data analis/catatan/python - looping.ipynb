List:

- Ketika Anda memerlukan koleksi elemen yang dapat diubah (mutable) setelah pembuatan, seperti menambah, mengubah, atau menghapus elemen-elemen.
- Untuk menyimpan urutan data yang mungkin memiliki duplikasi.
- Ketika Anda ingin menjaga elemen-elemen dalam urutan tertentu berdasarkan indeks.

Contoh penggunaan list:

- Daftar belanjaan yang dapat berubah.
- Log data yang terus bertambah.
- Daftar nama-nama pengguna pada sebuah platform.

Tuple:

- Ketika Anda ingin melindungi data dari perubahan setelah pembuatan.
- Untuk menyimpan data yang memiliki arti atau makna tetap, seperti koordinat geografis.

Contoh penggunaan tuple:

- Koordinat geografis (latitude, longitude).
- Data yang merepresentasikan titik dalam ruang.

Set:

- Ketika Anda perlu menyimpan kumpulan elemen unik (tanpa duplikasi).
- Ketika Anda ingin melakukan operasi matematika himpunan seperti gabungan, irisan, atau perbedaan.

Contoh penggunaan set:

- Menyimpan daftar alamat email unik.
- Mengetahui elemen yang sama antara dua kelompok data.
- Menghilangkan elemen duplikat dari koleksi data.

```python
# list dan tuple (bisa duplikasi) dan set (tidak bisa duplikasi)

angka_list = [1,1,2,3,4,5]

print(angka_list)

angka_tuple = (1,1,2,3,4,5)

print(angka_tuple)

angka_set = {1,1,2,3,4,5}

print(angka_set)
```

```python
# list dan tuple (bisa di akses dengan index) dan set (tidak bisa di akses dengan index)

print(angka_list[0])
print(angka_tuple[0])
print(angka_set[0])
```

```python
# list (bisa di rubah) dan set atau tuple (tidak bisa di rubah)

angka_list[0] = 5

print(angka_list[0])

angka_set[0] = 5
angka_tuple[0] = 5
```

```python
# list serta tuple (data nya beraturan) dan set (data nya tidak beraturan)

nama = input("masukan nama :")

list_nama = list(map(str, nama.split()))
set_nama = set(map(str, nama.split()))
tuple_nama = tuple(map(str, nama.split()))

print(list_nama)
print(tuple_nama)
print(set_nama)
```

```python
# iterasi dalam list

for index in angka_list:
  print(index)

print("\n")

# iterasi dalam tuple

for index in angka_tuple:
  print(index)

print("\n")

# iterasi dalam set

for index in angka_set:
  print(index)
```
