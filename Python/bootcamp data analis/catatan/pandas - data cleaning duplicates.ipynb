```python
import pandas as pd

data = pd.read_csv('user_age (1).csv')

data
```

```python
data['date'] = pd.to_datetime(data['date'])

data
```

```python
data['date'] = data['date'].dt.strftime('%d-%m-%Y')

data
```

```python
from datetime import datetime

tanggal_sekarang = datetime.now()

data['age'] = (tanggal_sekarang - data['date']).dt.days

data
```

```python
bins = [0, 1400, 2000, 3000, 4000]

labels = ['< 1400 day', '1400 to 2000 day', '2000 to 3000 day', '4000 > day']

data['age_bin'] = pd.cut(data['age'], bins = bins, labels = labels)

data
```
