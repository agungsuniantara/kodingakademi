```python
# assignment function

def penjumlahan(angka1, angka2):
  hasil = angka1 + angka2
  return hasil

def pengurangan(angka1, angka2):
  hasil = angka1 - angka2
  return hasil

def pembagian(angka1, angka2):
  hasil = angka1 / angka2
  return hasil

def perkalian(angka1, angka2):
  hasil = angka1 * angka2
  return hasil

while True:
  print("Silahkan memilih menu : ")
  print("1. Penjumlahan")
  print("2. Pengurangan")
  print("3. Perkalian")
  print("4. Pembagian")

  try:
    masukan = int(input("masukan pilihan : "))

    if masukan == 1:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      print("hasil jumlah =",penjumlahan(angka1, angka2))
    elif masukan == 2:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      print("hasil pengurangan =",pengurangan(angka1, angka2))
    elif masukan == 3:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      print("hasil perkalian =",perkalian(angka1, angka2))
    elif masukan == 4:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      print("hasil pembagian =",pembagian(angka1, angka2))
    else:
      print("pilihan tidak ada")

    kondisi = input("masih ingin memilih ?  ")

    if kondisi == "tidak":
      print("Terimakasih, selamat datang kembali")
      break
    elif kondisi == "ya":
      print("")
    else:
      print("kondisi salah")

  except ValueError:
    print("masukan angka dengan benar")
```

```python
# assignment 2

def terbesar(a):
  terbesar = max(a)
  return terbesar

def terkecil(a):
  terkecil = min(a)
  return terkecil

data_list = [2,3,1,5,4,2,1]

urut_terbesar = sorted(data_list, reverse = True)
urut_terkecil = sorted(data_list)

print(f"nilai terbesar dari list {data_list} adalah {terbesar(data_list)}")
print(f"nilai terbesar dari list {data_list} adalah {terkecil(data_list)}")

print(f"data diurutkan dari terbesar ke terkecil = {urut_terbesar}")
print(f"data diurutkan dari terkecil ke terbesar = {urut_terkecil}")

```

```python
def terbesar(a):
    hasil = a[0]
    for angka in a:
        if angka > hasil:
            hasil = angka
    return hasil

def terkecil(a):
    hasil = a[0]
    for angka in a:
        if angka < hasil:
            hasil = angka
    return hasil

data_list = [2, 3, 1, 5, 4, 2, 1]

print(f"Nilai terbesar dari list {data_list} adalah {terbesar(data_list)}")
print(f"Nilai terkecil dari list {data_list} adalah {terkecil(data_list)}")

```
