```python
# assignment 6

mystr = "Welcome to koding akademi"

masukan = input("masukan kata : ") #koding

if masukan in mystr:
  index_masukan = mystr.index(masukan)
  print(index_masukan)
  ukuran_masukan = len(masukan)
  print(ukuran_masukan)

  print(mystr[index_masukan: index_masukan + ukuran_masukan]) #mystr[11:17]
else:
  print("masukan kata yang sesuai")
```

```python
# membalikkan kata

# cara 1 :

print(mystr[::-1])

# cara 2 :

pisah = mystr.split()
print(pisah)
terbalik = reversed(pisah)
print(terbalik)

kata_baru = " ".join(terbalik)

print(kata_baru)
```

Fungsi join() adalah metode bawaan dalam bahasa pemrograman Python yang digunakan untuk menggabungkan (atau menggabungkan kembali) elemen-elemen dalam suatu iterable menjadi sebuah string. Metode ini sangat berguna ketika Anda ingin menggabungkan elemen-elemen dalam list, tuple, atau set ke dalam sebuah string, dengan menyisipkan pemisah tertentu di antara setiap elemen.

Fungsi reversed() adalah fungsi bawaan dalam Python yang digunakan untuk menghasilkan iterator yang membalik urutan elemen-elemen dalam suatu iterable. Iterator yang dihasilkan dapat digunakan untuk mengakses elemen-elemen dalam urutan terbalik tanpa harus mengubah struktur data asli. Fungsi reversed() biasanya digunakan dengan struktur data yang mendukung iterasi, seperti list, tuple, dan string.
