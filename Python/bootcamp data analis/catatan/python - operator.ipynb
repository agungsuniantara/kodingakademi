```python
#is operator

a = 100
print(id(a))
b = a
print(id(b))
c = 100
print(id(c))
d = 200
print(id(d))

print("\n")
print(id(b == a))
print(id(a == c))
print(id(a == d))
print(id(b == d))

print("\n")
print(id(b is a))
print(id(a is c))
print(id(a is d))
print(id(b is d))
```

```python
#bitwise (berhubungan dengan bit) = hanya angka, logical = string dan angka

angka1 = 1
angka2 = 1

print(bool(angka1 ^ angka2))
print(bool(angka1 & angka2))
print(bool(angka1 | angka2))
print(bool(angka1 and angka2))
print(bool(angka1 or angka2))

# shift

num = 5       # 101 dalam biner
shifted = num << 2   # Geser 2 bit ke kiri, hasil: 10100 (20 dalam desimal)
print(shifted)

num = 16      # 10000 dalam biner
shifted = num >> 3   # Geser 3 bit ke kanan, hasil: 00010 (2 dalam desimal)
print(shifted)

# unary

num = 5
positive_num = +num  # Tidak umum digunakan dalam praktik
print(positive_num)

num = 10
negative_num = -num
print(negative_num)  # Output: -10

# bitwise negation (kalau 1 jadi 0 , kalau 0 jadi 1)

num = 5   # 101 dalam biner
negated_num = ~num
print(negated_num)  # Output: -6 (11111111111111111111111111111010 dalam biner)

# floor division

a = 20
b = 3
c = a // b

print(c)

```
