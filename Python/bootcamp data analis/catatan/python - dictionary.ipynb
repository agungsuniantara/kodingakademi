```python
buku = {"action": "avenger", "romance": "twilight"}

while True:
    print("Pilihlah pilihan dibawah ini : ")
    print("1. hapus")
    print("2. edit")
    print("3. tambah")
    print("4. keluar")

    masukan = input("Ketik pilihan: ")

    if masukan == "1":
        hapus = input("Ketik jenis buku yang ingin dihapus: ")
        if hapus in buku:
            del buku[hapus]
            print("Berhasil dihapus")
            print(buku)
        else:
            print("Jenis buku tidak ada")

    elif masukan == "2":
        edit = input("Ketik jenis buku yang ingin diedit: ")
        if edit in buku:
            proses = input(f"Ganti nama jenis {edit} menjadi: ")
            buku[proses] = buku.pop(edit)
            print(buku)
        else:
            print("Jenis buku tidak ada")

    elif masukan == "3":
        jenis_baru = input("Ketik jenis buku baru: ")
        judul_baru = input("Ketik judul buku baru: ")
        buku[jenis_baru] = judul_baru
        print("Buku baru ditambahkan")
        print(buku)

    elif masukan == "4":
        break

    else:
        print("Pilihan tidak ada")
```
