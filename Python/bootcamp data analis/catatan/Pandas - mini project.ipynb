```python
data = pd.read_csv("bali_tourist.csv")

data
```

```python
data_2018 = data[data['year'] == 2018]
data_2019 = data[data['year'] == 2019]
```

```python
data_2018
```

```python
data_2019
```

```python
kolom = ['January','February','March','April','May','June','July','August','September','October','November','December']
total_2018 = data_2018[kolom].sum()
total_2019 = data_2019[kolom].sum()

total_kedatangan_2018 = total_2018.sum()
total_kedatangan_2019 = total_2019.sum()

print("total 2018 =", total_kedatangan_2018)
print("total 2019 =", total_kedatangan_2019)
```

```python
rata_2018 = total_2018.mean()
rata_2019 = total_2019.mean()

print(rata_2018)
print(rata_2019)
```

```python
max = total_2018.max()

for key, value in total_2018.items():
  if max == value:
    print(key)
```

```python
min = total_2018.min()

for key, value in total_2018.items():
  if min == value:
    print(key)
```
