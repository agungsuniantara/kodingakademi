```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 22]}

df = pd.DataFrame(data)

print(df)

# Mengakses baris 'row1' dan kolom 'Name' menggunakan loc
print(df.loc[0, 'Name'])  # Output: Alice

# Mengakses beberapa baris dan kolom menggunakan loc
print(df.loc[[0, 2], ['Name', 'Age']])

```

```python
import pandas as pd

# perbedaan loc dan iloc

data = {1: ['Alice', 'Bob', 'Charlie'],
        "umur": [25, 30, 22]}

data_series = ['scratch','roblox','php','python']

df = pd.DataFrame(data, index = [0, "row2", "row3"])
series = pd.Series(data_series, index = [0,1,2,3])

print(df)

print("\n")

print(df.loc[0, "umur"])

print(df.iloc[0, 1])

print(series.loc[0 : 3])

print(series.iloc[0 : 3])
```

```python
# assignment

assignment = ['jan','feb','mar']

series = pd.Series(assignment)

print(series.loc[0:2])
print(series.iloc[0:3])
```

```python
assignment_2 = {
    'month' : assignment
}

series_2 = pd.DataFrame(assignment_2)

print(series_2.loc[:"month"])
print(series_2.iloc[0:, 0])
# print(series_2[0:, 0])

print(series_2.to_string(index = False))
```
