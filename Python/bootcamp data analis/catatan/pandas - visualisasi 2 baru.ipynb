```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data1 = pd.read_csv('bali_tourist (1).csv')

data1
```

```python
plt.figure(figsize=[15,5])
x = np.linspace(5,20,100)
plt.plot(x, np.sin(x));
```

```python
plt.figure(figsize=[15,5])
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x),linestyle=":",linewidth=4)
plt.plot(x,np.cos(x),linestyle="--",linewidth=4)
plt.plot(x,np.cos(2*x),linestyle="-.",linewidth=4);
```

```python
x1 = np.random.laplace(1, 0.8, 100)
x2 = np.random.normal(1, 2, 100)

plt.hist(x1, alpha=0.5, density=True, bins=20)
plt.hist(x2, alpha=0.5, density=True, bins=20)
```

```python
january = data1['January']
februari = data1['February']

plt.hist(january, alpha=0.5, density=True, bins=10)
plt.hist(februari, alpha=0.5, density=True, bins=10)
```

```python
plt.boxplot(data1['January']);
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x = data1['January'], y = data1['area'],alpha=0.5,label='January')
plt.scatter(x = data1['March'], y = data1['area'],alpha=0.5,label='March')

plt.xlabel("Tourist")
plt.ylabel("Area")
plt.title("Death Rates");
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x = data1['January'], y = data1['April'],alpha=0.5,label='January')
plt.scatter(x = data1['March'], y = data1['May'],alpha=0.5,label='March')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Data")
plt.legend()
```
