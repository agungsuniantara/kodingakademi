```python
# pakai global

contoh = 10

def tampil():
  global contoh
  contoh += 1
  print(contoh)

def coba():
  print("nilai = ", contoh)

tampil()
coba()
```

```python
# assignment function

def penjumlahan():
  global angka1, angka2
  hasil = angka1 + angka2
  print("hasil penjumalahn = ", hasil)

def pengurangan():
  global angka1, angka2
  hasil = angka1 - angka2
  print("hasil pengurangan = ", hasil)

def pembagian():
  global angka1, angka2
  hasil = angka1 / angka2
  print("hasil pembagian = ", hasil)

def perkalian():
  global angka1, angka2
  hasil = angka1 * angka2
  print("hasil perkalian = ", hasil)

while True:
  print("Silahkan memilih menu : ")
  print("1. Penjumlahan")
  print("2. Pengurangan")
  print("3. Perkalian")
  print("4. Pembagian")

  try:
    masukan = int(input("masukan pilihan : "))

    if masukan == 1:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      penjumlahan()
    elif masukan == 2:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      pengurangan()
    elif masukan == 3:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      perkalian()
    elif masukan == 4:
      angka1 = int(input("masukan angka 1 : "))
      angka2 = int(input("masukan angka 2 : "))
      pembagian()
    else:
      print("pilihan tidak ada")

    kondisi = input("masih ingin memilih ?  ")

    if kondisi == "tidak":
      print("Terimakasih, selamat datang kembali")
      break
    elif kondisi == "ya":
      print("")
    else:
      print("kondisi salah")
      break

  except ValueError:
    print("masukan angka dengan benar")

```
