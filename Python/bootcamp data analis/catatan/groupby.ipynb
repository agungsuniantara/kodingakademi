```python
import pandas as pd

data = {
    'Kategori': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Nilai1': [10, 15, 20, 25, 30, 35],
    'Nilai2': [100, 150, 200, 250, 300, 350],
    'Nilai3': [1000, 1500, 2000, 2500, 3000, 3500]
}

df = pd.DataFrame(data)

df
```

```python
# Mengelompokkan berdasarkan 1 kolom kategori

grouped = df.groupby('Kategori')

grouped.first()
```

```python
# Mengelompokkan berdasarkan 1 kolom kategori dan 1 kolom data

grouped = df.groupby('Kategori')['Nilai1']

grouped.first()
```

```python
# Mengelompokkan berdasarkan 2 kolom kategori

double = df.groupby(['Nilai1', 'Nilai2'])

double.first()
```

```python
# Mengelompokkan berdasarkan 2 kolom kategori dan 1 kolom data

double = df.groupby(['Nilai1', 'Nilai2'])['Kategori']

double.first()
```

```python
# Menampilkan 1 kategori berdasarkan value

grouped.get_group('A')
```

```python
# Menampilkan 2 kategori berdasarkan value

double.get_group((10, 100))
```

```python
# Contoh Penggunaan groupby().mean()

mean_values = grouped.mean()

print(mean_values)
```

```python
# Contoh Penggunaan groupby().sum()

sum_values = grouped.sum()

print(sum_values)
```

```python
# Contoh Penggunaan groupby().max() dan groupby().min()

max_values = grouped.max()
min_values = grouped.min()

print(max_values)
print(min_values)
```

```python
# Contoh Penggunaan Iterasi melalui Kelompok

for name, group in grouped:
    print(f'\nKelompok {name}:')
    print(group)
```

```python
# Contoh Penggunaan groupby().transform()

df['Rata-rata_Nilai1'] = grouped['Nilai1'].transform('mean')
df['Total_Nilai2'] = grouped['Nilai2'].transform('sum')

print(df)
```
