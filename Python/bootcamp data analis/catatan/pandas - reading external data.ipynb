```python
url = 'https://raw.githubusercontent.com/mediatechid/bootcamp-data-analyst/main/zomato_resto_file1.json'

data = pd.read_json(url)

data
```

```python
data.shape
```

```python
data.head(10)
```

```python
data.tail(10)
```

```python
restoran = data[['restaurants']]

restoran
```

```python
nama_restoran = []

for key, value in restoran.items():
  for a,b in value.items():
    if isinstance(b, list):
      for x in b:
        for z,y in x['restaurant'].items():
          # print(f" key = {z}, value = {y}")
          if z == 'name':
            nama_restoran.append(y)
            print(y)
```
