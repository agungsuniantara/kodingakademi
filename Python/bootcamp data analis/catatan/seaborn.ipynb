```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Contoh DataFrame pandas
data = {'Nama': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'Nilai': [85, 92, 78, 88, 95]}

df = pd.DataFrame(data)

# Membuat grafik bar menggunakan Seaborn
plt.figure(figsize=(8, 5))
sns.barplot(x='Nama', y='Nilai', data=df, palette='viridis')

# Menambahkan label dan judul
plt.xlabel('Nama')
plt.ylabel('Nilai')
plt.title('Grafik Nilai')

# Menampilkan nilai di atas setiap batang bar
for index, row in df.iterrows():
    plt.text(index, row['Nilai'] + 2, str(row['Nilai']), ha='center')

# Menampilkan grafik
plt.show()
```
