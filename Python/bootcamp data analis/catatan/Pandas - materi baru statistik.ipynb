1. MEAN ( RATA - RATA )

Mean adalah jumlah dari semua nilai dalam sebuah dataset, dibagi dengan jumlah total nilai tersebut. Ini menggambarkan nilai rata-rata dari sebuah data.

```python
# cara 1

data = [10, 15, 20, 25, 30]
mean = sum(data) / len(data)
print("Mean:", mean)
```

```python
# cara 2

rata = np.mean(data)
print("Mean:", rata)
```

2. MEDIAN ( NILAI TENGAH )

Median adalah nilai tengah dalam dataset setelah diurutkan. Jika jumlah data ganjil, median adalah nilai tengah, jika jumlah data genap, median adalah rata-rata dari dua nilai tengah.

```python
# cara 2

tengah = data_frame['penderekan'].median()

print("tengah:", tengah)
```

4. VARIANS

Varians mengukur sejauh mana setiap nilai dalam dataset menyebar dari rata-rata. Varians lebih tinggi jika nilai-nilai lebih tersebar.

contoh perhitungan :

data: [10, 15, 20, 25, 30]

mean: (10 + 15 + 20 + 25 + 30) / 5 = 20

varians:
((10 - 20)^2 + (15 - 20)^2 + (20 - 20)^2 + (25 - 20)^2 + (30 - 20 )^ 2) / 5 = 50

maka rata-rata perbedaan setiap data adalah 50

berikut diterapkan dalam kode :

```python
# cara 1

data = [10, 15, 20, 25, 30]

mean = np.mean(data)

perbedaan = [(x - mean) ** 2 for x in data]

varian = np.mean(perbedaan)

print("varian:",varian)
```

```python
# cara 2

data = [10, 15, 20, 25, 30]

varian = np.var(data)

print("varian:",varian)
```

```python
# cara 3

varian = data_frame['penderekan'].var()

print("varian:", varian)
```

5. STANDAR DEVIASI

Standar deviasi adalah akar kuadrat dari varians. Ini mengukur dispersi data dan memberikan gambaran tentang seberapa jauh nilai-nilai data dari rata-rata.

standar deviasi = akar kuadrat(varian)

```python
# cara 1

data = [10, 15, 20, 25, 30]

mean = np.mean(data)

perbedaan = [(x - mean) ** 2 for x in data]

varian = np.mean(perbedaan)

standar_deviasi = np.sqrt(varian)

print("standar deviasi:", standar_deviasi)
```

```python
# cara 2

data = [10, 15, 20, 25, 30]

standar_deviasi = np.std(data)

print("standar deviasi:", standar_deviasi)
```

```python
# cara 3

standar_deviasi = data_frame['penderekan'].std()

rata = data_frame['penderekan'].mean()

print("rata-rata:", round(rata))
print("standar_deviasi:", round(standar_deviasi))
```

didapatkan hasil std = 86 dan rata-rata = 160. berarti ini mengartikan rentang nilai dari data frame adalah (160-86) sampai dengan (160+86), sebaran data 74 - 246. nilai dari standar deviasi akan dipengaruhi oleh nilai outlier.

Outlier adalah nilai yang sangat ekstrem atau tidak biasa yang berbeda secara signifikan dari nilai-nilai lain dalam dataset. Outlier dapat menyebabkan standar deviasi yang lebih tinggi atau lebih rendah daripada yang seharusnya.

```python
# contoh penggunaan std dalam saham

import numpy as np

# Harga saham harian dalam dolar
harga_saham = [45.2, 46.5, 44.8, 47.2, 46.0, 47.8, 48.5, 49.2, 45.9, 47.1,
               44.5, 43.8, 42.2, 43.5, 44.8, 46.2, 47.5, 48.8, 47.2, 49.0,
               46.5, 45.2, 44.0, 42.8, 41.5, 43.2, 42.9, 41.7, 43.0, 44.5]

mean_price = np.mean(harga_saham)
variance = np.var(harga_saham)
std_deviation = np.std(harga_saham)

print("Harga Saham Rata-rata:", mean_price)
print("Varians Harga Saham:", variance)
print("Standar Deviasi Harga Saham:", std_deviation)
```

6. KUARTIL

Kuartil adalah nilai-nilai yang membagi dataset menjadi empat bagian yang sama besar. Kuartil digunakan dalam statistik untuk mengukur sebaran data dan memberikan gambaran tentang letak nilai-nilai dalam distribusi data. Ada tiga kuartil utama yang biasanya dihitung: kuartil pertama (Q1), kuartil kedua (Q2), dan kuartil ketiga (Q3).

Q1 = (N + 1) / 4

Q2 = median

Q3 = (3 X ( N + 1 )) / 4



```python
import numpy as np

# Data contoh
data = [10, 15, 20, 25, 30, 40, 50, 60, 70, 500]

# Menghitung kuartil
q1 = np.percentile(data, 25) # 18
q2 = np.percentile(data, 50)  # 35
q3 = np.percentile(data, 75)  # 65

# Menghitung IQR (rentang interkuartil)
iqr = q3 - q1

# Mencari data outlier
lower_bound = q1 - 1.5 * iqr
# lower_bound = max(q1 - 1.5 * iqr, min(data))
upper_bound = q3 + 1.5 * iqr

outliers = []

for x in data:
  if x < lower_bound or x > upper_bound:
    outliers.append(x)

data_bersih = []

for x in data:
  if x not in outliers:
    data_bersih.append(x)

print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
print("IQR:", iqr)
print("batas bawah:", lower_bound)
print("batas atas:", upper_bound)
print("outliers:",outliers)
print("data bersih:",data_bersih)
```

7. SKEWNESS

Skewness adalah ukuran statistik yang digunakan untuk mengukur sejauh mana distribusi data cenderung miring atau tidak simetris terhadap nilai tengahnya. Skewness mengindikasikan apakah ekor distribusi data condong lebih ke kanan (positif skewness) atau lebih ke kiri (negatif skewness) dari nilai tengahnya.

skewness = (3 x ( mean - median )) / standar deviasi

```python
import numpy as np
from scipy.stats import skew

# Dataset untuk skewness positif (condong ke kanan)
data_pos = [100, 50, 40, 30, 20, 10]

# Dataset untuk skewness negatif (condong ke kiri)
data_neg = [-100, 50, 40, 30, 20, 10]

# Dataset untuk skewness mendekati nol (simetris)
data_sym = [10, 20, 30, 40, 50, 60]

# Menghitung skewness
skewness_pos = skew(data_pos)
skewness_neg = skew(data_neg)
skewness_sym = skew(data_sym)

print("Skewness Positif:", skewness_pos)
print("Skewness Negatif:", skewness_neg)
print("Skewness Simetris:", skewness_sym)
```
