```python
import random as rd
from matplotlib import pyplot as plt
import numpy as np

rd.seed(1) # digunakan untuk mencatat nilai random
for _ in range(5): # _ digunakan jika tidak peduli dengan index
  print(rd.random())

print("\n")

for _ in range(5):
  print(rd.random())

print("\n")

rd.seed(1)
for _ in range(5):
  print(rd.random())
```

```python
rvs = np.array([rd.random() for _ in range(1000)])

plt.figure(figsize=(8,4))
plt.hist(rvs, bins=100)
plt.title("histogram")
plt.xlabel('Rentang nilai')
plt.ylabel('Banyak nilai')

# bins merupakan jumlah grafik batang pada histogram yang setiap batangnya mengandung rentang nilai
```

Modus adalah nilai dalam kumpulan data yang muncul paling sering atau memiliki frekuensi tertinggi.

```python
r1 = [rd.normalvariate(0.5,0.2) for _ in range(10000)]
r2 = [rd.normalvariate(0.2,0.1) for _ in range(5000)]
r3 = [rd.normalvariate(0.8,0.2) for _ in range(5000)]

fig, axes = plt.subplots(1,2,figsize=(12,2)) # (baris, kolom, ukuran gambar), memiliki 1 baris dan 2 kolom

axes[0].hist(r1,bins=100)
axes[0].set_title("Unimodal")
axes[1].hist(r2+r3,bins=100)
axes[1].set_title("Bimodal");
```

```python
r1 = [rd.normalvariate(0.5,0.2) for _ in range(10000)]
r2 = [rd.normalvariate(0.2,0.1) for _ in range(5000)]
r3 = [rd.normalvariate(0.8,0.2) for _ in range(5000)]

fig, axes = plt.subplots(2,2,figsize=(12,5)) # (baris, kolom, ukuran gambar), memiliki 2 baris dan 2 kolom

axes[0,0].hist(r1, bins = 100) # baris 1, kolom 1
axes[0,1].hist(r2, bins = 100) # baris 1, kolom 2
axes[1,0].hist(r1, bins = 100) # baris 2, kolom 1
axes[1,1].hist(r2, bins = 100) # baris 2, kolom 2

axes[0,0].set_title("data 1")
axes[0,1].set_title("data 2")
axes[1,0].set_title("data 3")
axes[1,1].set_title("data 4")

plt.subplots_adjust(hspace=0.4) # hspace berfungsi untuk memberikan space plot setiap baris
```

```python
r1 = [random.normalvariate(0.5,0.3) for _ in range(10000)] # (rata-rata, standar deviasi)
r2 = [random.normalvariate(0.5,0.1) for _ in range(10000)] # (rata-rata, standar deviasi)
fig, axes = plt.subplots(1,2,figsize=(12,5))

axes[0].hist(r1,bins=100)
axes[0].set_xlim([-1,2]) # mengatur batas sumbu x dari -1 hingga 2
axes[0].set_title("Big Variance")
axes[1].hist(r2,bins=100)
axes[1].set_title("Small Variance")
axes[1].set_xlim([-1,2]) # mengatur batas sumbu x dari -1 hingga 2
```

Varians adalah ukuran statistik yang mengukur seberapa jauh data tersebar dari nilai rata-rata.

```python
import random

mean = 0.5
std_dev = 0

random_number = random.normalvariate(mean, std_dev) # (rata-rata, standar deviasi)
print("Random Number from Normal Distribution:", random_number)
```

```python
import random

mean = 0.5
std_dev = 0.2

random_number = random.normalvariate(mean, std_dev) # (rata-rata, standar deviasi)
print("Random Number from Normal Distribution:", random_number)
```

Deviasi standar (standard deviation) adalah ukuran statistik yang digunakan untuk mengukur sejauh mana data tersebar dari nilai rata-ratanya. Deviasi standar memberikan informasi tentang dispersi atau variasi data dalam suatu kumpulan data. Semakin besar deviasi standar, semakin besar variasi data.

```python
np.quantile(rvs,[0.25,0.5,0.75])
```
