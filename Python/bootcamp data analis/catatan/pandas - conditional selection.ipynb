```python
data1 = pd.Series([1,2,3,4,5], index = ['a','b','c','d','e'])
data2 = pd.Series([1,2,3,4,5], index = ['f','g','h','i','j'])

total = data1.add(data2, fill_value = 0)
total
```

```python
data1 = pd.DataFrame([1,2,3,4,5], index = ['a','b','c','d','e'])
data2 = pd.DataFrame([1,2,3,4,5], index = ['a','b','c','d','e'])

total = data1.add(data2, fill_value = 0)
total
```
