```python
data_1 = [1,2,3,4,5,20]

def kuartil(*a):
  data = list(a)
  q1 = np.percentile(data, 25)
  q2 = np.percentile(data, 50)
  q3 = np.percentile(data, 75)

  iqr = q3 - q1

  lower_bound = q1 - 1.5 * iqr
  high_bound = q3 + 1 * iqr

  hasil_bersih = [x for x in data if x > lower_bound and x < high_bound]
  outlier = [x for x in data if x < lower_bound or x > high_bound]

  return hasil_bersih, outlier

x1_bersih, x1_outlier = kuartil(*data_1)

print(x1_outlier)
```
