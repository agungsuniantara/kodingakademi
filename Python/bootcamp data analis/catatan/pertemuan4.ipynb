```python
data_diri = {}

nama = input("masukan nama anda : ")
umur = input("masukan umur anda : ")
alamat = input("masukan alamat anda : ")
username = input("masukan username anda : ")
password = input("masukan password anda : ")

data_diri.update({"nama" : nama, "umur" : umur, "alamat" : alamat, "username" : username, "password" : password})

diri = dict(data_nama = nama, data_umur = umur, data_alamat = alamat, data_username = username, data_password = password)

print(f"nama = {data_diri['nama']}")
print(f"umur = {data_diri['umur']}")
print(f"alamat = {data_diri['alamat']}")
print(f"username = {data_diri['username']}")
print(f"password = {data_diri['password']}")

for x in data_diri:
  print(f"nama = {data_diri[x+1]}")
  print(f"umur = {data_diri[x+2]}")
  print(f"umur = {data_diri[x+3]}")
  print(f"umur = {data_diri[x+4]}")
  print(f"umur = {data_diri[x+5]}")
```

```python
buku_action = {"judul" : "marvel", "rating" : "9", "genre" : "action"}
# buku_romance = {"judul" : "cinta fitri", "rating" : "7", "genre" : "romance"}

while True:
  print(buku_action)
  # print(buku_romance)
  print("pilihlah opsi berikut ini : ")
  print("1. tambah")
  print("2. hapus")
  print("3. selesai")

  masukan = input("masukan pilihan : ")

  if masukan == "tambah":
    jenis = input("masukan jenis buku : ")

    if jenis == "action":
      data_key = input("masukan data key : ")
      isi_data = input("masukan isi data : ")
      buku_action[data_key] = isi_data

    # elif jenis == "romance":
    #   data_baru = input("masukan data : ")
    #   isi_data = input("masukan isi data : ")
    #   buku_romance[data_baru] = isi_data

    else:
      print("jenis buku tidak tersedia")

  elif masukan == "hapus":
    jenis = input("masukan jenis buku : ")
    if jenis == "action":
      data_hapus = input("masukan data key : ")
      del [buku_action[data_hapus]]

    # elif jenis == "romance":
    #   data_hapus = input("masukan data : ")
    #   del [buku_romance[data_hapus]]

    else:
      print("jenis buku tidak tersedia")

  elif masukan == "selesai":
    break

  else:
    print("pilihan tidak tersedia")
```

```python
# fungsi keys()
for key in nama_course.keys():
  print(f"key = {key}")

print("\n")

# fungsi values()
for values in nama_course.values():
  print(f"value = {values}")

print("\n")

# fungsi items()
for keys, values in nama_course.items():
  print(f"key = {keys}, value = {values}")
```
