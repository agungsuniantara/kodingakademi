```python
data = pd.read_csv("Cause-of-Death-Indonesia (1).csv")
data
```

```python
manusia = data[['Total Deaths', 'Year']]

total = pd.DataFrame(manusia.groupby('Year')['Total Deaths'].sum())

total
```

```python
total.idxmax()
```

```python
bencana = data[data['Type'] == 'Bencana Sosial']

bencana
```

```python
filter = bencana[['Total Deaths', 'Cause']]

filter_gabung = filter.groupby('Cause')['Total Deaths'].sum()

filter_gabung
```

```python
sns.barplot(x='Total Deaths', y='Cause', data = filter)
```

```python
filter_gabung.idxmax()
```
