FOR

```python
print("Perulangan manual : ")
print("hello")
print("hello")
print("hello")
print("hello")

# range (start, end, step)

print("Perulangan for 1 : ")
for i in range(4): # i = sebagai index for
    print("hello")

print("Perulangan for 2 : ")
for i in range(1,4):
    print(i)

print("Perulangan for 3 : ")
for i in range(1,10,2):
    print(i)

print("Perulangan list : ")
data = ["apel", "jeruk", "mangga"] # list
for i in data:
    print(i)

```

FOR DI DALAM FOR

```python
for i in range(4): # i = 0
    for z in range(4): # z = 0
        print(f"[{i}][{z}]")
```

WHILE

```python
# contoh 1

ulang = 0 

while ulang < 3:
    print(ulang)
    ulang += 1

# contoh 2

while True:
    print("sekali")
    break

# contoh 3

while True:
    cek = input("apakah ingin mengulangi ? ")
    if cek == "tidak":
        break
    elif cek == "iya":
        print("ulang lagi")
    else:
        print("masukan salah !")
```

DO WHILE

```python
ulang = 0 # 3

do {
    print(ulang)
    ulang += 1
}while (ulang < 3)

# output : 

# 0
# 1
# 2
# 3
```

ASSIGNMENT

```python
data1 = ["* * * * *", "* * * * *", "* * * * *", "* * * * *"]
data2 = ["*", "**", "***", "****", "*****"]

for i in data1:
    print(i)
    
for i in data2:
    print(i)
```

```python
print("bintang 1 : ")
for i in range(5):
    for z in range(5):
        print("* ", end = "")
    print("")

print("bintang 2 : ")
for i in range(5): 
    for z in range(5):
        if i >= z:
            print("* ", end = "")
    print("")
```

```python
for i in range(5):
    for z in range(5):
        print("*", end="")
    print("")
```
