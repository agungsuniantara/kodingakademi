VARIABEL TIPE DATA

```python
nama = "dyah 20" # string (tulisan)
angka = 10 # int (bilangan bulat)
angka_float = 10.5 # float (bilangan desimal)
benar = True # boolean

print("Nama =", nama)
print("Angka int =", angka)
print("Angka float =", angka_float)
print("Boolean =", benar)
```

KONVERSI TIPE DATA

```python
angka_int = 100
angka_float = 20.5676767

jadi_float = float(angka_int) # int to float
jadi_int = int(angka_float) # float to int
jadi_str = str(angka_int)
print(jadi_float)
print(jadi_int)
print(jadi_str)
print(type(jadi_float)) # fungsi type = cek tipe data
print(type(jadi_int))
print(type(jadi_str))
```
