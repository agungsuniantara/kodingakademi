DATA SERIES

LIST

```python
import pandas as pd

data = ["satu", "dua", "tiga", "empat"]
series = pd.Series(data)
series
```

DICTIONARY

```python
data_dict = {
    "nama" : "wahyu", 
    "alamat" : "denpasar"
}
series = pd.Series(data_dict)
series
```

```python
series = pd.Series([1,2,3,4], index = ["satu", "dua", "tiga", "empat"])
series
```

INDEXING

```python
data = ["satu", "dua", "tiga", "empat"]
series = pd.Series(data)

print(series[1])
print(series[0])
print(series)
```

```python
print(series[1:3]) # [awal, batas]
print(series[1:])
```

SLICING (LOC & ILOC)

```python
# LOC (mengakses value dengan indeks ditentukan)
# ILOC (mengakses value dengan indeks yang sebenarnya)

data = ["senin", "selasa", "rabu", "kamis"]
series = pd.Series(data, index = ["satu", "dua", "tiga", "empat"])
print(series.loc["satu" : "tiga"])
print(series.iloc[0:3])
```

CONDITIONAL STATEMENT

```python
data = [1500, 1250, 2224, 1302, 4000]
series = pd.Series(data)
print(series)
print(series > 1500)
print(series[series > 1500])
print(series[(series > 1500) & (series < 4000)])
```

```python
kondisi = series[(series > 1500) & (series < 4000)]
print(kondisi)
```

KONVERSI

```python
print(series)
series = series.astype(str)
print(series)
```

CHANGING DATA SERIES

```python
print(series)
```

```python
series[2] = 10000
print(series)
```

DATAFRAME

LIST

```python
data = [
    [10,20],
    [40,50],
    [60,70]
]

df = pd.DataFrame(data, columns=["angka1", "angka2"], index = ["satu", "dua", "tiga"])
df
```

DICTIONARY

```python
data = {
    "populasi" : [10,20,30],
    "negara" : ["amerika", "indonesia", "german"]
}

df = pd.DataFrame(data)
df
```

AKSES DATAFRAME

```python
# kurung siku 1 = output series
# kurung siku 2 = output dataframe

print(df["populasi"])
print(df[["populasi", "negara"]])
```

INDEXING DATAFRAME

```python
df.index = ["satu", "dua", "tiga"]
df
```

```python
print(df.loc["satu":"dua", "populasi":"negara"]) # [baris, kolom]
print(df.iloc[0:2, 0:2]) # [baris, kolom]
```

```python
print(df.columns)
print(df.index)
```
