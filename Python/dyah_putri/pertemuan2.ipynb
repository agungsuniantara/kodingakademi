OPERATOR

ARITMATIKA

```python
angka1 = int(input("masukan angka 1 : "))
angka2 = int(input("masukan angka 2 : "))

tambah = angka1 + angka2
kurang = angka1 - angka2
bagi = angka1 / angka2 
kali = angka1 * angka2 
modulo = angka1 % angka2 
pangkat = angka1 ** angka2 

print(f"Hasil tambah : {tambah}")
print(f"Hasil kurang : {kurang}")
print(f"Hasil bagi : {bagi}")
print(f"Hasil kali : {kali}")
print(f"Hasil modulo : {modulo}")
print(f"Hasil pangkat : {pangkat}")
```

PERBANDINGAN

```python
angka1 = int(input("masukan angka 1 : "))
angka2 = int(input("masukan angka 2 : "))

lebih_besar = angka1 > angka2
lebih_kecil = angka1 < angka2
sama_dengan = angka1 == angka2 
tidak_sama_dengan = angka1 != angka2
lebihbesar_samadengan = angka1 >= angka2
lebihkecil_samadengan = angka1 <= angka2

print(f"angka1 > angka2 : {lebih_besar}")
print(f"angka1 < angka2 : {lebih_kecil}")
print(f"angka1 == angka2 : {sama_dengan}")
print(f"angka1 != angka2 : {tidak_sama_dengan}")
print(f"angka1 >= angka2 : {lebihbesar_samadengan}")
print(f"angka1 <= angka2 : {lebihkecil_samadengan}")
```

OPERATOR LOGIKA

```python
angka1 = int(input("masukan angka 1 : ")) # 10
angka2 = int(input("masukan angka 2 : ")) # 5
angka3 = int(input("masukan angka 3 : ")) # 7

operator_and = (angka1 > angka2) & (angka1 < angka3) & (angka1 == angka2) # true and false and false = false
operator_or = (angka1 > angka2) | (angka1 < angka3) | (angka1 == angka2) # true or false or false = true
operator_not = (angka1 > angka2) & (not(angka1 < angka3)) # true and true = true
print(operator_and)
print(operator_or)
print(operator_not)
```

CONDITIONAL STATEMENT

```python
username = input("masukan username : ")
password = input("masukan password : ")

if (username == "diah") & (password == "admin") :
    print("halo admin")
elif (username == "agung") & (password == "user") :
    print("halo user")
elif (username == "wahyu") & (password == "staff") :
    print("halo staff")
else :
    print("bukan user / admin / staff")
```
