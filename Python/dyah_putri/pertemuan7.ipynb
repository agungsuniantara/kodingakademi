DATAFRAME CONDITIONAL

```python
import pandas as pd

data = {
    "Athlete": ["Michael Phelps","Michael Phelps","Michael Phelps","Natalie Coughlin","Aleskey Nemov"],
    "Age": [23,19,27,25,24],
    "Country": ["United States","Canada","United States","Norway","China"],
    "Year": [2008,2004,2012,2008,2000],
    "Gold Medals": [8,6,4,1,2]
}

df = pd.DataFrame(data)
df
```

```python
df[(df["Age"] <= 25) & (df["Country"] == "Canada")]
```

```python
df.drop(columns = ["Gold Medals"], inplace=True)
```

```python
df
```

```python
df["Hari"] = ["senin", "selasa", "rabu", "kamis", "jumat"]
df
```

```python
df.rename(
    columns={
        "Country" : "Negara",
        "Athlete" : "Atlit"
    }, inplace=True
)

df
```

```python
df["Negara"].replace({"United States" : "America", "China" : "Indo"}, inplace=True)
df
```
