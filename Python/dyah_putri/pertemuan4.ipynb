ASSIGNMENT

```python
n = int(input("masukan n : "))

angka = 0
end = 0
jumlah = 0

while end < n:
    if angka % 2 == 0:
        jumlah += angka
    angka += 1
    end += 1

print(f"Hasil = {jumlah}")
```

```python
n = int(input("Masukkan angka n:"))
jumlah = 0 

for i in range(n): 
    if i % 2 == 0:
        jumlah += i 
    
print(n)
print("Jumlah angka genap sampai n: ", jumlah)
```

```python
jumlah = 0
z = 0 

for i in range(5):
    angka = int(input("masukan angka : ")) 
    jumlah += angka
    z += 1

print(f"rata - rata = {jumlah / z}")
```

LIST

```python
data = [1,2,3,4,"test",True] 
print(data)

angka1 = data[1] # 2
angka2 = data[2] # 3

data[5] = 100
print(data)

print(f"hasil = {angka1 + angka2}")
```

```python
# append

data = [1,2,3,4,"test",True] 
print(data)

data.append("coba")
print(data)
```

```python
# insert

data = [1,2,3,4,"test",True] 
print(data)

data.insert(1, 90)
print(data)
```

```python
# del

data = [1,2,3,4,"test",True] 
print(data)

del data[2]
print(data)
```

```python
# remove

data = [1,2,3,4,"test",True] 
print(data)

data.remove("test")
print(data)
```

LIST MULTIDIMENSI

```python
data_1d = [1,2,3,4]
data_2d = [
    [1,2,3,4], # 0
    [5,6,7,8], # 1
    [9,10,11,12] # 2
]

print(data_1d[0])
print(data_2d[0])
print(data_2d[0][3]) # [baris][kolom]
```

```python
for baris in data_2d:
    for kolom in baris:
        print(f"{kolom} ", end = "")
    print("")
```

DICTIONARY

```python
data = {
    "nama" : "agung",
    "umur" : 20,
    "alamat" : "denpasar"
} # cara 1

data2 = dict(nama = "agus", umur = 40) # cara 2

print(data)
print(data["nama"])
print(data2)
```

```python
for i in data:
    print(f"key = {i}, value = {data[i]}")
```

```python
data = {
    "nama" : "agung",
    "umur" : 20,
    "alamat" : "denpasar"
} 
print(data)

data["alamat"] = "canggu"
print(data)
```

```python
# pop

data = {
    "nama" : "agung",
    "umur" : 20,
    "alamat" : "denpasar"
} 
print(data)
hapus = data.pop("alamat")
print(hapus)
print(data)
```

```python
# del

data = {
    "nama" : "agung",
    "umur" : 20,
    "alamat" : "denpasar"
} 
print(data)
del data["alamat"]
print(data)
```

```python
# update

data = {
    "nama" : "agung",
    "umur" : 20,
    "alamat" : "denpasar"
} 
print(data)
data.update({"phone" : "081212121", "cek" : "satu"})
print(data)
```
