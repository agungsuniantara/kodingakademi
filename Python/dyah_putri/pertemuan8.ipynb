```python
import pandas as pd 

df = pd.read_csv("data/penyakit_jantung.csv")
df.head()
```

```python
df = df[["sex", "age", "cp", "trestbps", "chol", "thalach"]]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

SEABORN NUMERIC

HISTPLOT

```python
plt.figure(figsize=(10,5)) # (lebar, tinggi)
sns.histplot(x = "trestbps", data = df, bins = 10)
plt.title("Histplot trestbps")
plt.ylabel("Jumlah")
plt.show()
```

KDEPLOT

```python
plt.figure(figsize=(10,5)) # (lebar, tinggi)
sns.kdeplot(x = "trestbps", data = df, fill=True)
plt.title("KDEPLOT trestbps")
plt.show()
```

SCATTERPLOT

```python
plt.figure(figsize=(10,5)) # (lebar, tinggi)
sns.scatterplot(x = "trestbps", y = "chol", data = df)
plt.title("Scatterplot trestbps")
plt.show()
```

PAIRPLOT

```python
sns.pairplot(data = df[["thalach", "chol", "trestbps"]])
plt.show()
```

SEABORN CATEGORY

BARPLOT

```python
sns.barplot(x = "sex", y = "chol", data = df)
plt.title("Barplot")
plt.show()
```

COUNTPLOT

```python
sns.countplot(x = "sex", data = df)
plt.title("Count plot")
plt.show()
```

BOXPLOT

```python
sns.boxplot(x = "sex", y = "trestbps", data = df)
plt.title("Box plot")
plt.show()
```

HEATMAP

```python
pivot = df.pivot_table(index = "sex", columns="age", values="chol")
pivot = pivot.fillna(0)
pivot
```

```python
sns.heatmap(pivot)
plt.title("Heatmap plot")
plt.show()
```

HUE SEABORN

```python
sns.barplot(x = "sex", y = "chol", hue = "cp", data = df, errorbar=None)
plt.title("Barplot hue")
plt.show()
```

SUBPLOT SEABORN

```python
fig, axes = plt.subplots(1, 2) # (baris, kolom)

# kolom 1
sns.histplot(x = "trestbps", data = df, ax = axes[0])
axes[0].set_xlabel("Sumbu x")
axes[0].set_title("Histplot")

# kolom 2
sns.kdeplot(x = "trestbps", data = df, ax = axes[1], fill=True)
axes[1].set_xlabel("Sumbu x")
axes[1].set_title("KDE plot")

plt.tight_layout()
plt.show()
```
