```python
data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

while True : 
    print(data)
    kondisi = input("ingin tambah / hapus : ")
    if kondisi == "tambah":
        jenis = input("masukan jenis buku : ")
        buku = input("masukan buku : ")
        data.update({jenis : buku})
    elif kondisi == "hapus":
        jenis = input("masukan jenis buku : ")
        data.pop(jenis)
    
    stop = input("apakah masih ingin memilih lagi ? ")
    if stop == "tidak":
        break

print(data)
```

DICTIONARY FUNCTION

COPY

```python
data = {
    "action" : "marvel",
    "romance" : "cinta fitri"
}

backup = data.copy()
print(backup)
```

CLEAR

```python
backup.clear()
print(backup)
```

FROMKEY

```python
nama = {"sapi", "ayam", "kambing"}
hewan = dict.fromkeys(nama, "hewan")
print(hewan)
```

LEN

```python
print(data)
print(f"banyak data = {len(data)}")
```

GET

```python
print(data)
action = data.get("action")
print(f"action = {action}")
```

ITEMS

```python
print(data)
data_items = data.items()
print(data_items)
```

```python
for a,b in data.items():
    print(f"key = {a}, value = {b}")
```

KEYS

```python
print(data)
key = data.keys()
print(key)
```

```python
for i in data.keys():
    print(i)
```

VALUES

```python
print(data)
value = data.values()
print(value)
```

```python
for i in data.values():
    print(i)
```

FUNCTION PYTHON

VOID

```python
# function void adalah function yang menyelesaikan prosesnya sendiri

def hitung():
    a = 10
    b = 20
    print(f"hasil = {a + b}")

hitung()
```

```python
# function parameter adalah function yang dapat menerima value di luar dari functionnya sendiri

def tambah(a, b):
    hasil = a + b
    print(f"hasil = {hasil}")

tambah(10,20)
tambah(100,200)
```

```python
# function return adalah function yang dapat mengembalikan sebuah nilai

def kali(a, b):
    hasil = a * b 
    return hasil

result = kali(10,20)
print(result)
```

```python
def kali(a, b):
    hasil = a * b
    return hasil 

def kubus(a, b, c):
    volume = kali(a, b) * c 
    print(f"volume kubus = {volume}")

kubus(10,20,30)
```
