```python
# pandas = library to work with dataframes
# pandas have 2 main data structures: Series and DataFrame
# series = 1-dimensional labeled array (row)
# DataFrame = 2-dimensional labeled data structure (row & column)

import pandas as pd

data_series = [420, 380, 390]
df_series = pd.Series(data_series)
print(df_series)
print(df_series[0])

data_dataframe = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45],
    "name" : ["John", "Anna", "Peter"]
}

df_dataframe= pd.DataFrame(data_dataframe)
print(df_dataframe)
```

```python
print(df_dataframe["calories"])  # access column by column name
```

```python
# loc & iloc = access a group of rows and columns by label or integer position
# loc = label based
# iloc = integer position based

print(df_dataframe.loc[0:1, "calories" : "duration"])  # loc[row_start : row_end, column_start : column_end]
print(df_dataframe.iloc[0:, 0:2])  # iloc[row_start : row_end, column_start : column_end]
```

```python
# conditional dataframe 

print(df_dataframe.loc[df_dataframe["calories"] > 390])  
print(df_dataframe.loc[df_dataframe["calories"] > 390, ["name", "duration"]])  
print(df_dataframe.loc[ (df_dataframe["calories"] < 400) & (df_dataframe["duration"] > 40) ])
```

```python
result = df_dataframe.loc[df_dataframe["calories"] > 390]
print(result)
```

```python
# modifying dataframe 

data_dataframe = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45],
    "name" : ["John", "Anna", "Peter"]
}

df_dataframe= pd.DataFrame(data_dataframe)
print(df_dataframe)

df_dataframe.at[1, "calories"] = 400  # modify single value
print(df_dataframe)

df_dataframe.loc[df_dataframe["name"] == "Peter", "duration"] = 60  # modify value based on condition
print(df_dataframe)

df_dataframe["calories"] = df_dataframe["calories"] + 50  # modify entire column
print(df_dataframe)

df_dataframe["charge"] = [10, 8, 9]  # add new column
print(df_dataframe)

df_dataframe["total"] = df_dataframe["calories"] + df_dataframe["charge"]  # add new column based on existing columns
print(df_dataframe)
```

```python
print(df_dataframe["calories"].sum())
print(df_dataframe["duration"].max())
print(df_dataframe["charge"].min())
print(df_dataframe["total"].mean())
```

```python
# delete dataframe

print(df_dataframe)
df_dataframe.drop("total", axis=1, inplace=True)  # delete column (axis=1 for column, axis=0 for row) inplace=True to modify original dataframe
print(df_dataframe)
df_dataframe.drop([0,1], axis=0, inplace=True)  # delete rows
print(df_dataframe)
```

```python
df_dataframe.rename(columns={"charge": "fee", "name": "customer"}, inplace=True)  # rename column
print(df_dataframe)
```

```python
data_dataframe = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45],
    "name" : ["John", "Anna", "Peter"]
}

df_dataframe= pd.DataFrame(data_dataframe)

df_dataframe.replace({390: 395, "John": "Jane"}, inplace=True)  # replace value
print(df_dataframe)
```

```python
data_dataframe = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45],
    "name" : ["John", "Anna", "Peter"]
}

df_dataframe= pd.DataFrame(data_dataframe)

mean_calories = df_dataframe["calories"].mean()

df_dataframe.loc[df_dataframe["calories"] > mean_calories, "calories"] = mean_calories.astype(int)  # set calories above mean to mean value
print(df_dataframe)
```

```python
row_index = df_dataframe.loc[df_dataframe["calories"] > 390].index
df_dataframe.drop(row_index, axis=0, inplace=True)
```
