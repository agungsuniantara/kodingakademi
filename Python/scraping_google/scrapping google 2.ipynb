```python
import requests
from bs4 import BeautifulSoup
import urllib.parse
from newspaper import Article
import nltk

# Ensure punkt tokenizer is downloaded
nltk.download('punkt')

def google_search_bs(query, num_results=10, num_pages=3):
    search_results = []
    base_url = "https://www.google.com/search"

    for page in range(num_pages):
        start = page * 10  # 10 results per page
        params = {
            'q': query,
            'start': start
        }
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.121 Safari/537.36"
        }

        response = requests.get(base_url, headers=headers, params=params)
        soup = BeautifulSoup(response.text, 'html.parser')

        for g in soup.find_all('div', class_='g'):
            anchors = g.find_all('a')
            if anchors:
                link = anchors[0]['href']
                title = g.find('h3').text if g.find('h3') else 'No title'
                snippet = g.find('span', class_='aCOpRe').text if g.find('span', class_='aCOpRe') else 'No snippet'
                search_results.append({
                    'title': title,
                    'link': link,
                    'snippet': snippet
                })
    return search_results

def fetch_and_summarize(url):
    try:
        article = Article(url)
        article.download()
        article.parse()
        article.nlp()
        return article.summary
    except Exception as e:
        return None

def search(query):
    results = google_search_bs(query)

    # Displaying search results and summaries
    no = 0
    for result in results:
        summary = fetch_and_summarize(result['link'])
        if summary:
            no += 1
            print(f"{no}. Title: {result['title']}")
            print(f"   URL: {result['link']}")
            print(f"   Snippet: {result['snippet']}")
            print(f"   Summary: {summary}\n")

```

```python
# Example usage

search("Gempa kalimantan agustus")
```
