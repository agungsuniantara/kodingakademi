Variables

```python
# Declaring variables
name = "John"
age = 30
height = 175.5

# Displaying variable values
print("Name:", name)
print("Age:", age)
print("Height:", height)
```

Data Types

```python
# Basic data types

number = 10               # int
pi = 3.14                 # float
text = "Learning Python"  # str
is_true = True            # bool

print(type(number))       # Output: <class 'int'>
print(type(pi))           # Output: <class 'float'>
print(type(text))         # Output: <class 'str'>
print(type(is_true))       # Output: <class 'int'>
```
